# EX — Backpropagation From Scratch

**Goal:** manually derive and implement gradient descent for a tiny 1-layer network,
then verify against PyTorch autograd.


````python
import numpy as np

# y = sigmoid(w*x + b), loss = (y - target)^2
def sigmoid(z): return 1/(1+np.exp(-z))

x, target = 2.0, 1.0
w, b = 0.5, 0.0
lr = 0.1

def forward(w, b):
    z = w*x + b
    y = sigmoid(z)
    loss = (y - target)**2
    return y, loss

# Manual gradients (chain rule):
# dloss/dy = 2*(y-target)
# dy/dz = y*(1-y)          [sigmoid derivative]
# dz/dw = x, dz/db = 1
for step in range(10):
    y, loss = forward(w, b)
    dloss_dy = 2*(y - target)
    dy_dz = y*(1-y)
    dz_dw = x
    dz_db = 1
    grad_w = dloss_dy * dy_dz * dz_dw
    grad_b = dloss_dy * dy_dz * dz_db
    w -= lr * grad_w
    b -= lr * grad_b
    if step % 2 == 0:
        print(f"step {step}: loss={loss:.4f} w={w:.4f} b={b:.4f}")
````

### TODO
Verify these manual gradients match PyTorch autograd for the *first* step (starting from `w=0.5, b=0.0`).

````python
import torch

w_t = torch.tensor(0.5, requires_grad=True)
b_t = torch.tensor(0.0, requires_grad=True)
x_t = torch.tensor(2.0)
target_t = torch.tensor(1.0)

# TODO: compute y_t, loss_t, call backward(), print w_t.grad and b_t.grad
# Compare to grad_w / grad_b computed manually above for the FIRST step (w=0.5,b=0.0)
````

```python
y_t = torch.sigmoid(w_t * x_t + b_t)
loss_t = (y_t - target_t) ** 2
loss_t.backward()
print(w_t.grad.item(), b_t.grad.item())
# Should closely match the first manually computed grad_w, grad_b
```

## Key Takeaways
- Backprop = chain rule, applied layer by layer, from the loss backward to each weight.
- PyTorch's `.backward()` automates exactly this — useful to have done it by hand once.
- If gradients don't match your manual math, double check the derivative of your activation function.
