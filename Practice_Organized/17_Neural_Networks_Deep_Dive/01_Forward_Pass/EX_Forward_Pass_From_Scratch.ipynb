# EX — Forward Pass From Scratch (NumPy) and in PyTorch

**Goal:** implement a tiny 2-layer neural network's forward pass by hand with NumPy,
then confirm it matches PyTorch's `nn.Linear` + activation output.


````python
import numpy as np

# A single "layer": weights, bias, activation
def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

np.random.seed(0)
X = np.random.rand(4, 3)          # 4 samples, 3 features
W1 = np.random.randn(3, 5) * 0.1  # 3 -> 5
b1 = np.zeros(5)
W2 = np.random.randn(5, 1) * 0.1  # 5 -> 1
b2 = np.zeros(1)

def forward(X):
    z1 = X @ W1 + b1
    a1 = relu(z1)
    z2 = a1 @ W2 + b2
    a2 = sigmoid(z2)   # binary classification style output
    return a2

output = forward(X)
print(output)
````

### Cross-check with PyTorch
Build the same 3-3-...-1 shaped network with `nn.Sequential` and confirm shapes match (weights differ since they're randomly initialized separately — the point is shape/flow, not identical numbers).

````python
import torch
import torch.nn as nn

torch_model = nn.Sequential(
    nn.Linear(3, 5), nn.ReLU(),
    nn.Linear(5, 4), nn.ReLU(),
    nn.Linear(4, 1), nn.Sigmoid(),
)
X_t = torch.tensor(X, dtype=torch.float32)
print(torch_model(X_t).shape)  # expect torch.Size([4, 1])
````
