## 1. Big-O Warm-up
**Pointer:** before optimizing, always get a correct brute-force solution first.

````python
# Brute-force: find if any two numbers in a list sum to a target (O(n^2))
def has_pair_sum_bruteforce(nums, target):
    n = len(nums)
    for i in range(n):
        for j in range(i+1, n):
            if nums[i] + nums[j] == target:
                return True
    return False

print(has_pair_sum_bruteforce([2,7,11,15], 9))
````

## 3. Sorting & Searching
**Pointer:** Python's built-in sort is already O(n log n) — you rarely hand-implement sorting, but binary search on sorted data is a pattern worth knowing cold.

````python
import bisect

sorted_prices = [10, 25, 40, 55, 70, 90]

def cheapest_at_least(prices, budget_min):
    idx = bisect.bisect_left(prices, budget_min)
    return prices[idx] if idx < len(prices) else None

print(cheapest_at_least(sorted_prices, 42))  # expect 55
````

## 4. Graph Traversal — BFS for Shortest Path
Real-world use: shortest number of hops between users in a social graph, or steps in a workflow/dependency graph.

````python
from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C", "E"],
    "E": ["D"],
}

def bfs_shortest_path(graph, start, goal):
    queue = deque([[start]])
    visited = {start}
    while queue:
        path = queue.popleft()
        node = path[-1]
        if node == goal:
            return path
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(path + [neighbor])
    return None

print(bfs_shortest_path(graph, "A", "E"))
````

## 5. Dynamic Programming — Recognizing Overlapping Subproblems
Real-world use: computing things like 'minimum cost to reach state N' where naive recursion recomputes the same subproblem repeatedly.

````python
# Naive recursive Fibonacci -- exponential time, recomputes the same values many times
def fib_naive(n):
    if n <= 1: return n
    return fib_naive(n-1) + fib_naive(n-2)

print(fib_naive(20))
````

    if cache is None:
        cache = {}
    if n <= 1:
        return n

## Key Takeaways
- A hash map/set is the go-to tool for turning O(n^2) 'have I seen this' problems into O(n).
- Binary search (`bisect`) is O(log n) on sorted data — much faster than scanning.
- BFS finds shortest paths in unweighted graphs; DFS explores depth-first (good for connectivity/cycles).
- Memoization turns exponential naive recursion into efficient top-down dynamic programming.
