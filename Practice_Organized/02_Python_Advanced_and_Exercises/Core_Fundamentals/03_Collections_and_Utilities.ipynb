## 🧠 Concept Notes

### When to use which collection?

| Type | Ordered | Mutable | Duplicates | Key-Value | Use Case |
|------|---------|---------|------------|-----------|----------|
| `list` | ✅ | ✅ | ✅ | ❌ | Sequence of items |
| `tuple` | ✅ | ❌ | ✅ | ❌ | Immutable record |
| `dict` | ✅ (3.7+) | ✅ | keys: ❌ | ✅ | Lookups by key |
| `set` | ❌ | ✅ | ❌ | ❌ | Membership / dedup |
| `frozenset` | ❌ | ❌ | ❌ | ❌ | Hashable set |

### Big-O Complexity

| Operation | list | dict | set |
|-----------|------|------|-----|
| Access by index | O(1) | O(1) | — |
| Search | O(n) | O(1) | O(1) |
| Insert/Delete end | O(1) | O(1) | O(1) |
| Insert/Delete start | O(n) | — | — |

### collections Module Highlights
- **Counter** — frequency counting
- **defaultdict** — dict with automatic default values
- **deque** — O(1) append/pop from both ends
- **namedtuple** — tuple with named fields
- **OrderedDict** — preserves insertion order (less needed in 3.7+)
- **ChainMap** — view over multiple dicts

## ✅ Executable Code

````python
# ── 1. Lists — methods & operations ──
lst = [5, 3, 1, 4, 2]
lst.append(6)
lst.insert(0, 0)
lst.extend([7, 8])
print("After append/insert/extend:", lst)

lst.remove(0)           # remove first occurrence of value
popped = lst.pop(-1)    # remove & return last
print(f"After remove/pop({popped}):", lst)

lst.sort()
print("Sorted:", lst)
print("Reversed:", lst[::-1])

# List as stack (LIFO)
stack = []
for i in range(5): stack.append(i)
print("Stack:", stack)
while stack: print(stack.pop(), end=" ")
print()
````

````python
# ── 2. Tuples & Unpacking ──
point = (3, 4)
x, y = point
print(f"Point: ({x}, {y})")

# Extended unpacking
first, *middle, last = [1, 2, 3, 4, 5]
print(f"first={first}, middle={middle}, last={last}")

# Swapping (Pythonic, uses tuple under the hood)
a, b = 10, 20
a, b = b, a
print(f"After swap: a={a}, b={b}")

# Tuples are hashable → usable as dict keys
distance = {(0,0): 0, (1,0): 1, (0,1): 1, (1,1): 1.414}
print("Distance (1,1):", distance[(1,1)])
````

````python
# ── 3. Dictionaries ──
# Creation
person = {"name": "Alice", "age": 30, "city": "NYC"}

# Safe access
print(person.get("email", "N/A"))   # default if missing

# Iterate
for key, val in person.items():
    print(f"  {key}: {val}")

# Dict comprehension
squares = {n: n**2 for n in range(1, 6)}
print("Squares:", squares)

# Merge dicts (Python 3.9+)
defaults = {"timeout": 30, "retries": 3}
overrides = {"retries": 5, "debug": True}
config = defaults | overrides          # merge, overrides win
print("Config:", config)

# Nested dict
users = {
    "alice": {"role": "admin", "active": True},
    "bob":   {"role": "user",  "active": False},
}
active_admins = [u for u, d in users.items() if d["role"] == "admin" and d["active"]]
print("Active admins:", active_admins)
````

````python
# ── 4. Sets ──
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7}

print("Union:        ", a | b)
print("Intersection: ", a & b)
print("Difference:   ", a - b)
print("Sym Diff:     ", a ^ b)
print("Subset:       ", {1,2}.issubset(a))
print("Superset:     ", a.issuperset({1,2}))

# Deduplication
words = ["apple","banana","apple","cherry","banana","date"]
unique = sorted(set(words))
print("Unique words:", unique)
````

````python
# ── 5. collections Module ──
from collections import Counter, defaultdict, deque, namedtuple, ChainMap

# Counter
text = "the quick brown fox jumps over the lazy dog"
freq = Counter(text.split())
print("Top 3 words:", freq.most_common(3))

# defaultdict
graph = defaultdict(list)
edges = [(1,2),(1,3),(2,4),(3,4)]
for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)
print("Graph adjacency:", dict(graph))

# deque — efficient queue
queue = deque(maxlen=5)
for i in range(8):
    queue.append(i)    # auto-evicts oldest when full
print("Deque (maxlen=5):", queue)
queue.appendleft(99)
print("After appendleft:", queue)

# namedtuple
Point3D = namedtuple("Point3D", ["x", "y", "z"])
p = Point3D(1, 2, 3)
print(f"Point: {p}, x={p.x}, z={p.z}")
print(p._asdict())

# ChainMap
base = {"a": 1, "b": 2}
override = {"b": 99, "c": 3}
cm = ChainMap(override, base)
print("ChainMap b:", cm["b"])   # 99 — first found wins
````

````python
# ── 6. itertools ──
from itertools import (chain, product, groupby, combinations,
                        permutations, cycle, islice, accumulate)
import operator

# chain — flatten iterables
flat = list(chain([1,2], [3,4], [5]))
print("Chain:", flat)

# product — cartesian product
suits = "♠♥♦♣"
ranks = ["A","K","Q","J"]
deck_sample = list(islice(product(ranks, suits), 8))
print("Deck sample:", deck_sample)

# groupby
data = [("fruit","apple"),("fruit","banana"),("veg","carrot"),("veg","daikon")]
for category, items in groupby(data, key=lambda x: x[0]):
    print(f"  {category}: {[i[1] for i in items]}")

# accumulate — running total
nums = [1, 2, 3, 4, 5]
running = list(accumulate(nums))
running_product = list(accumulate(nums, operator.mul))
print("Running sum:", running)
print("Running product:", running_product)

# combinations & permutations
print("Combinations(4,2):", list(combinations("ABCD", 2)))
print("Permutations(ABC,2):", list(permutations("ABC", 2)))
````

## 📝 Exercises

1. **Word Frequency** — Given a paragraph, return the 5 most common words (ignore case & punctuation).
2. **Anagram Grouper** — Group a list of words into anagram clusters.
3. **Sliding Window Max** — Given a list and window size k, return the max in each window.
4. **Inventory Manager** — Use a defaultdict to build a category → [items] inventory from a flat list.
5. **Unique Pairs** — Given a list of numbers and target sum, find all unique pairs that sum to target.


## ✔️ Solutions

````python
import re
from collections import Counter, defaultdict
from collections import deque

# 1. Word Frequency
def top_words(text, n=5):
    words = re.findall(r"[a-z]+", text.lower())
    return Counter(words).most_common(n)

sample = "To be or not to be that is the question To be is to exist"
print(top_words(sample))

# 2. Anagram Grouper
def group_anagrams(words):
    groups = defaultdict(list)
    for w in words:
        groups[tuple(sorted(w))].append(w)
    return list(groups.values())

print(group_anagrams(["eat","tea","tan","ate","nat","bat"]))

# 3. Sliding Window Max (deque approach O(n))
def sliding_max(nums, k):
    dq = deque()  # stores indices
    result = []
    for i, n in enumerate(nums):
        while dq and dq[0] < i - k + 1:
            dq.popleft()
        while dq and nums[dq[-1]] < n:
            dq.pop()
        dq.append(i)
        if i >= k - 1:
            result.append(nums[dq[0]])
    return result

print(sliding_max([1,3,-1,-3,5,3,6,7], 3))  # [3,3,5,5,6,7]

# 4. Inventory Manager
raw = [("Electronics","Laptop"),("Clothing","Shirt"),
       ("Electronics","Phone"),("Clothing","Jacket"),("Food","Apple")]
inventory = defaultdict(list)
for cat, item in raw:
    inventory[cat].append(item)
print(dict(inventory))

# 5. Unique Pairs
def unique_pairs(nums, target):
    seen, result = set(), set()
    for n in nums:
        complement = target - n
        if complement in seen:
            result.add((min(n, complement), max(n, complement)))
        seen.add(n)
    return list(result)

print(unique_pairs([1,3,2,2,4,5,3,1], 5))
````

## 💼 Enterprise Examples

````python
# Enterprise: Event analytics pipeline using collections
from collections import Counter, defaultdict
from itertools import groupby
from datetime import datetime

# Simulated event log
events = [
    {"user": "alice", "action": "login",   "page": "/dashboard", "ts": "2024-01-15 09:00"},
    {"user": "bob",   "action": "view",    "page": "/products",  "ts": "2024-01-15 09:05"},
    {"user": "alice", "action": "view",    "page": "/products",  "ts": "2024-01-15 09:10"},
    {"user": "alice", "action": "purchase","page": "/checkout",  "ts": "2024-01-15 09:15"},
    {"user": "bob",   "action": "logout",  "page": "/",          "ts": "2024-01-15 09:20"},
    {"user": "carol", "action": "login",   "page": "/dashboard", "ts": "2024-01-15 09:25"},
    {"user": "carol", "action": "view",    "page": "/products",  "ts": "2024-01-15 09:30"},
    {"user": "carol", "action": "purchase","page": "/checkout",  "ts": "2024-01-15 09:35"},
]

# Metrics
action_counts = Counter(e["action"] for e in events)
page_visits   = Counter(e["page"]   for e in events)
user_actions  = defaultdict(list)
for e in events:
    user_actions[e["user"]].append(e["action"])

print("Action breakdown:", dict(action_counts))
print("Page visits:", dict(page_visits))
print("Conversion rate:", f"{action_counts['purchase']/action_counts['login']*100:.0f}%")
print()
for user, actions in user_actions.items():
    print(f"  {user}: {' → '.join(actions)}")
````

## ❓ Interview Questions

**Q1: When would you use a `deque` over a `list`?**
> When you need O(1) append/pop from **both** ends. `list.pop(0)` is O(n); `deque.popleft()` is O(1).

**Q2: Are dicts ordered in Python 3?**
> Yes — since **Python 3.7**, dicts maintain insertion order as part of the language spec.

**Q3: What makes a type hashable?**
> It must implement `__hash__` and `__eq__`. Mutable containers (list, dict, set) are not hashable. Tuples **are** hashable if all their elements are.

**Q4: `defaultdict` vs `dict.setdefault`?**
> `defaultdict` auto-creates missing keys on access. `setdefault` creates only on explicit call. `defaultdict` is cleaner for repeated use.

**Q5: How does `Counter` arithmetic work?**
> `Counter` supports `+`, `-`, `&` (intersection/min), `|` (union/max). Subtraction drops ≤0 counts.


## ⚠️ Common Mistakes

| Mistake | Problem | Fix |
|---------|---------|-----|
| Modifying list while iterating | Skips elements | Iterate over a copy: `for x in lst[:]` |
| Using list for membership test | O(n) | Use `set` for O(1) |
| dict key ordering assumption | Only guaranteed in 3.7+ | Add `from __future__` note for older code |
| `defaultdict` leaks in APIs | Creates key on `d[missing]` | Use `.get()` for optional access |
| Copying with `=` | Same reference | Use `lst.copy()` or `lst[:]` |
| `Counter` with floats | Unexpected results | Use integer counts |
