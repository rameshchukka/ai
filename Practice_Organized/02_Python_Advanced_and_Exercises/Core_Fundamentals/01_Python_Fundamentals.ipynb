## 🧠 Concept Notes

### Variables & Dynamic Typing
Python is **dynamically typed** — the interpreter infers the type at runtime.
```
x = 10        # int
x = "hello"   # now str — perfectly valid
```

### Key Built-in Types
| Type | Example | Notes |
|------|---------|-------|
| `int` | `42` | Unlimited precision |
| `float` | `3.14` | IEEE 754 double |
| `str` | `"hi"` | Immutable, Unicode |
| `bool` | `True/False` | Subclass of int |
| `NoneType` | `None` | Singleton null |

### Truthy / Falsy
`False`, `0`, `0.0`, `""`, `[]`, `{}`, `None` → all falsy. Everything else → truthy.

### String Formatting — 3 styles
```python
name = "Alice"
# 1. f-string (Python 3.6+) ← preferred
f"Hello, {name}!"
# 2. .format()
"Hello, {}!".format(name)
# 3. % operator (legacy)
"Hello, %s!" % name
```

## ✅ Executable Code

````python
# ── 1. Variables & Types ──
x = 42
pi = 3.14159
name = "Python"
is_fun = True
nothing = None

print(type(x), type(pi), type(name), type(is_fun), type(nothing))
print(f"x={x}, pi={pi:.2f}, name={name}, is_fun={is_fun}, nothing={nothing}")
````

````python
# ── 2. Arithmetic & Integer Division ──
a, b = 17, 5
print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
print(f"{a} / {b}  = {a / b}")    # true division → float
print(f"{a} // {b} = {a // b}")   # floor division → int
print(f"{a} % {b}  = {a % b}")    # modulus
print(f"{a} ** {b} = {a ** b}")   # power
````

````python
# ── 3. String operations ──
s = "Hello, World!"
print(s.upper(), s.lower(), s.title())
print(s[0:5])            # slicing
print(s[::-1])           # reverse
print(s.replace("World", "Python"))
print(s.split(", "))
print("  spaces  ".strip())
print(f"Length: {len(s)}")
````

````python
# ── 4. Control Flow ──
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")

# Ternary / conditional expression
status = "Pass" if score >= 60 else "Fail"
print(f"Status: {status}")
````

````python
# ── 5. Loops ──
# for loop with range
total = 0
for i in range(1, 11):
    total += i
print(f"Sum 1-10: {total}")

# while loop
n = 10
factorial = 1
while n > 0:
    factorial *= n
    n -= 1
print(f"10! = {factorial}")

# break / continue
evens = []
for i in range(20):
    if i % 2 != 0:
        continue
    if i > 10:
        break
    evens.append(i)
print(f"Evens up to 10: {evens}")
````

````python
# ── 6. Comprehensions ──
squares   = [x**2 for x in range(1, 6)]
even_sq   = [x**2 for x in range(1, 11) if x % 2 == 0]
word_lens = {w: len(w) for w in ["python", "is", "awesome"]}

print("Squares:", squares)
print("Even squares:", even_sq)
print("Word lengths:", word_lens)
````

## 📝 Exercises

1. **Temperature Converter** — Write a program that converts Celsius to Fahrenheit and Kelvin.
2. **FizzBuzz** — Print 1-100; multiples of 3 → "Fizz", 5 → "Buzz", both → "FizzBuzz".
3. **Palindrome Checker** — Check whether a string reads the same forwards and backwards.
4. **Digit Sum** — Given an integer, return the sum of its digits.
5. **Star Pattern** — Print a right-angled triangle of stars with n rows.


## ✔️ Solutions

````python
# 1. Temperature Converter
def celsius_to_all(c):
    f = c * 9/5 + 32
    k = c + 273.15
    return f, k

c = 100
f, k = celsius_to_all(c)
print(f"{c}°C = {f}°F = {k}K")

# 2. FizzBuzz
for i in range(1, 21):
    if i % 15 == 0:  print("FizzBuzz")
    elif i % 3 == 0: print("Fizz")
    elif i % 5 == 0: print("Buzz")
    else:            print(i)
````

````python
# 3. Palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

for word in ["racecar", "hello", "A man a plan a canal Panama"]:
    print(f"'{word}' → {is_palindrome(word)}")

# 4. Digit Sum
def digit_sum(n):
    return sum(int(d) for d in str(abs(n)))

print(digit_sum(12345))   # 15

# 5. Star Pattern
def star_triangle(n):
    for i in range(1, n+1):
        print("*" * i)

star_triangle(5)
````

## 💼 Enterprise Examples

### Config Parsing with Type Coercion


````python
# Enterprise: safe config loader with type coercion
import os

DEFAULTS = {"MAX_RETRIES": 3, "TIMEOUT": 30.0, "DEBUG": False, "SERVICE_NAME": "api"}

def get_config(key, cast=str):
    """Read env var with fallback to defaults and type casting."""
    raw = os.environ.get(key, str(DEFAULTS.get(key, "")))
    try:
        if cast is bool:
            return raw.lower() in ("1", "true", "yes")
        return cast(raw)
    except (ValueError, TypeError):
        return DEFAULTS.get(key)

# Simulate reading configuration
config = {
    "max_retries": get_config("MAX_RETRIES", int),
    "timeout":     get_config("TIMEOUT", float),
    "debug":       get_config("DEBUG", bool),
    "service":     get_config("SERVICE_NAME", str),
}
print("App config:", config)
````

## ❓ Interview Questions

**Q1: What is the difference between `==` and `is`?**
> `==` compares **value**; `is` compares **identity** (same object in memory).  
> `[] == []` → True, but `[] is []` → False.

**Q2: Are strings mutable in Python?**
> No. Strings are **immutable** — every "modification" creates a new object.

**Q3: What does `//` do?**
> **Floor division** — divides and rounds *down* to the nearest integer.  
> `-7 // 2 = -4` (not -3!).

**Q4: Difference between `break`, `continue`, `pass`?**
> `break` → exits loop; `continue` → skips to next iteration; `pass` → no-op placeholder.

**Q5: What is the GIL?**
> The **Global Interpreter Lock** allows only one thread to execute Python bytecode at a time, limiting true parallelism for CPU-bound tasks (use `multiprocessing` instead).


## ⚠️ Common Mistakes

| Mistake | Wrong | Right |
|---------|-------|-------|
| Integer division | `7/2 == 3` | `7/2 == 3.5`; use `7//2` for int |
| Mutable default arg | `def f(lst=[]):` | `def f(lst=None): lst = lst or []` |
| `is` vs `==` | `x is 1000` | `x == 1000` |
| Chained assignment confusion | `a = b = []` (same object!) | `a, b = [], []` |
| Off-by-one in range | `range(1, n)` misses n | `range(1, n+1)` |
| String concatenation in loop | `s += x` inside loop O(n²) | `"".join(list)` |
