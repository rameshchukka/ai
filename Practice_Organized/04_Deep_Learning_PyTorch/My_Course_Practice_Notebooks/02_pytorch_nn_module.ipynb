```python
# create model class
import torch
import torch.nn as nn
```

```python


class ModelSequential(nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
                                  nn.Linear(num_features, 3),
                                  nn.ReLU(),
                                  nn.Linear(3,2),
                                  nn.LeakyReLU(),
                                  nn.Linear(2, 1),
                                  nn.Sigmoid()
                                )

  def forward(self, features):
    out = self.network(features)
    return out
```

```python
features = torch.rand(10,5)
```

```python
modelobj = ModelSequential(num_features=5)

modelobj.forward(features)

modelobj(features)
```

```python
import torch
import torch.nn as nn

# A second way to build the SAME kind of network: explicit named layers
# instead of nn.Sequential. Renamed to ModelManual so it doesn't overwrite
# ModelSequential above -- both stay usable side by side.
class ModelManual(nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear1(features)
    out = self.relu(out)
    out =self.linear2(out) 
    out = self.sigmoid(out)
    return out
```

```python
# create dataset


# create model
model = ModelSequential(5)

# call model for forward pass
# model.forward(features)
model(features)
```

```python
# show model weights
# ModelManual (cell above) defines layers as separate named attributes, so you'd
# access weights via model.linear1.weight directly on that class.
# ModelSequential (used here) wraps layers in nn.Sequential, so we index into
# the container instead: model.network[2] is the second nn.Linear layer.
model.network[2].weight  # the second Linear layer inside the Sequential container
```
