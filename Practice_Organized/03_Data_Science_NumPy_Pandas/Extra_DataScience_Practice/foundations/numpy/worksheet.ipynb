## 1. Creating arrays — the five ways you'll use constantly

```python
# From a Python list
a = np.array([10, 20, 30, 40, 50])
print("from list:", a)
print("shape:", a.shape, "  dtype:", a.dtype, "  ndim:", a.ndim)

# Range-based
b = np.arange(0, 20, 3)      # start, stop, step
print("arange:", b)

# Evenly spaced (useful for plot x-axes)
c = np.linspace(0, 1, 6)     # start, stop, n_points
print("linspace:", c)

# Filled
print("zeros:", np.zeros((2, 3)))
print("ones:", np.ones((2, 3)))

# Random (set seed for reproducibility)
np.random.seed(42)
d = np.random.randn(3, 3)    # standard normal (mean=0, std=1)
print("random:\n", d)
```

## 2. Indexing and slicing — getting at specific values

```python
np.random.seed(42)
m = np.arange(1, 13).reshape(3, 4)  # shape (3,4) — numbers 1 to 12
print("matrix m:\n", m)
print()

# Single element
print("m[0, 0] =", m[0, 0])    # row 0, col 0
print("m[2, 3] =", m[2, 3])    # row 2, col 3  (last element)

# Entire row or column
print("row 1:", m[1])           # or m[1, :]
print("col 2:", m[:, 2])        # ALL rows, col 2

# Slices
print("top-left 2x2:\n", m[:2, :2])    # rows 0-1, cols 0-1
print("bottom row:", m[-1, :])           # -1 = last row

# Boolean indexing — powerful
print("elements > 7:", m[m > 7])
print("even elements:", m[m % 2 == 0])
```

### Observe:
- What does `m[-1, :]` give? Try `m[-2, :]`.
- `m[m > 7]` returns a 1D array. Why? (Where did the 2D shape go?)
- Try `m[m > 7] = 0` — does it modify `m` in place? Print `m` to confirm.

## 3. Vectorized operations — NO loops needed

```python
a = np.array([1, 2, 3, 4, 5], dtype=float)
b = np.array([10, 20, 30, 40, 50], dtype=float)

print("a + b =", a + b)
print("a * b =", a * b)
print("a ** 2 =", a ** 2)
print("np.sqrt(a) =", np.sqrt(a).round(3))
print("np.log(a) =", np.log(a).round(3))

# Comparison — produces bool array
print("a > 3:", a > 3)
print("a == 3:", a == 3)

# Combine with & (and), | (or)
print("2 < a < 5:", (a > 2) & (a < 5))

# Time comparison: loop vs vectorized (try this live)
import time
big = np.random.randn(1_000_000)

t0 = time.time()
result_loop = [x**2 for x in big]
print(f"Loop:       {time.time()-t0:.4f}s")

t0 = time.time()
result_vec = big ** 2
print(f"Vectorized: {time.time()-t0:.4f}s")
```

### Observe:
- How much faster is vectorized vs loop on 1 million elements?
- What happens if you try `a + b` when `a` has shape `(5,)` and `b` has shape `(3,)`? Try it.

## 4. Aggregations — summarizing data

```python
np.random.seed(42)
data = np.random.randint(1, 100, size=(4, 5))
print("data:\n", data)
print()

# Whole array
print("sum:", data.sum())
print("mean:", data.mean())
print("std:", data.std().round(2))
print("min:", data.min(), "  max:", data.max())
print("argmin:", data.argmin(), "  (flattened index)")

# Per row or column — axis parameter (see diagrams.md section 2)
print("\ncolumn means (axis=0):", data.mean(axis=0).round(2))
print("row means    (axis=1):", data.mean(axis=1).round(2))
print("column sums  (axis=0):", data.sum(axis=0))
print("row sums     (axis=1):", data.sum(axis=1))
```

### Observe:
- `data.mean(axis=0)` gives 5 numbers (one per column) or 4 (one per row)? Confirm against the diagram.
- `data.argmin()` gives a single number even though `data` is 2D. What does it mean?

## 5. Reshaping — same data, different shape

```python
a = np.arange(1, 13)   # [1, 2, 3, ..., 12]
print("original:", a.shape, a)

b = a.reshape(3, 4)    # 3 rows, 4 cols
print("reshaped (3,4):\n", b)

c = a.reshape(2, 6)    # 2 rows, 6 cols
print("reshaped (2,6):\n", c)

# -1 means "infer this dimension"
d = a.reshape(-1, 4)   # "however many rows needed, 4 cols"
print("reshape(-1, 4):", d.shape, "\n", d)

# Transpose
print("original shape:", b.shape)
print("transposed shape:", b.T.shape)
print("transposed:\n", b.T)

# Flatten
print("flattened:", b.flatten())
```

## 6. Broadcasting — operations on different-shaped arrays
Read `diagrams.md` section 3 before this cell.

```python
# Scalar broadcast (simplest case)
a = np.array([1, 2, 3, 4, 5])
print("a + 10:", a + 10)    # 10 "broadcasts" to every element

# 1D broadcast across rows of a 2D array
matrix = np.arange(1, 13).reshape(3, 4)
row_to_add = np.array([100, 200, 300, 400])   # shape (4,)
print("matrix:\n", matrix)
print("matrix + row_to_add:\n", matrix + row_to_add)

# Column broadcast (needs shape (3,1) not (3,))
col_to_add = np.array([[10], [20], [30]])   # shape (3,1) — KEY DIFFERENCE
print("matrix + col_to_add:\n", matrix + col_to_add)

# Broadcasting failure — see what the error says
try:
    bad = np.array([1, 2, 3])   # shape (3,) — 3 doesn't align with 4 cols
    print(matrix + bad)
except ValueError as e:
    print("Expected error:", e)
```

### Observe:
- Why does `row_to_add` need shape `(4,)` (matching the number of columns), not `(3,)` (matching the number of rows)?
- Try `col_to_add = np.array([10, 20, 30])` (shape `(3,)` instead of `(3,1)`) — does it broadcast correctly across columns now? Why not?

## 7. Aggregations useful for embeddings — std, sorting, argsort

```python
np.random.seed(42)
scores = np.random.rand(6)   # simulated similarity scores
print("scores:", scores.round(3))

# Sort
print("sorted:", np.sort(scores).round(3))
print("argsort (indices that would sort):", np.argsort(scores))

# Top-k indices (like 'retrieve top 3')
top3_indices = np.argsort(scores)[-3:][::-1]   # last 3, reversed (descending)
print("top-3 indices (descending score):", top3_indices)
print("top-3 scores:", scores[top3_indices].round(3))

# Stacking arrays (often needed when combining results)
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("hstack:", np.hstack([a, b]))      # [1 2 3 4 5 6]
print("vstack:\n", np.vstack([a, b]))   # [[1 2 3], [4 5 6]]
```

### Observe:
- `np.argsort(scores)[-3:][::-1]` — trace through this step by step. What does `[-3:]` do? What does `[::-1]` do?
- This exact pattern (`argsort` to get top-k indices) is used internally by `SimpleVectorStore.search()` in `rag_pure_python.py`. Look at that file and find the equivalent line.

## 8. The teaser: normalize rows of a 2D array (real embedding use case)

```python
# Simulated embedding matrix: 5 documents, 4 dimensions each
np.random.seed(42)
embeddings = np.random.randn(5, 4)
print("embeddings shape:", embeddings.shape)
print("embeddings:\n", embeddings.round(3))

# WRONG: this gives scalar norm of the whole matrix
scalar_norm = np.linalg.norm(embeddings)
print("\nscalar norm (wrong):", scalar_norm)

# CORRECT: norm per row, keeping shape for broadcasting
row_norms = np.linalg.norm(embeddings, axis=1, keepdims=True)
print("row norms shape:", row_norms.shape)  # (5,1) not (5,)
print("row norms:\n", row_norms.round(3))

normalized = embeddings / row_norms
print("\nnormalized:\n", normalized.round(3))

# Verify: every row should now have length 1.0
verify = np.linalg.norm(normalized, axis=1)
print("row magnitudes after normalization:", verify.round(6))
print("all close to 1.0:", np.allclose(verify, 1.0))

# Why this matters: cosine similarity becomes just a dot product
# when both vectors are unit-normalized
a_norm = normalized[0]
b_norm = normalized[1]
cos_sim = np.dot(a_norm, b_norm)
print("\nCosine similarity (dot product of normalized vectors):", cos_sim.round(4))
```
