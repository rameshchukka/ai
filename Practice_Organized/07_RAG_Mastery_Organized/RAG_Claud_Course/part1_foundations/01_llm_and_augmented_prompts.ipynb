# 01 · LLM Calls & Augmented Prompts
The heart of RAG: take a question, add retrieved context, and build the prompt you send to the LLM. Here we use a MOCK 'LLM' so it runs offline — the prompt-building is the real lesson (that part is identical with a real model).

## 1. A mock LLM
Swap `mock_llm` for a real call (OpenAI/Together/your in-house model) later; the prompt you build is exactly the same.

```python
def mock_llm(prompt):
    """Pretend model: it just echoes which context it was given, to prove grounding.
    A real LLM would read the CONTEXT and answer the QUESTION in natural language."""
    # find the CONTEXT block and the QUESTION
    ctx = prompt.split("CONTEXT:")[-1].split("QUESTION:")[0].strip()
    q   = prompt.split("QUESTION:")[-1].strip()
    return f"[mock answer to '{q}' using {len(ctx.splitlines())} context lines]"
print("mock LLM ready")
```

## 2. The augmented prompt = instructions + context + question
This structure is the whole idea of RAG — the model answers from the *context you retrieved*, not its memory.

```python
def build_prompt(question, context_docs):
    context = "\n".join(f"- {d}" for d in context_docs)
    return f"""You are a helpful assistant. Answer using ONLY the context below.
If the answer is not in the context, say you don't know.

CONTEXT:
{context}

QUESTION: {question}"""

retrieved = [
    "Refunds are processed back to your original payment method in 5 business days.",
    "You can request a refund within 30 days of purchase for a full return.",
]
prompt = build_prompt("How long do refunds take?", retrieved)
print(prompt)
print("\n--- model output ---")
print(mock_llm(prompt))
```

**Observe:** the prompt has three parts — instructions, retrieved CONTEXT, and the QUESTION. Change `retrieved` to unrelated docs and the (real) model would say it doesn't know. Grounding the model to the context is what stops hallucination.
