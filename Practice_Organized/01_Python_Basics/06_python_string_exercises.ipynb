## Concepts you'll need

This set is a deep dive into **string manipulation** — one of the most common tasks in everyday Python.

- **Indexing & slicing** — `s[0]` (first), `s[-1]` (last), `s[a:b]` (substring, `b` exclusive), `s[::-1]` (reversed). Slicing never raises an index error even at boundaries.
- **Search methods** — `.find()` / `.rfind()` (first/last position of a substring, or -1), `.count()`, `.startswith()` / `.endswith()`, and the `in` operator for membership.
- **Splitting & joining** — `.split(sep)` breaks a string into a list; `.partition(sep)` splits into exactly a (before, separator, after) 3-tuple; `"".join(list)` glues a list of strings back together.
- **Case & classification** — `.lower()`, `.upper()`, `.swapcase()`, `.capitalize()`; `.isalpha()`, `.isdigit()`, `.isalnum()`, `.isspace()`, `.islower()` for testing what kind of characters you're looking at.
- **Cleaning** — `.replace(old, new)` for global substitution, `.strip()` for trimming ends only (not used much here since several exercises need *all* whitespace removed, not just the edges).
- **Immutability** — strings can't be edited in place; every "modification" actually builds and returns a brand-new string, often via slicing + concatenation or list comprehension + `.join()`.
- **Frequency & lookup** — dictionaries (`dict.get(key, default)`) and sets (fast membership, automatic de-duplication) come up repeatedly for counting characters/words and preserving or dropping order.

Each exercise below gives a problem, a hint, a solution, and an explanation.

## Exercise 1. First, Middle, and Last Character

**Concept:** indexing, integer division

**Problem:** Build a new string from a word's first, middle, and last characters.

**Given:**
```
str1 = "James"
```

**Expected Output:**
```
Jms
```

**Hint:** str1[0] and str1[-1] give first/last; len(str1) // 2 (or int(len/2)) gives the middle index.

````python
str1 = "James"
print("Original String is", str1)

first_char = str1[0]

res = len(str1)
middle_index = int(res / 2)
mid_char = str1[middle_index]

last_char = str1[-1]

res_str = first_char + mid_char + last_char
print("New String:", res_str)
````

**Explanation:** len(str1) gives the character count, and dividing by 2 with int() truncates to a valid integer index (5/2 = 2.5 -> 2, landing on 'm'). Negative indexing makes str1[-1] a safe way to grab the last character regardless of string length.

## Exercise 2. Middle Three Characters

**Concept:** slicing around a computed midpoint

**Problem:** Extract the middle three characters of a string of odd length.

**Given:**
```
str1 = "JhonDipPeta"
```

**Expected Output:**
```
Dip
```

**Hint:** Find the middle index, then slice from middle-1 to middle+2 (slice stop is exclusive).

````python
def get_middle_three_chars(str1):
    print("Original String is", str1)
    mi = int(len(str1) / 2)
    res = str1[mi - 1:mi + 2]
    print("Middle three chars are:", res)

get_middle_three_chars("JhonDipPeta")
get_middle_three_chars("JaSon")
````

**Explanation:** mi is the midpoint index. Slicing from mi-1 to mi+2 captures three characters centered on that midpoint, since a slice's stop bound is exclusive — mi+2 ensures the character at mi+1 is the last one included.

## Exercise 3. Insert a String in the Middle of Another

**Concept:** split + concatenate

**Problem:** Insert one string into the exact middle of another.

**Given:**
```
s1 = "Ault", s2 = "Kelly"
```

**Expected Output:**
```
AuKellylt
```

**Hint:** Split s1 into two halves at its midpoint, then glue them back together with s2 in between.

````python
def append_middle(s1, s2):
    print("Original Strings are", s1, s2)
    mi = int(len(s1) / 2)
    x = s1[:mi]
    y = s1[mi:]
    res = x + s2 + y
    print("After appending new string in middle:", res)

append_middle("Ault", "Kelly")
````

**Explanation:** s1[:mi] takes everything up to the midpoint, and s1[mi:] takes everything from there onward. Concatenating x + s2 + y effectively inserts s2 right at the split point.

## Exercise 9. Character Balance Check

**Concept:** membership loop with an early-exit flag

**Problem:** Check whether every character of one string is present somewhere in another string (case-sensitive, position doesn't matter).

**Given:**
```
s1 = "yt", s2 = "PyNative" (True); s1 = "ytf" (False)
```

**Expected Output:**
```
True / False
```

**Hint:** Loop through s1; if any character isn't found in s2 via `in`, flip a flag to False and stop checking.

````python
def string_balance_check(s1, s2):
    flag = True
    for char in s1:
        if char in s2:
            continue
        else:
            flag = False
            break
    return flag

s1 = "yt"
s2 = "PyNative"
print("Is s1 and s2 balanced:", string_balance_check(s1, s2))

s1 = "ytf"
s2 = "PyNative"
print("Is s1 and s2 balanced:", string_balance_check(s1, s2))
````

**Explanation:** Only the shorter 'test' string needs to be looped over — each of its characters is checked for membership in the longer string via `in`. break stops the loop the instant a missing character is found, since the answer is already known.

## Exercise 17. Lowercase Letters First

**Concept:** two-bucket partitioning by character property

**Problem:** Rearrange a string's characters so all lowercase letters come first, then all uppercase letters.

**Given:**
```
str1 = "PyNaTive"
```

**Expected Output:**
```
yaivePNT
```

**Hint:** Route each character into one of two lists based on .islower(), then join both lists together.

````python
str1 = "PyNaTive"
lower = []
upper = []

for char in str1:
    if char.islower():
        lower.append(char)
    else:
        upper.append(char)

res = "".join(lower + upper)
print("Result:", res)
````

**Explanation:** Every character is sorted into exactly one of two buckets based on .islower(), preserving each bucket's relative order. Concatenating lower + upper before joining places all lowercase letters ahead of all uppercase ones.

## Exercise 18. Count Letters, Digits, and Symbols

**Concept:** isalpha()/isdigit() classification

**Problem:** Count how many letters, digits, and other symbols appear in a string.

**Given:**
```
str1 = "P@#yn26at^&i5ve"
```

**Expected Output:**
```
Chars = 8 Digits = 3 Symbol = 4
```

**Hint:** Check isalpha() and isdigit() in order; anything that fails both must be a symbol.

````python
def count_chars_digits_symbols(sample_str):
    char_count = 0
    digit_count = 0
    symbol_count = 0
    for char in sample_str:
        if char.isalpha():
            char_count += 1
        elif char.isdigit():
            digit_count += 1
        else:
            symbol_count += 1

    print("Chars =", char_count, "Digits =", digit_count, "Symbol =", symbol_count)

sample_str = "P@#yn26at^&i5ve"
print("total counts of chars, digits, and symbols")
count_chars_digits_symbols(sample_str)
````

**Explanation:** Every character falls into exactly one of three buckets: isalpha() catches letters, isdigit() catches digits, and the else branch catches everything else — since those two checks together cover every letter and number, whatever's left must be a symbol.

## Exercise 19. Alternate Characters From Two Strings

**Concept:** dual-direction indexing

**Problem:** Interleave characters starting from the front of one string and the back of another.

**Given:**
```
s1 = "Abc", s2 = "Xyz"
```

**Expected Output:**
```
AzbycX
```

**Hint:** Reverse s2 first, then pull matching-index characters from both strings using the same loop variable.

````python
s1 = "Abc"
s2 = "Xyz"

s1_length = len(s1)
s2_length = len(s2)

length = s1_length if s1_length > s2_length else s2_length
result = ""

s2 = s2[::-1]

for i in range(length):
    if i < s1_length:
        result = result + s1[i]
    if i < s2_length:
        result = result + s2[i]

print(result)
````

**Explanation:** Reversing s2 up front means the same index i can be used to pull 'next from the front' of s1 and 'next from the back' of the original s2 simultaneously. The two length checks (i < s1_length / i < s2_length) prevent an out-of-range error if the two strings differ in length.

## Exercise 20. Sum and Average of Digits in a String

**Concept:** filter with isdigit(), then int() conversion

**Problem:** Find the sum and average of the numeric digits embedded in a string, ignoring everything else.

**Given:**
```

```

**Expected Output:**
```

```

**Hint:** Loop through, check isdigit(), and convert matching characters to int() before adding them up.

````python
str1 = "PYnative29@#8496"
total_sum = 0
cnt = 0

for char in str1:
    if char.isdigit():
        total_sum += int(char)
        cnt += 1

avg = total_sum / cnt

print("Sum is:", total_sum, "Average is", round(avg, 2))
````

## Exercise 21. Character Frequency Dictionary

**Concept:** dictionary-based counting

**Problem:** Count how many times each character appears in a string.

```
{'a': 1, 'p': 2, 'l': 1, 'e': 1}
```

## Exercise 23. Remove Punctuation From a String

## Exercise 25. Find Words With Both Letters and Numbers

**Concept:** any() checks combined with `and`

**Problem:** Find words in a sentence that contain at least one letter AND at least one digit.

**Given:**
```
str1 = "Emma25 is Data scientist50 and AI Expert"
```

**Expected Output:**
```
Emma25
scientist50
```

**Hint:** For each word, require any(char.isalpha()...) and any(char.isdigit()...) to both be true.

````python
str1 = "Emma25 is Data scientist50 and AI Expert"
print("The original string is:", str1)

res = []
temp = str1.split()

for item in temp:
    if any(char.isalpha() for char in item) and any(char.isdigit() for char in item):
        res.append(item)

print("Words with alphabets and numbers:")
for i in res:
    print(i)
````

**Explanation:** any(condition for char in item) is true if at least one character in the word satisfies the condition. Requiring both the letter-check and the digit-check to pass filters out words that are purely letters (like 'Data') or purely numbers.

## Exercise 27. Palindrome Check

**Concept:** case-normalized reversal comparison

**Problem:** Check whether a string reads the same forwards and backwards.

**Given:**
```
"radar"
```

**Expected Output:**
```
True
```

**Hint:** Lowercase the string first, then compare it to its own [::-1] reversal.

````python
def is_palindrome(s):
    s = s.lower()
    return s == s[::-1]

print("Is 'radar' a palindrome?", is_palindrome("radar"))
print("Is 'Python' a palindrome?", is_palindrome("Python"))
````

**Explanation:** Lowercasing first means a palindrome like 'Racecar' isn't incorrectly rejected just because 'R' != 'r'. == then checks whether every character matches its mirrored counterpart in the reversed string.

## Exercise 28. Anagram Detector

**Concept:** sorted() as a canonical form

**Problem:** Check whether two strings are anagrams of each other.

**Given:**
```
s1 = "listen", s2 = "silent"
```

**Expected Output:**
```
True
```

**Hint:** sorted(s1.lower()) == sorted(s2.lower()) — anagrams produce identical sorted character lists.

````python
def check_anagram(s1, s2):
    return sorted(s1.lower()) == sorted(s2.lower())

s1 = "listen"
s2 = "silent"

if check_anagram(s1, s2):
    print(f"{s1} and {s2} are anagrams.")
else:
    print(f"{s1} and {s2} are NOT anagrams.")
````

**Explanation:** sorted() rearranges each string's characters into the same canonical alphabetical order. Two strings made of exactly the same characters, just shuffled, will always end up as identical sorted lists.

## Exercise 29. Check for All-Unique Characters

**Concept:** comparing lengths, set vs. string

**Problem:** Determine whether every character in a string is distinct.

**Given:**
```
"python" (True), "alphabet" (False)
```

**Expected Output:**
```
True / False
```

**Hint:** Converting to a set() drops duplicates automatically — compare its length to the original string's length.

````python
def is_unique(s):
    char_set = set(s)
    return len(char_set) == len(s)

str1 = "python"
str2 = "alphabet"

print(f"'{str1}' unique: {is_unique(str1)}")
print(f"'{str2}' unique: {is_unique(str2)}")
````

**Explanation:** A set can only hold each distinct value once, so converting the string to a set silently drops any repeated characters. If nothing was dropped — meaning the set is the same length as the original string — every character must have been unique.

**Given:**
```
str1 = "google"
```

**Expected Output:**
```
gole
```

**Hint:** Track already-seen characters in a set; only append a character to the result the first time it appears.

````python
str1 = "google"
seen = set()
result = []

for char in str1:
    if char not in seen:
        result.append(char)
        seen.add(char)

res_str = "".join(result)
print("Original:", str1)
print("Cleaned :", res_str)
````

**Explanation:** Checking membership in the seen set is fast regardless of its size. Because the loop moves left to right through the original string, result naturally collects each character in the order it was first encountered, unlike a plain set() conversion which would scramble that order.

**Hint:** Track a `longest` variable; only replace it when a word is strictly longer (>) than the current record.

## Exercise 36. Word Frequency Dictionary

**Concept:** dict.get() as a safe counter

**Problem:** Count how many times each word appears in a sentence.

**Given:**
```
str1 = "apple banana apple cherry banana apple"
```

**Expected Output:**
```
{'apple': 3, 'banana': 2, 'cherry': 1}
```

**Hint:** word_count.get(word, 0) + 1 increments a count safely, defaulting to 0 for a word's first appearance.

````python
str1 = "apple banana apple cherry banana apple"
words = str1.lower().split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
````

**Explanation:** .lower() ensures 'Apple' and 'apple' count as the same word. .get(word, 0) looks up the current count, returning 0 if the word hasn't been seen yet, so adding 1 always works without a separate existence check.

## Exercise 37. First Non-Repeating Character

**Concept:** two-pass algorithm: count, then check order

**Problem:** Find the first character in a string that doesn't repeat anywhere else in it.

**Given:**
```
"swiss"
```

**Expected Output:**
```
w
```

**Hint:** Build a frequency dictionary first, then loop the original string again looking for the first count-of-1 character.

````python
def first_unique(s):
    counts = {}
    for char in s:
        counts[char] = counts.get(char, 0) + 1

    for char in s:
        if counts[char] == 1:
            return char
    return None

print("First unique in 'swiss':", first_unique("swiss"))
````

**Explanation:** A single pass isn't enough because you can't know whether a character repeats until you've seen the whole string. The first loop builds a complete frequency count; the second loop then walks the string again in its original order, returning the first character whose count is exactly 1.

## Exercise 38. Check if One String Is a Rotation of Another

**Concept:** the double-concatenation trick

**Problem:** Determine whether one string is a rotation of another (like 'erbottlewat' is of 'waterbottle').

**Given:**
```
s1 = "waterbottle", s2 = "erbottlewat"
```

**Expected Output:**
```
True
```

**Hint:** s1 + s1 contains every possible rotation of s1 as a substring — just check if s2 is inside it.

````python
def is_rotation(s1, s2):
    if len(s1) != len(s2):
        return False

    combined = s1 + s1

    return s2 in combined

s1 = "waterbottle"
s2 = "erbottlewat"
print(f"Is '{s2}' a rotation of '{s1}'? {is_rotation(s1, s2)}")
````

**Explanation:** Concatenating s1 with itself produces a string containing every possible rotation of s1 as a contiguous substring. The length check up front is a quick way to rule out non-rotations before doing the (still fast, but unnecessary) substring search.
