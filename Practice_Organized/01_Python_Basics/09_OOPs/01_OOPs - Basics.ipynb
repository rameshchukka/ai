```python
class BasicClass:
    name="NareshIT"

    def print_something(self):
        print("hello intro to class")
        print(self.name)

    def assignname(self,newname):
        self.name=newname

bc1=BasicClass()
bc2=BasicClass()

bc1.assignname("Prudvi")
bc2.assignname("hello")

bc1.print_something()
bc2.print_something()
```

```python
class MyClass:
    def __init__(self, name, age):
        # These are instance-level variables
        self.name = name
        self.age = age
    
    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")

# Creating an instance of MyClass
person1 = MyClass("John", 30)
person2 = MyClass("Alice", 25)

# Accessing instance-level variables
person1.display_info()  # Output: Name: John, Age: 30
person2.display_info()  # Output: Name: Alice, Age: 25
```

```python
# Access Modifiers

class PersonalDetails:
    def __init__(self,name,age,bankpin):
        self.name=name # public
        self._age=age # protected
        self.__bankpin=bankpin # private

c1=PersonalDetails("prudvi",30,6765)

c1.name="krishna"

print(c1.name)

print(c1._age) # though it is protected can be used outside class as python donot restrict

#print(c1.__bankpin) # it is not accessible directly as python does mangling i.e. renaming variable to _classname__varname

print(c1._PersonalDetails__bankpin) # so you call as this way _classname__varname
```
