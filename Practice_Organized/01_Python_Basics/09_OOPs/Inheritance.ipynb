```python
#single inheritance
class Animal:
    def walk(self):
        print("Animal is walking")


class Dog(Animal):  # Dog inherits from Animal
    def bark(self):
        print("Dog barks")

dogobj= Dog()

dogobj.walk()
dogobj.bark()
```

```python
#Multiple Inheritance 

class Father:
    def speak(self):
        print("Father speaks")


class Mother:
    def speak(self):
        print("Mother speaks")


class Child(Father, Mother):  # Child inherits from both Father and Mother
    def speak(self):
        print("I am child")


childobj= Child()

childobj.speak()

```

```python
#multi level inheritance
class Grandparent:
    def greet(self):
        print("Hello from Grandparent")

class Parent(Grandparent):
    pass
    # def greet(self):
    #     print("Hello from Parent")

class Child(Parent):  # Inherits from Parent, which inherits from Grandparent
    pass

childobj= Child()

childobj.greet()

print(Child.mro()) # to the sequence of methods getting called

print(Child.__mro__)
```

```python
#hierarchial
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):  # Dog inherits from Animal
    def bark(self):
        print("Dog barks")

class Cat(Animal):  # Cat also inherits from Animal
    def meow(self):
        print("Cat meows")


dog = Dog()
dog.speak()  # Inherited from Animal
dog.bark()
cat = Cat()
cat.speak()  # Inherited from Animal
cat.meow()

```

```python
# hybrif - hierarchial & multiple
class Animal:
    def speak(self):
        print("Animal speaks")


class Mammal(Animal):
    def walk(self):
        print("Mammal walks")

class Bird(Animal):
    def fly(self):
        print("Bird flies")

# Inherits from both Mammal and Bird
class Bat(Mammal, Bird):
    def echo_location(self):
        print("Bat uses echo location")

bat = Bat()
bat.speak()         # Inherited from Animal
bat.walk()          # Inherited from Mammal
bat.fly()           # Inherited from Bird
bat.echo_location()  # Defined in Bat
```

```python
#Super()


class Animal():
    def __init__(self, name):
        self.name = name
    def display_info(self):
        return f"Animal Name: {self.name}"


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Calling the constructor of the parent class
        self.breed = breed
    def display_info(self):
        # Accessing the superclass's attribute and method
        animal_info = super().display_info()  # Call the parent's display_info method
        return f"{animal_info}, Breed: {self.breed}"
    

dog = Dog("Buddy", "Golden Retriever")

print(dog.display_info())  # Output: Animal Name: Buddy, Breed: Golden Retriever

```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name}, {self.age} years old"
    def __repr__(self):
        return f"Person('{self.name}', {self.age})"


p = Person("Alice", 30)

print(str(p))  # Output: Alice, 30 years old
print(repr(p))  # Output: Person('Alice', 30)

#to print all methods of person class including dunder methods
#print(dir(p))
```

```python
class Person:
    def __init__(self, newname, newage):
        self._name = newname            # Protected variable
        self.__age = newage             # Private variable
    def display_info(self):
        print(f"Name: {self._name}, Age: {self.__age}")

# Creating an instance of the Person class
person = Person("Alice", 30)

person.display_info()
```

```python
class Parent:
    def __init__(self):
        self._protected_attr = "I am a protected attribute"

    def show_protected(self):
        print(f"Parent class: {self._protected_attr}")


class Child(Parent):
    def __init__(self):
        super().__init__()
        #self._protected_attr ="name"
        print(f"Child class: {self._protected_attr}")  # Accessing the protected attribute of Parent

    def change_protected(self, new_value):
        self._protected_attr = new_value  # Modifying the protected attribute



# Create an instance of Child
child_instance = Child()

# Accessing and modifying the protected attribute in the child class
child_instance.show_protected()  # Displaying inherited protected attribute

child_instance.change_protected("New protected value")
child_instance.show_protected()  # Displaying modified value

child_instance._protected_attr
```
