```python
from abc import ABC, abstractmethod

class Shape(ABC):  # Abstract Base Class
    
    @abstractmethod
    def area(self):
        pass  # Abstract method, no implementation
    @abstractmethod
    def perimeter(self):
        pass  # Abstract method, no implementation

    def nonabstractmethod():
        print("i am non acstract method")
```

```python
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height  # Implementing abstract method
    
    def perimeter(self):
        return 2 * (self.width + self.height)  # Implementing abstract method
```

```python
# Creating an instance of Rectangle
rect = Rectangle(10, 20)
print("Area:", rect.area())         # Output: Area: 200
print("Perimeter:", rect.perimeter())  # Output: Perimeter: 60
```

```python
# This fucntion will call the fucntions of the class obj passed to it.
def function1(obj):
    return obj.area()


objrect= Rectangle(10,5)
print(objrect.area())
print(function1(objrect))
```

```python
# Using variable length parameters also you can achieve method overloading
class Dog:
    def __init__(self, *args):
        if len(args) == 2:
            self.name = args[0]
            self.age = args[1]
        elif len(args) == 1:
            self.name = args[0]
            self.age = 0
        else:
            self.name = "Unknown"
            self.age = 0

    def info(self):
        print(f"{self.name} is {self.age} years old.")

# Creating instances with different parameters
dog1 = Dog("Buddy", 5)
dog2 = Dog("Bella")
dog3 = Dog()  # Using default values

dog1.info()  # Output: Buddy is 5 years old.
dog2.info()  # Output: Bella is 0 years old.
dog3.info()  # Output: Unknown is 0 years old.
```

```python
#operator overloading
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        # Overloading the + operator
        return Vector(self.x + other.x, self.y + other.y)
    def __str__(self): # Overriding __str__ to print Vector in a readable format 
        return f"Vector({self.x}, {self.y})"
  
v1 = Vector(2, 4)
v2 = Vector(1, 3)

v3 = v1 + v2  # This calls v1.__add__(v2)
print(v3)  # Output: Vector(3, 7)
```

```python
#method Overriding
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):  # Overriding the speak method
        print("Dog barks")

class Cat(Animal):
    def speak(self):  # Overriding the speak method
        print("Cat meows")

# Polymorphic behavior
def animal_sound(animal):
    animal.speak()  # Calls the appropriate method based on the object type

dog = Dog()
cat = Cat()
animal_sound(dog)  # Output: Dog barks
animal_sound(cat)  # Output: Cat meows
```

```python
# Duck Typing
class Bird:
    def fly(self):
        return "Bird flies high in the sky!"

class Aerorplane:
    def fly(self):
        return "Aeroplane soars through the clouds!"
    def take_off(self): 
        return "The airplane is taking off!"

class Fish:
    def swim(self):
        return "Fish Swims in the ocean!"


def make_it_fly(flyable):
    print(flyable.fly())

# Creating instances of Bird and Airplane
bird = Bird()
airplane = Aerorplane()

# Both can be used interchangeably because they implement the fly method
make_it_fly(bird)      # Output: Flies high in the sky!
make_it_fly(airplane)  # Output: Soars through the clouds!

# The Fish class does not implement the fly method, so it would raise an error by calling fly
# Uncommenting the following line would raise an AttributeError
# fish = Fish()
# make_it_fly(fish)
```
