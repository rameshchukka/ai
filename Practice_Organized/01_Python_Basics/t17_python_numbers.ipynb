# Python Numbers

Python's numeric data comes in three built-in flavors — `int`, `float`, `complex` — plus two standard-library modules, `decimal` and `fractions`, for cases where binary floating-point isn't precise enough. This notebook covers creating each type, the `math` module's toolkit, several lesser-known number methods, and both `Decimal` and `Fraction`.

## Integer and Float

Create directly, or via the `int()`/`float()` constructors.

```python
roll_no = 33

print("Roll number is:", roll_no)

print(type(roll_no))



student_id = int(25)

print(student_id, type(student_id))
```

```python
x = -9999

y = 99.99

print(type(x))

print(type(y))
```

## Complex Numbers and the `cmath` Module

A complex number has a `.real` and `.imag` part. Beyond the basic `complex` type itself, the `cmath` module adds conversions between RECTANGULAR (real, imaginary) and POLAR (modulus `r`, phase angle `phi`) representations.

```python
import cmath



# phase angle (in radians) of a complex number

phi = cmath.phase(complex(-1.0, 1.0))

print(phi)



# build a complex number FROM polar coordinates (r=1, angle=2 radians)

c = cmath.rect(1, 2)

print("real:", c.real)

print("imaginary:", c.imag)

print("complex number:", c)



# convert a complex number back TO polar form: (modulus, phase angle)

p = cmath.polar(c)

print("polar:", p)
```

## Hexadecimal and Octal Conversion

`oct()` and `hex()` convert an integer to its string representation in those bases — note both come back as a `str`, prefixed with `0o` or `0x` respectively.

```python
i = 10

print("in octal form:", oct(i))

print("in hexadecimal form:", hex(i))
```

## `math` Module: Rounding

`ceil()` rounds UP to the nearest integer; `floor()` rounds DOWN.

```python
import math as m



a = 1.4

print(m.ceil(a))

print(m.floor(a))
```

## `math` Module: Angles and Trigonometry

`degrees()`/`radians()` convert between the two angle units; Python's trig functions (`sin`, `cos`, ...) always expect RADIANS.

```python
import math



pi = math.pi

print(pi)

print("Degree value of pi is ::", math.degrees(pi))

print("Radian value of 90 degree is ::", math.radians(90))
```

```python
radians_90 = math.radians(90)

print("sin value of pi/2 is ::", math.sin(radians_90))



radians_60 = math.radians(60)

print("cos value of 60 degrees is ::", math.cos(radians_60))
```

## `math` Module: Factorial

`factorial(n)` returns `n! = n x (n-1) x ... x 1`.

```python
import math

print("factorial of 6 is ::", math.factorial(6))
```

## `math` Module: `fabs()` and `trunc()`

`fabs()` always returns a FLOAT absolute value (works on ints too). `trunc()` chops off the decimal part entirely (toward zero) — different from `floor()`, which always rounds DOWN (toward negative infinity).

```python
import math



print("floating absolute value::", math.fabs(-99.999))

print("floating absolute value of an integer::", math.fabs(-11))

print("truncated integer value::", math.trunc(-99.99))
```

## `math` Module: `pow()` and `log()`

`math.pow(x, y)` returns `x` raised to `y` as a float; `math.log(x)` is the NATURAL log (base e).

```python
import math

print("The value of 9^3 is ::", math.pow(9, 3))

print("The log value of 5::", math.log(5))
```

## Checking Functions: `isfinite()`, `isinf()`, `isnan()`



```python
import math

print("Finite or not::", math.isfinite(-0.00006))

print("Infinite or not::", math.isinf(math.inf))

print("Is nan or not::", math.isnan(math.nan))
```

## `isclose()`: Safe Float Comparison

Never compare floats with `==` directly — tiny rounding errors make it unreliable. `isclose(a, b, rel_tol=..., abs_tol=...)` checks whether two numbers are 'close enough', which is almost always what you actually want.

```python
import math

print("with absolute tolerance difference::", math.isclose(1.5, 2.05, abs_tol=0.55))

print("with relative tolerance difference::", math.isclose(1.5, 1.75, rel_tol=0.5))
```

## The `decimal` Module

Python's `float` is stored internally as BINARY floating point — some decimal fractions (like 0.1) can't be represented exactly, just like 1/3 can't be written exactly in decimal (0.333...). For financial or precision-sensitive work, the `decimal` module represents numbers exactly in base 10 instead, avoiding that whole category of rounding surprise.

```python
from decimal import Decimal



print(Decimal(0.01))     # constructed from a FLOAT - shows the float's actual binary imprecision

print(Decimal('0.05'))   # constructed from a STRING - exact, no binary rounding noise

print(Decimal((1, (9, 9, 9, 9), -2)))  # (sign, digits, exponent) tuple form: -9999 * 10^-2 = -99.99
```

### Controlling Precision With `getcontext()`

`getcontext()` gives access to the CURRENT decimal precision/rounding settings, adjustable globally for all subsequent `Decimal` arithmetic.

```python
from decimal import Decimal, getcontext



print(Decimal(3) / Decimal(77))
```

```python
import decimal

from decimal import Decimal, getcontext



old_prec = getcontext().prec

old_rounding = getcontext().rounding



getcontext().prec = 2

getcontext().rounding = decimal.ROUND_UP

print(Decimal(3) / Decimal(77))



# restore the previous settings so later cells aren't affected

getcontext().prec = old_prec

getcontext().rounding = old_rounding
```

### Math Functions on `Decimal`

`Decimal` objects have their own built-in `.sqrt()`, `.ln()`, etc. `clear_flags()` resets internal state flags set by previous operations — good practice before changing precision requirements for a new calculation.

```python
import decimal

from decimal import Decimal, getcontext

getcontext().clear_flags()

print(Decimal(2).sqrt())
```

## The `fractions` Module

`Fraction` represents an exact rational number — a genuine numerator/denominator pair, automatically reduced to lowest terms. Can be built from two integers, a string, a `Decimal`, or a `float`.

```python
from fractions import Fraction

from decimal import Decimal



print(Fraction(25, -100))          # from two integers - auto-reduces to lowest terms

print(Fraction('-3/7'))            # from a string

print(Fraction(Decimal('9.9')))    # from a Decimal - exact

print(Fraction(9.1))               # from a float - shows the float's OWN binary imprecision as a huge fraction

print(Fraction())                  # empty -> defaults to 0/1
```

### `limit_denominator()` — Approximate With a Simpler Fraction

Useful for finding a 'nice' rational approximation of an irrational or messy number.

```python
from fractions import Fraction

import math



f = Fraction(math.pi).limit_denominator(100)

print("The fraction created is::", f)

print("Numerator::", f.numerator)

print("Denominator::", f.denominator)
```

### `Fraction.from_float()` and `ceil()`/`floor()` on a `Fraction`

```python
from fractions import Fraction

import math



print(Fraction.from_float(0.25))

print(math.ceil(Fraction(300, 124)))

print(math.floor(Fraction(300, 124)))
```

## Printing Numbers in Specific Formats

`str.format()`'s width/precision specifiers (covered in the Input/Output notebook) apply just as well to standalone number formatting — no `input()` needed.

```python
print("Total 6 digits with 4 after the decimal point:")

print('{:06.4f}'.format(3.141592653589793))



print("6 digits including the sign:")

print('{:=6d}'.format(-234))
```

### Exercise 2 solution

```python
import math

def is_probably_integer(f, tolerance=1e-9):

    return math.isclose(f, round(f), abs_tol=tolerance)



print(is_probably_integer(4.0))

print(is_probably_integer(4.0000000001))

print(is_probably_integer(4.5))
```
