# Python Comments

A comment is a note left in the source code for humans, not the interpreter — a short explanation of *why* the code does something, aimed at the next person reading it (often a future version of yourself). Python's interpreter skips comments entirely when it runs a script.

Python only has one native comment marker: the hash symbol `#`. Anything after a `#` (and a following space) is ignored, all the way to the end of that line. There's no separate multi-line comment syntax — you either repeat `#` on every line, or lean on an unassigned triple-quoted string (more on both below).

This notebook covers single-line comments, the 'multi-line via repeated #' pattern, writing genuinely *useful* comments, inline vs. block comments, docstrings, and the trick of temporarily commenting out code while debugging.

## A Basic Comment

```python
x = 10

y = 20



# adding two numbers

z = x + y

print('Sum:', z)
```

## Single-Line Comments

The most common form: a `#` at the start of a line, explaining the statement right below it. PEP 8 (Python's official style guide) suggests keeping comment lines to a maximum of about 79 characters for readability.

```python
# welcome message

print('Welcome to PYnative...')
```

## Multi-Line Comments (Repeated `#`)

Python has no dedicated multi-line comment syntax. The conventional workaround is simply repeating `#` at the start of every line that's part of the comment block.

```python
# This is a

# multiline

# comment

print('Welcome to PYnative...')
```

## Writing Sensible Comments

A good comment adds information the code *doesn't already make obvious*. Restating exactly what a line does in English, when the code is already self-explanatory, just adds noise.

**Don't do this** — the code already says everything the comment does:

```python
# Define list of student names

names = ['Jess', 'Emma', 'Kelly']



# iterates over name list and prints each name

for student in names:

    print(student, end=' ')
```

**Do this instead** — comments that explain *purpose*, parameters, and return value, things the code's syntax can't communicate on its own:

```python
# Returns welcome message for a customer by customer name and location

# param name - Name of the customer

# param region - location

# return - Welcome message



def greet(name, region):

    message = get_message(region)

    return message + " " + name



# Returns welcome message by location

# param region - location

def get_message(region):

    if region == 'USA':

        return 'Hello'

    elif region == 'India':

        return 'Namaste'



print(greet('Jessa', 'USA'))
```

## Inline Comments

An **inline comment** (also called a *trailing comment*) sits on the same line as the code it explains. Convention: separate it from the code by at least two spaces, then start with `#` and a single space. Inline comments work best for a short clarification — like the meaning of a magic number in a formula.

```python
def calculate_bonus(salary):

    salary = salary * 7.5  # yearly bonus percentage 7.5

    return salary



print(calculate_bonus(1000))
```

## Block Comments

A **block comment** describes an entire file, class, or function, and typically sits directly above the thing it describes — a header explaining what follows, rather than annotating a single line. A blank line afterward helps visually separate it from the surrounding code.

```python
# Returns welcome message for a customer by customer name and location

# param name - Name of the customer

# param region - location

# return - Welcome message



def greet(name, region):

    message = get_message(region)

    return message + " " + name
```

## Docstring Comments

A **docstring** is a special comment placed as the very first line inside a function, class, or module — written as a string literal (usually triple-quoted) rather than with `#`. Unlike an ordinary comment, a docstring is actually *stored* by Python and accessible at runtime via `.__doc__` or `help()`.

Conventions from [PEP 257](https://www.python.org/dev/peps/pep-0257): a docstring goes right after the `def`/`class` line, can be one line or several, and should begin with a capital letter and end with a period.

```python
def bonus(salary):

    """Calculate the bonus 10% of a salary."""

    return salary * 10 / 100



print(bonus(50000))

print(bonus.__doc__)

help(bonus)
```

## Commenting Out Code for Debugging

A very common practical use of `#`: temporarily disabling a line (or block) of code while tracking down a bug, without deleting it outright — so you can quickly restore it once you've isolated the problem.

```python
def greet(name, region):

    # below code is commented out for testing

    # message = get_message(region)

    message = 'Hello'

    return message + " " + name



def get_message(region):

    if region == 'USA':

        return 'Hello'

    elif region == 'India':

        return 'Namaste'



print(greet('Jessa', 'USA'))
```

## Using an Unassigned String Literal as a Multi-Line Comment

A triple-quoted string (`'''...'''` or `"""..."""`) that ISN'T assigned to any variable, and isn't the first statement in a function/class (which would make it a docstring), gets evaluated and then simply discarded by the interpreter — nothing happens with the value, since nothing captured it. This makes it usable as an ad-hoc multi-line comment.

**Important caveat:** this isn't technically a comment at all — it's a real string literal that Python actually builds in memory each time that line runs, then throws away. For genuinely large blocks of disabled code, most editors' 'toggle comment' shortcut (which just adds `#` to every selected line) is the more standard and more efficient approach.

```python
'''

I am a

multiline comment!

'''

print("Welcome to PYnative..")
```

## Summary

- Python's only native comment marker is `#` — everything after it, to the end of the line, is ignored by the interpreter.
- There's no dedicated multi-line comment syntax; repeat `#` on each line, or (less idiomatically) use an unassigned triple-quoted string.
- Good comments explain *why*, not *what* — avoid restating obvious code in English.
- **Inline** comments sit beside a line of code; **block** comments sit above a whole section, file, or function.
- **Docstrings** are a special first-statement string in a function/class/module — they're real, retrievable data (`.__doc__`, `help()`), not just a disregarded comment.
- `#`-ing out code temporarily is a common, quick debugging technique.
