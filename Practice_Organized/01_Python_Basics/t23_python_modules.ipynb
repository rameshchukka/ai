# Python Modules

A **module** is simply a `.py` file containing Python code — statements, functions, classes, variables — that can be IMPORTED and reused in other files. Splitting a large program into smaller modules keeps related code organized together and makes it reusable across multiple programs.

- **Built-in modules** — ship with Python itself: `math`, `os`, `sys`, `random`, `datetime`, ...
- **User-defined modules** — any `.py` file YOU write, importable by its filename (minus the `.py`).

This notebook creates and imports REAL module files on disk, so every example here is a genuine, working demonstration rather than a simulation.

## Importing a Built-in Module

`import module_name` loads the whole module; access its contents with `module_name.thing`.

```python
import math

print(math.sqrt(5))
```

### Importing Multiple Modules at Once

```python
import math, random



print(math.factorial(5))

print(random.randint(10, 20))
```

## `from ... import` — Importing Specific Names

Pulls specific functions/classes directly into your own namespace — you can then call them WITHOUT the module-name prefix.

```python
from math import factorial

print(factorial(5))
```

## Renaming on Import — `as`

Give an imported module (or a specific imported name) a different local name — commonly used for brevity (`import numpy as np`) or to avoid a naming clash.

```python
import random as rand

print(rand.randrange(10, 20, 2))
```

```python
from random import randint as random_number

print(random_number(10, 50))
```

## Importing Everything: `from module import *`

Pulls EVERY public name from a module directly into your namespace. Convenient, but generally discouraged in real code beyond quick scripts — it makes it unclear where a given name actually came from, and risks silently overwriting names you already have.

```python
from math import *



print(pow(4, 2))

print(factorial(5))

print(pi * 3)

print(sqrt(100))
```

### Variables in a Module

A module can hold variables too — lists, dicts, anything — accessible the same way, via `module_name.variable_name`.

```python
module_code = """cities_list = ['Mumbai', 'Delhi', 'Bangalore', 'Karnataka', 'Hyderabad']

"""



with open("test_module.py", "w") as f:

    f.write(module_code)



import test_module



# index 1 is the SECOND city (0-indexed) - 'Delhi'

city = test_module.cities_list[1]

print("Accessing 2nd city:", city)



cities = test_module.cities_list

print("Accessing all cities:", cities)
```

## The Python Module Search Path

When you `import` something, Python searches (in order): built-in modules first, then every directory listed in `sys.path` — populated from the running script's own directory, the `PYTHONPATH` environment variable, and the installation's default library locations.

```python
import sys

print(f"Number of search path entries: {len(sys.path)}")

for p in sys.path[:5]:

    print(" ", p if p else "(current directory)")
```

## Reloading a Module

A module is loaded (and its top-level code executed) only ONCE per program run, no matter how many times you `import` it afterward — later imports just reuse what's already loaded. `importlib.reload()` forces Python to re-run the module's code, picking up any changes made to the file since it was first imported.

```python
module_code_v1 = 'print("Welcome to PYnative - version 1")'

with open("reload_demo.py", "w") as f:

    f.write(module_code_v1)



import reload_demo          # runs the module's top-level code once

import reload_demo          # already loaded - does NOT print again
```

```python
from importlib import reload



# Now genuinely change the file on disk

module_code_v2 = 'print("Welcome to PYnative - version 2 (updated!)")'

with open("reload_demo.py", "w") as f:

    f.write(module_code_v2)



reload(reload_demo)   # forces the NEW code to actually run

reload(reload_demo)   # each reload() call re-runs it again
```

## The `dir()` Function

Lists every accessible name (functions, classes, variables, and special `__dunder__` attributes) belonging to a module or object — useful for exploring what's available without checking documentation.

```python
import math

print(dir(math))
```

```python
import mymodule

print(dir(mymodule))
```

Calling `dir()` with NO argument at all lists names currently available in your own local scope.

```python
x = 10

y = "hello"

def my_local_func():

    pass



names = dir()

print([n for n in names if not n.startswith('_')])
```

## Summary

- A module is just a `.py` file — anything you can write in a script, you can put in a module and import elsewhere.
- `import module` loads the whole module; `from module import name` pulls specific pieces directly into your namespace.
- `as` renames an import, for brevity or to avoid clashes.
- `from module import *` imports everything — convenient but can obscure where names came from; use sparingly.
- Modules are loaded (and their top-level code run) only ONCE per program run; `importlib.reload()` forces a re-run after changes.
- `dir()` lists a module's (or your own scope's) available names.

### Exercise 5 solution

```python
import sys

from importlib import reload



# Disabling bytecode caching for this demo avoids a subtle real-world gotcha:

# both file versions below are the same size and get written within the same

# wall-clock second, so a cached .pyc can look 'fresh' even after the source

# changed, causing reload() to silently reuse stale bytecode.

sys.dont_write_bytecode = True



with open("versioned.py", "w") as f:

    f.write("VERSION = 1\n")



import versioned

print("First import:", versioned.VERSION)



with open("versioned.py", "w") as f:

    f.write("VERSION = 2\n")



import versioned   # plain re-import - does NOT pick up the change

print("After plain re-import:", versioned.VERSION)



reload(versioned)   # forces the change to actually be picked up

print("After reload():", versioned.VERSION)
```
