# Python `for` Loop

A `for` loop iterates over a sequence or any iterable — string, list, tuple, dict, set, or `range()` — running the same block of code once per item. Use `for` when you know (or can determine) how many times you need to repeat something — a fixed or 'definite' number of iterations. Contrast this with `while` (its own dedicated notebook), which repeats based on a condition staying true, for a genuinely unknown number of iterations.

```
for i in sequence:
    statement 1
    statement 2
```

`i` (the **iterating variable**) takes on the current item's value on each pass, automatically, with no manual indexing required.

````python
for num in range(10):

    print(num)
````

## `for` Loop with `range()`

`range()` generates a sequence of integers — by default starting at 0, incrementing by 1 — and is the most common thing to iterate with `for` when you just need to repeat something N times, or count through a specific numeric range. `range(start, stop, step)` fully controls the sequence.

````python
# Sum of all even numbers from 2 to 20

total = 0

for i in range(2, 22, 2):   # step of 2 lands only on even numbers

    total = total + i

print(total)
````

## How `for` Loop Works

On each pass, the loop pulls the next item from the sequence, runs the body once with `i` set to that item, then checks whether more items remain — repeating until the sequence is exhausted.

for loop flowchart

````python
numbers = [1, 2, 3, 4, 5]

for i in numbers:

    square = i ** 2

    print("Square of:", i, "is:", square)
````

## Why Use `for` Loop?

- **Definite iteration** — when you know the exact count ahead of time (e.g. processing exactly 50 student records).
- **Reduces complexity** — one block of code handles every item, instead of duplicating the same logic manually.
- **Natural fit for sequences** — the idiomatic way to process lists, strings, tuples, dicts, and similar collections.

````python
numbers = [10, 20, 30, 40, 50]



total = 0

for i in numbers:

    total = total + i

average = total / len(numbers)

print(average)
````

## `if-else` Inside a `for` Loop

Combining a conditional with a loop lets you handle each item differently based on some test — checked fresh, every single iteration.

````python
for i in range(1, 11):

    if i % 2 == 0:

        print('Even Number:', i)

    else:

        print('Odd Number:', i)
````

**Practice problem from the source material:** generate numbers from 9 to 50 that are divisible by 2.

````python
for i in range(9, 51):

    if i % 2 == 0:

        print(i, end=" ")
````

## Loop Control Statements: `break`, `continue`, `pass`

### `break` — stop the loop entirely

````python
numbers = [1, 4, 7, 8, 15, 20, 35, 45, 55]

for i in numbers:

    if i > 15:

        break

    else:

        print(i)
````

**Note:** inside a nested loop, `break` only terminates the INNERMOST loop, not any outer loops around it.

### `continue` — skip to the next iteration

````python
name = "mariya mennen"

count = 0

for char in name:

    if char != 'm':

        continue

    else:

        count = count + 1



print('Total number of m is:', count)
````

### `pass` — a deliberate no-op placeholder

````python
num = [1, 4, 5, 3, 7, 8]

for i in num:

    # calculate multiplication in future if required

    pass

print("Loop finished, did nothing yet.")
````

## `else` Block in a `for` Loop

Python's `for` loops can have an `else` clause — it runs ONLY when the loop finishes NORMALLY (i.e. the sequence ran out on its own). If `break` cuts the loop short, `else` is skipped entirely. This is a handy way to detect 'the loop completed without needing to bail out early'.

````python
for i in range(1, 6):

    print(i)

else:

    print("Done")
````

````python
count = 0

for i in range(1, 6):

    count = count + 1

    if count > 2:

        break

    else:

        print(i)

else:

    print("Done")  # skipped! the loop was cut short by break
````

## Reverse `for` Loop

Two common ways to iterate backward: `reversed()`, or `range()` with a negative step.

````python
list1 = [10, 20, 30, 40]

for num in reversed(list1):

    print(num)
````

````python
num = 5

# start=5, stop=-1 (exclusive, so 0 is included), step=-1

for num in range(num, -1, -1):

    print(num, end=" ")
````

````python
numbers = [1, 2, 3, 4]

for i in numbers[::-1]:   # slicing with a negative step

    print(i, end=" ")
````

## Nested `for` Loops

A loop inside another loop — the inner loop runs to completion once for EVERY single iteration of the outer loop. Commonly used for anything grid-like: 2D data, tables, or pattern printing.

nested for loop diagram

````python
rows = 5

for i in range(1, rows + 1):     # outer loop: which row

    for j in range(1, i + 1):    # inner loop: how many stars in this row

        print("*", end=" ")

    print('')
````

### `while` Loop Inside a `for` Loop

Loop types can be freely mixed and nested — here a `for` loop picks each number, and an inner `while` loop prints that number's multiplication table.

````python
for i in range(1, 6):

    print('Multiplication table of:', i)

    count = 1

    while count < 11:

        print(i * count, end=' ')

        count = count + 1

    print('\n')
````

## `for` Loop in One Line

A comprehension condenses a simple `for` loop (with an optional filter) into one expression — see the dedicated Comprehensions material for the full picture. Quick preview:

````python
odd = [1, 5, 7, 9]

even = [i + 1 for i in odd if i % 2 == 1]

print(even)
````

## Accessing the Index While Looping

`enumerate()` is the idiomatic way to get both a value AND its position at once, without manual index bookkeeping.

````python
numbers = [4, 2, 5, 7, 8]

for i, v in enumerate(numbers):

    print('Numbers[', i, '] =', v)
````

An older, more manual alternative uses `range(len(...))` — works, but `enumerate()` is generally preferred as more direct and readable.

````python
numbers = [1, 2, 4, 6, 8]

size = len(numbers)

for i in range(size):

    print('Index:', i, ' ', 'Value:', numbers[i])
````

## Iterating a String

Strings are sequences of characters, so `for` walks through them one character at a time, just like a list.

````python
name = "Jessa"

for i in name:

    print(i, end=' ')
````

````python
name = "Jessa"

for i in name[::-1]:

    print(i, end=' ')
````

````python
name = "Jessa watson"

for char in name[2:7:1]:

    print(char, end=' ')
````

````python
dialogue = "Remember hope is a good thing maybe the best of things and no good thing ever dies"

# split() breaks the string into words at each whitespace

for word in dialogue.split():

    print(word)
````

## Iterating a List

Three equivalent styles for the same task — direct iteration is the most idiomatic; index-based and comprehension styles have their place too.

````python
numbers = [2, 3, 5, 6, 7]

for num in numbers:

    print(num)
````

````python
numbers = [1, 2, 3, 6, 7]

size = len(numbers)

for i in range(size):

    print(numbers[i])
````

````python
numbers = [1, 2, 3, 7, 8]

# list comprehension used purely for its side effect (printing) - unusual style,

# shown here because the source material demonstrates it, but a plain for loop

# is clearer when you're not actually building a new list from the result

[print(i) for i in numbers];
````

## Iterating a Dictionary

Looping a dict directly gives you its KEYS; use `.values()` for values only, or `.items()` (the most common choice) for both together.

````python
dict1 = {"Brand": "BMW", "Color": "Black", "Date": 1964}

for key in dict1:

    print(key)
````

````python
dict1 = {"Brand": "BMW", "Color": "Black", "Date": 1964}

for key in dict1:

    print(key, "->", dict1[key])
````

````python
dict1 = {"Brand": "BMW", "Color": "Black", "Date": 1964}

for value in dict1.values():

    print(value)
````

````python
dict1 = {"Brand": "BMW", "Color": "Black", "Date": 1964}

# .items() - the most common pattern - gives key AND value together in one step

for key, value in dict1.items():

    print(f"{key} -> {value}")
````

## Summary

- `for item in sequence:` iterates over any iterable, running the body once per item — ideal for a KNOWN or determinable number of repetitions.
- `range(start, stop, step)` is the most common thing to iterate for pure counting.
- `break` exits the loop early; `continue` skips to the next iteration; `pass` is a deliberate no-op.
- A `for` loop's `else` clause runs only if the loop was never `break`-ed out of.
- `reversed()` and negative-step slicing/`range()` both iterate backward.
- Nested loops run the inner loop to completion for every single outer-loop iteration — the standard tool for grid/table-shaped work.
- `enumerate()` is the idiomatic way to get index + value together.
- Iterating a dict directly gives keys; use `.values()` or `.items()` for the rest.

### Exercise 1 solution

````python
words = ["apple", "kiwi", "banana", "fig", "cherry"]

for i, word in enumerate(words):

    print(f"[{i}] {word} (length {len(word)})")
````

### Exercise 4 solution

````python
def find_target(items, target):

    for item in items:

        if item == target:

            print("found")

            break

    else:

        print("not found")



find_target([3, 7, 9, 12], 9)

find_target([3, 7, 9, 12], 100)
````

### Exercise 5 solution

````python
sentence = "the quick brown fox jumps over the lazy dog"

for word in sentence.split():

    if len(word) <= 4:

        continue

    print(word)
````
