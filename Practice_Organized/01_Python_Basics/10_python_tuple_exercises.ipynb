## Concepts you'll need

This set covers **tuples** — ordered, immutable sequences — and the ways their immutability shapes how you work with them.

- **Creation quirks** — parentheses alone don't make a tuple; `(50)` is just an int. A single-element tuple needs a trailing comma: `(50,)`.
- **Sequence operations shared with lists/strings** — indexing (`t[0]`, `t[-1]`), slicing (`t[2:5]`, `t[::-1]`), repetition (`t * 3`), concatenation (`t1 + t2`), and `len()`, `sum()`, `max()`, `min()` all work the same way.
- **Only two built-in methods** — `.count(value)` and `.index(value)`. That's it — tuples are deliberately minimal since they're not meant to be modified.
- **Unpacking** — `a, b, c = some_tuple` assigns each element to a name by position; the classic `a, b = b, a` swap works because Python builds the whole right-hand side as a temporary tuple before assigning anything.
- **Immutability, precisely defined** — you can never reassign a tuple's *slots* (`t[0] = x` fails), but if a slot holds a *mutable* object like a list, that inner object can still be changed in place. The tuple's identity and its other slots are unaffected.
- **Hashability** — a tuple is hashable (usable as a dict key or set member) only if every element inside it is also hashable; a tuple containing a list is not hashable.
- **Conversion patterns** — `tuple(some_list)` and `list(some_tuple)` convert between mutable and immutable; the classic "modify a tuple" workaround is converting to a list, changing it, and converting back.
- **Functional tools work identically to lists** — `map()`, `filter()`, `zip()`, and `sorted(key=lambda...)` all accept and can produce tuples.
- **`collections.namedtuple`** — a lightweight way to get named, attribute-style field access (`emp.salary`) while keeping all of a tuple's normal behavior and low memory footprint.

Each exercise below gives a problem, a hint, a solution, and an explanation.

## Exercise 1. Basic Tuple Operations

**Concept:** creation, indexing, len()

**Problem:** Create a tuple, access its first and last elements, and find its length.

**Given:**
```
fruits = ("apple", "banana", "cherry", "date")
```

**Expected Output:**
```
First element: apple
Last element: date
Length: 4
```

**Hint:** Index 0 is first; index -1 is always last regardless of length.

````python
fruits = ("apple", "banana", "cherry", "date")

print("First element:", fruits[0])
print("Last element:", fruits[-1])
print("Length:", len(fruits))
````

**Explanation:** Parentheses with comma-separated values create an ordered, immutable tuple. Negative indexing with -1 reliably reaches the last element no matter how many items the tuple holds, and len() counts them all directly.

## Exercise 6. Tuple Reversal

**Concept:** [::-1] slicing (tuples have no .reverse() method)

**Problem:** Reverse the order of a tuple's elements.

**Given:**
```
items = (1, 2, 3, 4, 5)
```

**Expected Output:**
```
(5, 4, 3, 2, 1)
```

**Hint:** [::-1] walks backward through the whole sequence; tuples don't have an in-place .reverse() like lists do.

````python
items = (1, 2, 3, 4, 5)
reversed_items = items[::-1]

print(reversed_items)
````

**Explanation:** Leaving start and stop blank means 'the whole tuple,' and a step of -1 tells Python to traverse it backward. Because slicing always builds a new object, the original items tuple is left unchanged — tuple(reversed(items)) would achieve the same result via an iterator.

## Exercise 12. The Swap Trick

**Concept:** a, b = b, a

**Problem:** Swap two variables' values without a temporary third variable.

**Given:**
```

```

**Expected Output:**
```
After swap: a = 200, b = 100
```

**Hint:** Python evaluates the entire right-hand side into a temporary tuple before assigning anything.

````python
a = 100
b = 200

a, b = b, a

print("After swap: a =", a, ", b =", b)
````

**Explanation:** Python first evaluates b, a as a whole, building the temporary tuple (200, 100), and only then unpacks it into a and b from left to right. Because both original values are captured before either variable is overwritten, no separate temp variable is ever needed.

## Exercise 22. Sort a Tuple of Tuples by a Field

**Concept:** sorted(key=lambda), then tuple() to restore type

**Problem:** Sort a tuple of (name, score) tuples by the score.

**Given:**
```
students = (("Alice", 88), ("Bob", 73), ("Charlie", 95), ("Diana", 61))
```

**Expected Output:**
```
(('Diana', 61), ('Bob', 73), ('Alice', 88), ('Charlie', 95))
```

**Hint:** sorted() always returns a list — wrap the result in tuple() to convert it back.

````python
students = (("Alice", 88), ("Bob", 73), ("Charlie", 95), ("Diana", 61))

sorted_students = tuple(sorted(students, key=lambda x: x[1]))
print("Sorted:", sorted_students)
````

**Explanation:** key=lambda x: x[1] tells sorted() to compare by each nested tuple's second element (the score) instead of comparing whole tuples. Since sorted() always hands back a list regardless of the input type, tuple(...) is needed afterward to restore the original tuple type.

## Exercise 23. Filter a Tuple

**Concept:** filter() vs. a generator expression

**Problem:** Keep only the tuple elements greater than 10, using two different approaches.

**Given:**
```
numbers = (3, 14, 7, 22, 9, 41, 18, 5)
```

**Expected Output:**
```
(14, 22, 41, 18)
```

**Hint:** filter(lambda x: x > 10, numbers) and (x for x in numbers if x > 10) both filter, then wrap either in tuple().

````python
numbers = (3, 14, 7, 22, 9, 41, 18, 5)

filtered_filter = tuple(filter(lambda x: x > 10, numbers))
print("Using filter():", filtered_filter)

filtered_comp = tuple(x for x in numbers if x > 10)
print("Using comprehension:", filtered_comp)
````

**Explanation:** filter() is a lazy iterator, needing tuple() to materialize its results, exactly like the generator expression. Both approaches process numbers left to right, so qualifying elements appear in the output in their original relative order.

## Exercise 24. Map a Transformation Over a Tuple

**Concept:** map() vs. a generator expression

**Problem:** Square every element of a tuple, using two different approaches.

**Given:**
```
numbers = (1, 2, 3, 4, 5, 6)
```

**Expected Output:**
```
(1, 4, 9, 16, 25, 36)
```

**Hint:** map(square, numbers) and (x ** 2 for x in numbers) both transform every element, then wrap either in tuple().

````python
numbers = (1, 2, 3, 4, 5, 6)

def square(n):
    return n * n

squared_map = tuple(map(square, numbers))
print("Using map():", squared_map)

squared_comp = tuple(x ** 2 for x in numbers)
print("Using comprehension:", squared_comp)
````

**Explanation:** map() applies square to each element lazily, one at a time, and like filter() needs tuple() to force full evaluation. The generator expression's x ** 2 achieves the identical transformation inline, without a separate named function.

## Exercise 25. Zip Two Tuples Into a Dictionary

**Concept:** zip() + dict()

**Problem:** Combine a tuple of keys and a tuple of values into a dictionary.

**Given:**
```
keys = ("name", "age", "city"), values = ("Alice", 30, "Pune")
```

**Expected Output:**
```
{'name': 'Alice', 'age': 30, 'city': 'Pune'}
```

**Hint:** zip(keys, values) pairs them by position; dict() converts the pairs directly.

````python
keys = ("name", "age", "city")
values = ("Alice", 30, "Pune")

mapping = dict(zip(keys, values))
print(mapping)

mapping_comp = {k: v for k, v in zip(keys, values)}
print(mapping_comp)
````

## Exercise 26. Tuple Intersection

**Concept:** set conversion for fast overlap, or an order-preserving alternative

**Problem:** Find the elements shared by two tuples.

**Given:**
```
t1 = (1, 2, 3, 4, 5, 6), t2 = (4, 5, 6, 7, 8, 9)
```

**Expected Output:**
```
(4, 5, 6)
```

**Hint:** Convert both to sets and use &, then sort and convert back to a tuple for a consistent order.

````python
t1 = (1, 2, 3, 4, 5, 6)
t2 = (4, 5, 6, 7, 8, 9)

common = tuple(sorted(set(t1) & set(t2)))
print("Common elements:", common)

common_ordered = tuple(x for x in t1 if x in set(t2))
print("Common elements (ordered):", common_ordered)
````

**Explanation:** set(t1) & set(t2) finds the overlap in roughly O(min(n, m)) time, but loses ordering, so sorted() restores a consistent sequence before converting back to a tuple. The generator-expression alternative instead keeps t1's original order by iterating it directly, while still checking membership against a converted set(t2) for speed.

## Exercise 28. Tuple 'Mutability' Nuance

**Concept:** a tuple can't reassign slots, but a slot's mutable contents can change

**Problem:** Show that a list stored inside a tuple can still be modified in place, without changing the tuple's identity.

**Given:**
```
t = (1, 2, [3, 4, 5]), then append 99 to the inner list
```

**Expected Output:**
```
Before: (1, 2, [3, 4, 5])
After:  (1, 2, [3, 4, 5, 99])
Same object? True
```

**Hint:** t[2] retrieves a reference to the list; .append() mutates that list object directly, not the tuple.

````python
t = (1, 2, [3, 4, 5])
print("Before:", t)
print("Tuple id before:", id(t))

t[2].append(99)

print("After: ", t)
print("Tuple id after: ", id(t))
print("Same object?", id(t) == id(t))
````

**Explanation:** t[2] hands back the actual list object stored at that position; .append() changes that list in place, while the tuple's slot still holds the exact same reference it always did. id(t) confirms the tuple itself was never replaced — tuple immutability means slots can't be reassigned (t[2] = something_else would fail), but it says nothing about whether the object a slot points to can itself be mutable.

## Exercise 29. Flatten a Deeply Nested Tuple

**Concept:** recursion with yield from

**Problem:** Flatten a tuple nested to an arbitrary depth into one flat tuple.

**Given:**
```
nested = (1, (2, 3), (4, (5, (6, 7))))
```

**Expected Output:**
```
(1, 2, 3, 4, 5, 6, 7)
```

**Hint:** If an item is a tuple, recurse into it with `yield from`; otherwise yield it directly.

````python
def flatten(t):
    for item in t:
        if isinstance(item, tuple):
            yield from flatten(item)
        else:
            yield item

nested = (1, (2, 3), (4, (5, (6, 7))))
result = tuple(flatten(nested))
print("Flattened:", result)
````

**Explanation:** isinstance(item, tuple) separates the recursive case (dig deeper) from the base case (yield the plain value directly). yield from flatten(item) forwards every value the recursive call produces up through the chain of calls without an explicit inner loop; wrapping the whole generator in tuple() forces it to run to completion and collects every yielded value.

## Exercise 30. Memory Efficiency: List vs. Tuple

**Concept:** sys.getsizeof()

**Problem:** Compare the memory footprint of a list and a tuple holding the same one million integers.

**Given:**
```
range(1_000_000), stored as both a list and a tuple
```

**Expected Output:**
```
The tuple uses noticeably less memory than the list
```

**Hint:** Lists over-allocate extra capacity for future .append() calls; tuples allocate exactly what's needed.

````python
import sys

data = range(1_000_000)
large_list  = list(data)
large_tuple = tuple(data)

list_size  = sys.getsizeof(large_list)
tuple_size = sys.getsizeof(large_tuple)
difference = list_size - tuple_size

print(f"List size:  {list_size:,} bytes")
print(f"Tuple size: {tuple_size:,} bytes")
print(f"Difference: {difference:,} bytes ({difference / 1024:.2f} KB)")
````

**Explanation:** sys.getsizeof() measures only the container itself, not the integers it points to — both collections reference the same integer objects, so that part of the memory cost is identical either way. Lists reserve extra spare capacity to make future .append() calls fast, while a tuple (which can never grow) allocates exactly the space it needs and no more, which is why it consistently comes out smaller.

## Exercise 31. NamedTuples for Readable Records

**Concept:** collections.namedtuple

**Problem:** Define a named tuple with labeled fields, create instances, and find the highest-paid one.

**Given:**
```
Employee records: ("Alice", "Engineering", 95000), ("Bob", "Marketing", 72000), ("Charlie", "Engineering", 88000)
```

**Expected Output:**
```
Highest paid: Alice ($95,000)
```

**Hint:** namedtuple() creates a tuple subclass whose fields can be accessed by name (emp.salary) as well as by index.

````python
from collections import namedtuple

Employee = namedtuple("Employee", ["name", "department", "salary"])

employees = (
    Employee("Alice",   "Engineering", 95000),
    Employee("Bob",     "Marketing",   72000),
    Employee("Charlie", "Engineering", 88000),
)

for emp in employees:
    print(f"{emp.name} works in {emp.department} and earns ${emp.salary:,}")

top = max(employees, key=lambda e: e.salary)
print(f"Highest paid: {top.name} (${top.salary:,})")
````

**Explanation:** namedtuple("Employee", [...]) dynamically builds a new tuple subclass with named fields. emp.salary is self-documenting compared to the equivalent plain-tuple access emp[2], but a named tuple still behaves exactly like a regular tuple for indexing, unpacking, and passing to functions like max() with a key.

## Exercise 32. The Hashability Paradox

**Concept:** a tuple is hashable only if all its elements are

**Problem:** Show that a tuple of only immutable values can be a dictionary key, while a tuple containing a list cannot.

**Given:**
```
(1, 2) as a key (works); (1, [2]) as a key (fails)
```

**Expected Output:**
```
(1, 2) as key: success
(1, [2]) as key: TypeError - unhashable type: 'list'
```

**Hint:** A tuple's hash is built from its elements' hashes — an unhashable element (like a list) makes the whole tuple unhashable.

````python
d = {}

try:
    d[(1, 2)] = "immutable tuple"
    print(f"(1, 2) as key: success, value = \"{d[(1, 2)]}\"")
except TypeError as e:
    print(f"(1, 2) as key: TypeError - {e}")

try:
    d[(1, [2])] = "tuple with list"
    print(f"(1, [2]) as key: success, value = \"{d[(1, [2])]}\"")
except TypeError as e:
    print(f"(1, [2]) as key: TypeError - {e}")

print("\nhash((1, 2))  =", hash((1, 2)))
try:
    print("hash((1, [2])) =", hash((1, [2])))
except TypeError as e:
    print(f"hash((1, [2])): TypeError - {e}")
````

**Explanation:** Dictionary keys must have a hash value that stays constant for their entire lifetime in the dictionary. (1, 2) contains only integers — themselves immutable — so its hash is stable and safe to use as a key. (1, [2]) contains a list, which could be mutated after insertion and would silently change the tuple's hash, breaking the dictionary's internal lookup — so Python refuses to hash any tuple containing a mutable element at all, raising TypeError immediately.
