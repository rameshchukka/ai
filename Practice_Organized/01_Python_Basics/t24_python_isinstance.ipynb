# Python `isinstance()`

`isinstance(object, classinfo)` checks whether `object` is an instance of `classinfo` — a specific type or class — returning `True` or `False`. Both arguments are required. `classinfo` can be a single type/class, OR a tuple of several, in which case `isinstance()` returns `True` if the object matches ANY of them.

Crucially, for a class involved in inheritance, `isinstance()` returns `True` not just for the object's own exact class, but for any of its PARENT classes too — this is the 'is-a' relationship at the heart of object-oriented programming, covered more fully in the OOP and Inheritance notebooks later in this series.

## Basic Usage

```python
num = 90

result = isinstance(num, int)

if result:

    print("Yes")

else:

    print("No")
```

An invalid `classinfo` (not a class, type, or tuple of types) raises `TypeError`.

```python
try:

    isinstance(90, "not a type")

except TypeError as e:

    print("TypeError:", e)
```

## `isinstance()` With Built-in Types

```python
number = 80

print(isinstance(number, int))

print(isinstance(number, float))



pi = 3.14

print(isinstance(pi, float))



complex_num = 1 + 2j

print(isinstance(complex_num, complex))



name = "PYnative.com"

print(isinstance(name, str))



names = ["Eric", "Scott", "Kelly"]

print(isinstance(names, list))



student_report = {"John": 80, "Eric": 70, "Donald": 90}

print(isinstance(student_report, dict))



names_tuple = ("Sam", "Kelly", 'Emma')

print(isinstance(names_tuple, tuple))



numbers = {11, 22, 33, 44, 55}

print(isinstance(numbers, set))
```

### `None` and `isinstance()`

`None` has its own distinct type (`NoneType`) — it's never considered an instance of any regular built-in type.

```python
var = None

s1 = ''   # empty string - NOT the same thing as None



print(isinstance(var, float))

print(isinstance(s1, str))

print(isinstance(var, type(None)))   # the correct way to isinstance-check for None
```

## `isinstance()` With Multiple Types

Pass a TUPLE of types to check against several possibilities in one call — `True` if the object matches ANY of them.

```python
def check_number(var):

    if isinstance(var, (int, float)):

        print('variable', var, 'is instance of numeric type')

    else:

        print('variable', var, 'is not instance of numeric type')



check_number(80)

check_number(55.70)

check_number('20')
```

## `isinstance()` With a User-Defined Class

Works exactly the same way for classes you write yourself — checking whether an object is genuinely built from a specific class.

```python
class Employee:

    def __init__(self, name, salary):

        self.name = name

        self.salary = salary



class Person:

    def __init__(self, name, sex):

        self.name = name

        self.sex = sex



emp = Employee("Emma", 11000)

per = Person("Brent", "male")



print(isinstance(emp, Employee))

print(isinstance(per, Employee))
```

## `isinstance()` and Inheritance

An object of a SUBCLASS is also considered an instance of its PARENT class — this is the 'is-a' relationship. If `PythonDeveloper` is a subclass of `Developer`, then a `PythonDeveloper` object genuinely 'is a' `Developer` too, and `isinstance()` reflects that.

```python
class Developer:

    def __init__(self, name):

        self.name = name



    def display(self):

        print("Developer:", self.name)



class PythonDeveloper(Developer):

    def __init__(self, name, language):

        self.name = name

        self.language = language



    def display(self):

        print("Python Developer:", self.name, "language:", self.language)



dev = PythonDeveloper("Eric", "Python")



print(isinstance(dev, PythonDeveloper))   # its own exact class

print(isinstance(dev, Developer))         # ALSO True - the parent class
```

This is genuinely useful for writing code that works correctly regardless of exactly WHICH subclass an object is, as long as it 'is-a' the expected parent type — a common pattern when a function is meant to accept any object from a whole family of related classes.

## `isinstance()` With Lists

### Checking the List Itself

```python
sample_list = ["Emma", "Stevan", "Brent"]

res = isinstance(sample_list, list)

print(sample_list, 'is instance of list?', res)
```

### Detecting a Nested List Among a List's Elements

```python
sample_list = ['Emma', 'Stevan', ['Jordan', 'Donald', 'Sam']]

for item in sample_list:

    if isinstance(item, list):

        print("Yes", item, 'is a nested list')
```

### Sorting a List's Elements by Type

```python
sample_list = ['Emma', 'Stevan', 12, 45.6, 1 + 2j, "Eric"]

number_list = []

string_list = []

for item in sample_list:

    if isinstance(item, (int, float, complex)):

        number_list.append(item)

    elif isinstance(item, str):

        string_list.append(item)



print("Strings:", string_list)

print("Numbers:", number_list)
```

## `isinstance()` vs. `type() ==` — Why `isinstance()` Is Usually Better

A very common alternative is `type(x) == SomeClass`, but this DOESN'T account for inheritance — it only matches the EXACT class, never a subclass. `isinstance()` is almost always the better choice, since it correctly handles the 'is-a' relationship.

```python
dev = PythonDeveloper("Eric", "Python")



print("type(dev) == Developer:", type(dev) == Developer)          # False! exact-type check misses inheritance

print("isinstance(dev, Developer):", isinstance(dev, Developer))  # True - correctly recognizes the is-a relationship
```

## Summary

- `isinstance(object, classinfo)` returns `True`/`False`; `classinfo` can be one type or a tuple of several.
- Works identically for built-in types and user-defined classes.
- For inherited classes, an object is `isinstance()` of BOTH its own class AND every parent class — the 'is-a' relationship.
- Prefer `isinstance()` over `type(x) == SomeClass` almost always, since the latter breaks for subclasses.
- A common real use: sorting a heterogeneous list's elements by type, or writing a function that accepts anything from a family of related classes.

## Practice Exercises

1. Write a function `describe(x)` that prints `"boolean"`, `"integer"`, `"float"`, `"string"`, or `"other"` for a given value — remember `bool` is a SUBCLASS of `int` in Python, so check for `bool` FIRST or it will be misreported as `"integer"`.
2. Given a mixed list `[1, "two", 3.0, [4], (5,), {6}, {"seven": 7}]`, count how many elements are instances of EACH of: `list`, `tuple`, `set`, `dict`.
3. Create a class hierarchy `Animal` -> `Dog` -> `Puppy` (each inheriting from the previous), create a `Puppy` instance, and show it's an instance of all three classes.
4. Write a function `safe_add(a, b)` that only performs `a + b` if BOTH are instances of `(int, float)`, otherwise returns `"Cannot add: not both numeric"`.
5. Using `isinstance()`, write a function `flatten_if_nested(lst)` that, given a list, returns a single flat list — expanding any element that is itself a list, but leaving non-list elements as-is (only one level deep).

### Exercise 1 solution

```python
def describe(x):

    if isinstance(x, bool):     # MUST check before int - bool is an int subclass

        return "boolean"

    elif isinstance(x, int):

        return "integer"

    elif isinstance(x, float):

        return "float"

    elif isinstance(x, str):

        return "string"

    else:

        return "other"



for value in [True, 42, 3.14, "hi", [1, 2]]:

    print(value, "->", describe(value))
```

### Exercise 2 solution

```python
items = [1, "two", 3.0, [4], (5,), {6}, {"seven": 7}]



counts = {"list": 0, "tuple": 0, "set": 0, "dict": 0}

for item in items:

    if isinstance(item, list):

        counts["list"] += 1

    elif isinstance(item, tuple):

        counts["tuple"] += 1

    elif isinstance(item, set):

        counts["set"] += 1

    elif isinstance(item, dict):

        counts["dict"] += 1



print(counts)
```

### Exercise 3 solution

```python
class Animal:

    pass



class Dog(Animal):

    pass



class Puppy(Dog):

    pass



p = Puppy()

print(isinstance(p, Puppy))

print(isinstance(p, Dog))

print(isinstance(p, Animal))
```

### Exercise 4 solution

```python
def safe_add(a, b):

    if isinstance(a, (int, float)) and isinstance(b, (int, float)):

        return a + b

    return "Cannot add: not both numeric"



print(safe_add(3, 4.5))

print(safe_add(3, "4.5"))
```

### Exercise 5 solution

```python
def flatten_if_nested(lst):

    result = []

    for item in lst:

        if isinstance(item, list):

            result.extend(item)

        else:

            result.append(item)

    return result



print(flatten_if_nested([1, [2, 3], 4, [5, 6, 7]]))
```
