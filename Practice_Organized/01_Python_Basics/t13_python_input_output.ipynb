# Python Input and Output

Python has two core built-ins for interacting with a user: `input(prompt)` to READ a line of text, and `print()` to WRITE output. This notebook covers both directions in depth — reading strings/numbers/multiple values/multiline input, command-line arguments via `sys.argv`, and the full range of `str.format()`-based output formatting.

**Note on this notebook:** every cell that would normally call the real, interactive `input()` instead simulates it with a small helper that feeds in pre-set values and echoes them — so every cell runs unattended, top to bottom, exactly as it would with a real user typing those same answers.

## The `input()` Function

`input([prompt])` displays an optional prompt message, waits until the user types something and presses Enter, then returns whatever was typed — ALWAYS as a plain `str`, even if it looks like a number.

```python
# Helper used throughout this notebook to simulate input() non-interactively

import builtins

def simulate_input(*responses):

    it = iter(responses)

    def fake_input(prompt=""):

        val = next(it)

        print(f"{prompt}{val}")

        return val

    builtins.input = fake_input



def restore_input():

    builtins.input = builtins.__dict__.get('_real_input', input)



import builtins as _b

_b._real_input = _b.input
```

```python
simulate_input("Jessa", "8000", "Google")



name = input("Enter Employee Name: ")

salary = input("Enter salary: ")

company = input("Enter Company name: ")



print()

print("Printing Employee Details")

print("Name", "Salary", "Company")

print(name, salary, company)
```

### `input()` Always Returns a String

Even if the user types something that LOOKS numeric, `input()` hands it back as a `str` — no automatic conversion happens.

```python
simulate_input("22", "Jessa")



number = input("Enter roll number ")

name = input("Enter age ")



print()

print('Roll number:', number, 'Name:', name)

print("Printing type of a input values")

print("type of number", type(number))

print("type of name", type(name))
```

## Reading Numbers: Explicit Type Conversion

To get an actual number, wrap `input()` with `int()` or `float()`.

```python
simulate_input("28", "12")



first_number = int(input("Enter first number "))

second_number = int(input("Enter second number "))



print()

print("First Number:", first_number)

print("Second Number:", second_number)

sum1 = first_number + second_number

print("Addition of two number is: ", sum1)
```

```python
simulate_input("74.65")



marks = float(input("Enter marks "))

print()

print("Student marks is: ", marks)

print("type is:", type(marks))
```

## Multiple Inputs on One Line

`.split()` lets one `input()` call collect several space-separated values at once, unpacked directly into separate variables.

```python
simulate_input("Jessa 18 72.50")



name, age, marks = input("Enter your Name, Age, Percentage separated by space ").split()

print()

print("User Details: ", name, age, marks)
```

## Multiline Input

`input()` only ever reads ONE line — pressing Enter ends that call immediately. To collect several lines, loop, calling `input()` repeatedly and collecting the results (here, stopping on the first blank line, mimicking 'press Enter twice to finish').

```python
simulate_input("My Name is Jessa", "I am a software engineer", "")



data = []

print("Tell me about yourself")

while True:

    line = input()

    if line:

        data.append(line)

    else:

        break

finalText = '\n'.join(data)

print()

print("Final text input")

print(finalText)
```

## `input()` vs. the Old Python 2 `raw_input()`

This distinction only matters if you ever encounter legacy Python 2 code:
- Python 2 had BOTH `input()` (auto-converted to the 'appropriate' type) and `raw_input()` (always returned a string).
- Python 3 removed the old `input()` entirely and renamed `raw_input()` to `input()` — so Python 3's `input()` behaves like Python 2's `raw_input()`: always a string.
- To reproduce Python 2's old auto-converting `input()` behavior in Python 3, you *could* wrap it as `eval(input(...))` — but this is a security risk, since `eval()` will execute ANY Python expression the user types, not just a number. Prefer explicit `int()`/`float()` conversion instead, always.

## Command-Line Input via `sys.argv`

Outside of interactive `input()`, a program can also receive input as **command-line arguments** — values typed after the script's name when launching it from a terminal, e.g. `python sample.py 20 30 40`.

`sys.argv` is a plain list: `sys.argv[0]` is always the script's own name/path; `sys.argv[1]` onward are the actual arguments, space-separated, and ALWAYS strings (just like `input()`, they need explicit conversion if you want numbers).

```python
import sys



# Simulating a terminal invocation: python sample.py 20 30 40

sys.argv = ["sample.py", "20", "30", "40"]



print("Total argument passed :", len(sys.argv))

print("Type of argv:", type(sys.argv))
```

```python
print("All command line inputs")

for value in sys.argv:

    print(value)
```

```python
# Command-line args need explicit type conversion, just like input()

sys.argv = ["sample.py", "20", "30"]



add = int(sys.argv[1]) + int(sys.argv[2])

print('Argument one:', sys.argv[1])

print('Argument Two:', sys.argv[2])

print('Addition is:', add)
```

```python
# Accessing an argument index that wasn't actually provided raises IndexError

sys.argv = ["sample.py", "20"]

try:

    print(sys.argv[2])

except IndexError as e:

    print("IndexError:", e)
```

## Output with `print()`

`print()` writes to standard output (normally the screen/console). Multiple arguments are joined with a space by default; `sep=` overrides that separator.

```python
simulate_input("Jessa")

name = input("Enter Name: ")

print('User Name:', name)

restore_input()
```

```python
simulate_input("Jessa", "412365", "abc street", "234")



name = input('Enter Name ')

zip_code = int(input('Enter zip code '))

street = input('Enter street name ')

house_number = int(input('Enter house number '))



print(name, zip_code, street, house_number, sep="-")

restore_input()
```

## Output Formatting With `str.format()`

`"template {0} {1}".format(a, b)` fills `{0}`, `{1}`, ... placeholders with the corresponding positional arguments. Text outside `{}` is copied through literally.

```python
print('FirstName - {0}, LastName - {1}'.format('Ault', 'Kelly'))
```

### Formatting by Position

Positional indices can be REUSED and REORDERED freely within the same template.

```python
simulate_input("Ault", "Kelly", "Google")



firstName = input("Enter First Name ")

lastName = input("Enter Last Name ")

organization = input("Enter Organization Name ")



print()

print('{0}, {1} works at {2}'.format(firstName, lastName, organization))

print('{1}, {0} works at {2}'.format(firstName, lastName, organization))

print('FirstName {0}, LastName {1} works at {2}'.format(firstName, lastName, organization))

print('{0}, {1} {0}, {1} works at {2}'.format(firstName, lastName, organization))

restore_input()
```

### Formatting by Name

Keyword arguments to `.format()` can be referenced by NAME in the template instead of by position — often more readable for templates with several fields.

```python
simulate_input("Jhon", "74")



name = input("Enter Name ")

marks = input("Enter marks ")



print()

print('Student: Name:  {firstName}, Marks: {percentage}%'.format(firstName=name, percentage=marks))

restore_input()
```

### Alignment: Width, Left/Right/Center

`{:<N}` left-aligns, `{:>N}` right-aligns, `{:^N}` centers — all within a field N characters wide.

```python
simulate_input("This is a sample text")

text = input("Enter text ")

print()

print('[{:<25}]'.format(text))   # left aligned

print('[{:>25}]'.format(text))   # right aligned

print('[{:^25}]'.format(text))   # centered

restore_input()
```

### Forcing a Sign on Numbers

`{:+f}` always shows a sign (`+` for positive, `-` for negative); the default `{:f}` only shows `-` for negatives, nothing for positives.

```python
simulate_input("25.25", "-15.50")



positive_number = float(input("Enter Positive Number "))

negative_number = float(input("Enter Negative Number "))



print()

print('{:+f}; {:+f}'.format(positive_number, negative_number))

print('{:f}; {:f}'.format(positive_number, negative_number))

restore_input()
```

### Displaying Integers in Different Number Bases

`d` decimal, `o` octal, `b` binary, `x`/`X` lowercase/uppercase hexadecimal.

```python
simulate_input("356")

number = int(input("Enter number "))

print()

print("The number is:{:d}".format(number))

print('Output number in octal format : {0:o}'.format(number))

print('Output number in binary format: {0:b}'.format(number))

print('Output number in hexadecimal format: {0:x}'.format(number))

print('Output number in HEXADECIMAL: {0:X}'.format(number))

restore_input()
```

### Formatting Floats

`f` fixed-point, with an optional `WIDTH.PRECISION` (e.g. `5.2f`); `e`/`E` scientific notation.

```python
simulate_input("234.567")

number = float(input("Enter float Number "))

print()

print("Output Number in The float type :{:f}".format(number))

print('padding for output float number {:8.2f}'.format(number))

print('Output Exponent notation {:e}'.format(number))

print('Output Exponent notation {:E}'.format(number))

restore_input()
```

### String Alignment via `.ljust()` / `.rjust()` / `.center()`

An older-style alternative to `.format()`'s alignment specifiers, called directly as string methods, with a custom fill character.

```python
simulate_input("Jessa")

text = input("Enter String ")

print()

print("Left justification  :", text.ljust(30, "*"))

print("Right justification :", text.rjust(30, "*"))

print("Center justification:", text.center(30, "*"))

restore_input()
```

## Summary

- `input([prompt])` reads one line and ALWAYS returns a `str` — wrap with `int()`/`float()` to get a number.
- `.split()` on the result of `input()` collects several space-separated values in one call.
- Reading multiple lines requires a loop — `input()` itself only ever reads a single line.
- `sys.argv` gives command-line arguments as a list of strings; `sys.argv[0]` is always the script name.
- `print()` joins its arguments with a space by default (`sep=` to change it).
- `str.format()` fills `{0}`, `{1}`, ... or `{name}` placeholders; format specs after a `:` control alignment (`<`/`>`/`^`), sign (`+`), width, precision, and number base (`d`/`o`/`b`/`x`/`X`/`f`/`e`).
- `.ljust()`/`.rjust()`/`.center()` are simpler string-method alternatives to `.format()`'s alignment options.
