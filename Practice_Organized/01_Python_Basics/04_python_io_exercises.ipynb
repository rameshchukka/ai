## Concepts you'll need

This set is about the two ends of every program: getting data in, and presenting data out.

- **`input()`** always returns a string — wrap it in `int()` or `float()` before doing math with it. `.split()` breaks one input line into multiple values at once.
- **f-strings** (`f"{value}"`) are the modern way to build formatted output. Inside the braces you can add a **format spec** after a colon:
  - `:.2f` — fixed to 2 decimal places
  - `:,` — thousands separator
  - `:>20`, `:<10`, `:^40` — right/left/center align within a field width, optionally with a custom fill character (`:-^40`)
  - `:05` — zero-pad to a total width of 5
  - `:b`, `:x`, `:X`, `:o` — binary, lowercase hex, uppercase hex, octal
- **Older formatting styles** — `%o`, `%.2f` (percent-style) and `"{0} {1}".format(...)` (the `.format()` method) still appear in a lot of real code, so a couple of exercises use them directly.
- **`print()`'s `sep` argument** controls what's inserted between multiple printed items (default is a space).
- **File I/O** — `open(path, mode)` with `"r"` (read), `"w"` (write/overwrite), or `"a"` (append); always prefer `with open(...) as f:` so the file closes automatically. `readlines()` loads every line into a list; looping `for line in f:` reads one line at a time, which is more memory-efficient for large files.
- **`os` module** — `os.path.exists()`, `os.stat().st_size`, and `os.remove()` for checking and manipulating files at the filesystem level.
- **`getpass`** — like `input()`, but hides what's typed, for sensitive fields like passwords.

## Exercise 1. Multiply Two User Numbers

**Concept:** input(), type casting with int()

**Problem:** Accept two integers from the user and print their product.

**Given:**
```
10, 20
```

**Expected Output:**
```
The multiplication is: 200
```

**Hint:** input() always returns a string — wrap it in int() before multiplying.

````python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

res = num1 * num2

print("The multiplication is:", res)
````

**Explanation:** int(input(...)) converts the typed text into a whole number in one step, so num1 * num2 can perform real multiplication instead of accidentally repeating a string.

## Exercise 4. Decimal to Binary

**Concept:** the :b format spec in an f-string

**Problem:** Display an integer in binary (base 2).

**Given:**
```
45
```

**Expected Output:**
```
The binary representation of 45 is 101101
```

**Hint:** f'{num:b}' converts an integer to binary without the '0b' prefix that bin() adds.

````python
num = int(input("Enter a decimal number: "))

binary_val = f"{num:b}"

print(f"The binary representation of {num} is {binary_val}")
````

## Exercise 5. Split One Input Line Into Three Variables

**Concept:** str.split() with unpacking

**Problem:** Accept three names in a single input() call and assign each to its own variable.

**Given:**
```
"Emma Jessa Kelly"
```

**Expected Output:**
```
Name1: Emma
Name2: Jessa
Name3: Kelly
```

**Hint:** input().split() breaks the line into a list wherever there's whitespace; unpack that list into three variables at once.

**Explanation:** .split() breaks a string into a list at each space by default. Iterable unpacking (str1, str2, str3 = ...) then distributes that three-item list into three separate variables, in order.

## Exercise 8. Percentage With 2 Decimal Places

**Concept:** combining a calculation with :.2f

**Problem:** Calculate a percentage from a numerator and denominator, showing exactly two decimal places and a percent sign.

**Given:**
```
numerator 22, denominator 29
```

**Expected Output:**
```
The result is: 75.86%
```

**Hint:** Multiply the ratio by 100 first, then format with :.2f and append a literal % after the closing brace.

````python
num = int(input("Enter numerator: "))
den = int(input("Enter denominator: "))

percentage = (num / den) * 100

print(f"The result is: {percentage:.2f}%")
````

**Explanation:** percentage:.2f formats the computed ratio-times-100 value to two decimal places. Since % isn't a special character inside an f-string (unlike in %-style formatting), it can just be typed directly after the closing brace.

## Exercise 9. Right-Align Text in a Fixed Width

**Concept:** the :>N alignment spec

**Problem:** Print a word right-aligned within a 20-character field, followed by a version number.

**Given:**
```
"Python", "3.14"
```

**Expected Output:**
```
              Python 3.14
```

**Hint:** f'{word:>20}' reserves 20 characters and pushes the text to the right edge.

````python
word = input("Enter a word: ")
version = input("Enter a version number: ")

print(f"{word:>20} {version}")
````

**Explanation:** :>20 reserves a 20-character-wide field and pads it with spaces on the left, pushing the actual text to the right edge of that field.

## Exercise 10. Center Text With a Fill Character

**Concept:** the :{fill}^{width} spec

**Problem:** Center a title within a 40-character-wide field, padded with hyphens.

**Given:**
```
"REPORT SUMMARY"
```

**Expected Output:**
```
-------------REPORT SUMMARY--------------
```

**Hint:** The syntax is :{fill_char}{align_char}{width}, e.g. :-^40.

````python
title = "REPORT SUMMARY"

formatted_title = f"{title:-^40}"

print(formatted_title)
````

**Explanation:** - is the fill character, ^ means center-align, and 40 is the total field width. Python works out how many hyphens belong on each side so the whole result is exactly 40 characters long.

## Exercise 11. Zero-Pad a Number

**Problem:** Pad a number with leading zeros so it's always 5 digits wide.

**Given:**
```
42
```

**Expected Output:**
```

```

**Hint:** The zfill(width) string method adds leading zeros until the string reaches the target length.

## Exercise 12. Format With .format() and Positional Indices

**Concept:** str.format() with {0}, {1}, {2}

**Problem:** Use .format() to build a sentence from three variables, formatting one of them as a 2-decimal float.

**Given:**
```
quantity=3, totalMoney=450, price=150
```

**Expected Output:**
```
I have 450 dollars so I can buy 3 football for 150.00 dollars.
```

**Hint:** {0}, {1}, {2} refer to the .format() arguments by position, in whatever order you place them in the string.

````python
quantity = 3
totalMoney = 450
price = 150
statement = "I have {1} dollars so I can buy {0} football for {2:.2f} dollars."

print(statement.format(quantity, totalMoney, price))
````

**Explanation:** The numbers inside {0}, {1}, {2} refer to the position of each argument passed to .format(), letting the template reorder them freely. {2:.2f} additionally formats that particular argument as a float with two decimal places.

## Exercise 13. Currency Formatting With Commas

**Concept:** the :,.2f spec

**Problem:** Display a large number as currency: dollar sign, comma thousands separators, two decimal places.

**Given:**
```

```

**Expected Output:**
```
Total Balance: $1,250,500.70
```

**Hint:** Combine a comma and .2f in one format spec: :,.2f.

````python
balance = 1250500.7

formatted_balance = f"${balance:,.2f}"

print(f"Total Balance: {formatted_balance}")
````

**Explanation:** The comma in the format spec inserts a thousands separator every three digits, and .2f guarantees exactly two decimal places. The dollar sign is just a literal character placed outside the braces.

## Exercise 14. Collect 5 Floats Into a List

**Concept:** looping input() with float() and append()

**Problem:** Accept 5 float numbers from the user, one at a time, and store them in a list.

**Given:**
```
78.6, 78.6, 85.3, 1.2, 3.5
```

**Expected Output:**
```
[78.6, 78.6, 85.3, 1.2, 3.5]
```

**Hint:** Loop 5 times, converting each input() call to float() and appending it to a growing list.

````python
numbers = []

for i in range(0, 5):
    print("Enter number at location", i, ":")
    item = float(input())
    numbers.append(item)

print("User List:", numbers)
````

**Explanation:** range(0, 5) guarantees exactly five prompts. Each entered value is converted with float() (so decimals are allowed) and appended to the numbers list, growing it one item per loop pass.

## Exercise 15. Tabular Output From Two Parallel Lists

**Concept:** zip() with left-aligned columns

**Problem:** Print two parallel lists (names and scores) as a neatly aligned table.

**Given:**
```
names = ["Alice", "Bob", "Charlie"], scores = [85, 92, 78]
```

**Expected Output:**
```
Name       Score
----------------
Alice      85
Bob        92
Charlie    78
```

**Hint:** zip() pairs the two lists; a left-align width spec like :<10 keeps the Score column starting at the same position.

````python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

print(f"{'Name':<10} {'Score'}")
print("-" * 15)

for name, score in zip(names, scores):
    print(f"{name:<10} {score}")
````

**Explanation:** zip(names, scores) pairs each name with its corresponding score by position. :<10 reserves a 10-character-wide, left-aligned field for the name, so every row's Score column lines up at the same horizontal position regardless of name length.

## Exercise 16. Interactive Menu

**Concept:** if/elif/else mapped to user choices

**Problem:** Build a 3-option menu (say hello / calculate square / exit) that acts on the user's numeric choice.

**Given:**
```
choice 2, number 5
```

**Expected Output:**
```
The square is: 25
```

**Hint:** Compare choice as a string ('1', '2', '3') and always include an else for invalid input.

````python
print("1. Say Hello\n2. Calculate Square\n3. Exit")
choice = input("Enter choice (1-3): ")

if choice == "1":
    print("Hello there! Hope you're having a great day.")
elif choice == "2":
    val = int(input("Enter number to square: "))
    print(f"The square is: {val * val}")
elif choice == "3":
    print("Exiting... Goodbye!")
else:
    print("Invalid choice. Please pick 1, 2, or 3.")
````

**Explanation:** Since input() returns a string, comparing choice == "1" works directly without converting to int first. The elif chain guarantees only one branch runs, and the trailing else catches anything that doesn't match 1, 2, or 3.

## Exercise 17. Masked Password Input

**Concept:** the getpass module

**Problem:** Ask for a username normally, but a password with the typed characters hidden.

**Given:**
```
username "admin", password "SecretPassword123"
```

**Expected Output:**
```
Login successful for admin!
```

**Hint:** import getpass, then use getpass.getpass() in place of input() for the password field.

````python
import getpass

user = input("Enter Username: ")
pwd = getpass.getpass("Password: ")

if user == "admin" and pwd == "SecretPassword123":
    print(f"Login successful for {user}!")
else:
    print("Access Denied.")
````

**Explanation:** getpass.getpass() suppresses terminal echo while the user types, so the password never appears on screen or in scrollback — unlike input(), which shows every keystroke. The typed value is still captured normally in the pwd variable for comparison.

## Exercise 18. Read a File Into a List of Lines

**Concept:** file.readlines()

**Problem:** Read an existing text file and store each line as one item in a list.

**Given:**
```
test.txt with several lines
```

**Expected Output:**
```
['Line 1\n', 'Line 2\n', ...]
```

**Hint:** readlines() reads the whole file and splits it into a list at each newline character.

````python
with open("test.txt", "r") as f:
    lines = f.readlines()
    print(lines)
````

**Explanation:** readlines() returns one list element per line, including the trailing \n on each. To get clean text without those newline characters, you'd typically follow up with a comprehension using .strip().

## Exercise 19. Write a List of Strings to a File

**Concept:** looping .write() with '\n'

**Problem:** Save each item of a list of fruits onto its own line in a text file.

**Given:**
```
["Apple", "Banana", "Cherry", "Date"]
```

**Expected Output:**
```
fruits.txt created with one fruit per line
```

**Hint:** Open in 'w' mode, then loop the list writing each item followed by a newline character.

````python
fruit_list = ["Apple", "Banana", "Cherry", "Date"]

with open("fruits.txt", "w") as f:
    for fruit in fruit_list:
        f.write(fruit + "\n")

print("File 'fruits.txt' has been created successfully.")
````

**Explanation:** "w" mode creates the file (overwriting it if it already exists). Since .write() doesn't add line breaks automatically, "\n" is appended manually after every fruit so each one lands on its own line.

## Exercise 20. Count the Lines in a File

**Concept:** memory-efficient line iteration

**Problem:** Count how many lines a text file contains.

**Given:**
```
sample.txt with 12 lines
```

**Expected Output:**
```
The file contains 12 lines.
```

**Hint:** for line in f: reads one line at a time rather than loading the whole file, and a counter tracks how many passes the loop makes.

````python
line_count = 0

with open("sample.txt", "r") as f:
    for line in f:
        line_count += 1

print(f"The file contains {line_count} lines.")
````

**Explanation:** for line in f reads the file lazily, one line at a time, which keeps memory usage low even for very large files (unlike readlines(), which loads everything at once). The counter increments once per line encountered.

## Exercise 22. Check if a File Is Empty

**Concept:** os.stat().st_size

**Problem:** Check a file's size in bytes and report whether it's empty.

**Given:**
```
an empty file empty_log.txt
```

**Expected Output:**
```
Status: empty_log.txt is empty.
```

**Hint:** os.stat(filename).st_size returns the file size in bytes; 0 means empty.

````python
import os

filename = "empty_log.txt"

if os.path.exists(filename):
    file_info = os.stat(filename)
    if file_info.st_size == 0:
        print(f"Status: {filename} is empty.")
    else:
        print(f"Status: {filename} size is {file_info.st_size} bytes.")
else:
    print("File not found.")
````

**Explanation:** os.stat() asks the operating system for metadata about the file — including .st_size, its size in bytes — without needing to open and read its contents. Checking os.path.exists() first avoids a crash if the file doesn't exist at all.

## Exercise 23. Delete a File Safely

**Concept:** os.remove() with a confirmation prompt

**Problem:** Ask the user for a filename, confirm the deletion, then permanently remove it.

**Given:**
```
"old_data.csv", confirmed with "y"
```

**Expected Output:**
```
File 'old_data.csv' has been deleted.
```

**Hint:** Check os.path.exists() first, ask for confirmation, and only then call os.remove().

````python
import os

target = input("Enter filename to delete: ")

if os.path.exists(target):
    confirm = input(f"Are you sure you want to delete '{target}'? (y/n): ")
    if confirm.lower() == 'y':
        os.remove(target)
        print(f"File '{target}' has been deleted.")
    else:
        print("Operation cancelled.")
else:
    print("Error: File does not exist.")
````

**Explanation:** os.remove() permanently unlinks the file from the filesystem — there's no recycle bin involved. Checking os.path.exists() first avoids a FileNotFoundError, and requiring explicit 'y' confirmation guards against accidental deletion.
