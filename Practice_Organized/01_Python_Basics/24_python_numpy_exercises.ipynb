## Concepts you'll need

This set covers **NumPy**, the core array library for numerical computing in Python.

- **Creating arrays** — `np.array(list)`, `np.arange(n)` (like `range()` but returns an array), `np.zeros()`/`np.ones()`/`np.full()`, `np.eye(n)` (identity matrix), `np.linspace(start, stop, num)` (evenly spaced values, *inclusive* of both ends), `np.random.rand()`/`randint()`.
- **Array attributes** — `.shape` (dimensions as a tuple), `.ndim` (number of dimensions), `.dtype` (element type), `.itemsize` (bytes per element), `.nbytes` (total memory).
- **Indexing & slicing** — works like Python lists but extends to multiple dimensions with a comma: `arr[row, col]`, `arr[:, -1]` (last column), `arr[1::2, ::2]` (every-other rows/columns). Slicing returns a *view* (shares memory), not a copy.
- **Boolean masking** — `arr[arr > 5]` filters using a condition; combine conditions with `&` (and) / `|` (or), each condition in its own parentheses; assignment through a mask (`arr[mask] = value`) modifies in place.
- **Reshaping & combining** — `.reshape(rows, cols)`, `.flatten()` (copy) / `.ravel()` (view), `np.hstack()`/`vstack()` (combine arrays), `np.split()` (divide into equal parts).
- **Element-wise vs. matrix operations** — `+`, `-`, `*`, `/`, `**` all operate element-by-element automatically ("vectorization" — no loop needed); `np.dot()` / `@` is real matrix multiplication, a fundamentally different operation from `*`.
- **Set-like operations** — `np.unique()`, `np.intersect1d()`, `np.union1d()`, `np.in1d()` (membership test) treat arrays like mathematical sets.
- **Aggregations** — `np.mean()`, `np.median()`, `np.std()`, `np.min()`/`max()`, often combined with `axis=` to aggregate along rows (`axis=1`) or columns (`axis=0`) instead of the whole array.
- **Linear algebra** — `np.linalg.inv()` (matrix inverse), `np.linalg.eig()` (eigenvalues/eigenvectors), `np.linalg.solve(A, b)` (solves `Ax = b` directly, better than computing an inverse manually).

## Exercise 2. Convert 1D Array to 2D

**Concept:** .reshape()

**Problem:** Convert a 1D array of 6 elements into a 2D array with 2 rows.

**Given:**
```

```

**Expected Output:**
```
Original: [0 1 2 3 4 5]
Reshaped 2x3:
 [[0 1 2]
  [3 4 5]]
```

**Hint:** reshape(rows, cols) needs rows * cols to exactly equal the original element count.

````python
import numpy as np

arr = np.arange(6)
print("Original Array:", arr)

arr_2d = arr.reshape(2, 3)
print("Reshaped 2x3 Array:\n", arr_2d)
````

## Exercise 13. Extract Odd Rows and Even Columns

**Concept:** combined row/column step-slicing

**Problem:** Extract the elements at odd-indexed rows and even-indexed columns.

**Given:**
```
a 5x4 array of multiples of 3
```

**Expected Output:**
```
[[15 21]
 [39 45]]
```

**Hint:** sampleArray[1::2, ::2] combines a row slice and a column slice in one indexing operation.

````python
import numpy as np

sampleArray = np.array([
    [3, 6, 9, 12],
    [15, 18, 21, 24],
    [27, 30, 33, 36],
    [39, 42, 45, 48],
    [51, 54, 57, 60]
])

newArray = sampleArray[1::2, ::2]

print("Original Array:")
print(sampleArray)
print("\nNew Array with Odd Rows and Even Columns:")
print(newArray)
````

Combining both slices with a comma, sampleArray[1::2, ::2], selects exactly the intersection: the values sitting at an odd row AND an even column simultaneously.

## Exercise 20. Matrix Multiplication (Dot Product)

**Concept:** np.dot() / @ — distinct from element-wise *

**Problem:** Compute the dot product of two 1D arrays.

**Given:**
```
a = [1,2,3], b = [4,5,6]
```

```
Dot Product: 32
```

````python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot_product = np.dot(a, b)
print("Dot Product:", dot_product)
print("Using @ operator:", a @ b)
````

## Exercise 22. Remove Common Items from an Array

**Concept:** np.in1d() (membership test) combined with the ~ NOT operator

**Problem:** Remove from array a every element that also appears in array b.

**Given:**
```
a = [1,2,3,4,5], b = [5,6,7,8,9]
```

**Expected Output:**
```
[1 2 3 4]
```

**Hint:** ~ is the bitwise NOT operator — it flips every True to False and vice versa in a boolean mask.

````python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([5, 6, 7, 8, 9])

mask = np.isin(a, b)
result = a[~mask]
print(result)
````

**Explanation:** np.isin(a, b) (the modern name for the same functionality as the older np.in1d()) checks each element of a for membership in b, returning a boolean array — True at position i means a[i] is also somewhere in b. ~mask inverts every value in that boolean array (True becomes False and vice versa) via the bitwise NOT operator, so a[~mask] selects exactly the elements of a that are NOT found in b.

## Exercise 23. Normalize an Array to [0, 1]

**Concept:** min-max normalization

**Problem:** Rescale an array's values so the minimum becomes 0 and the maximum becomes 1.

**Given:**
```
arr = [10, 20, 30, 40, 50]
```

**Expected Output:**
```
Normalized Array: [0.   0.25 0.5  0.75 1.  ]
```

**Hint:** Subtracting the min shifts the smallest value to 0; dividing by the range scales the largest value to 1.

````python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
normalized = (arr - arr.min()) / (arr.max() - arr.min())
print("Normalized Array:", normalized)
````

**Explanation:** Subtracting arr.min() from every element shifts the whole array so its smallest value becomes exactly 0. Dividing by the range (arr.max() - arr.min()) then rescales everything so the largest value lands at exactly 1 — every other value falls proportionally in between. Both operations broadcast automatically across the whole array with no explicit loop.

## Exercise 50. Descriptive Statistics Summary (Mean, Median, Std, Variance, Percentiles)

**Concept:** a fuller statistical summary combining several np.* functions

**Problem:** Compute a fuller statistical summary of an array: mean, median, std, variance, and quartiles.

**Given:**
```
arr = np.array([10, 20, 30, 100, 200, 300])
```

**Expected Output:**
```
Mean, Median, Std, Variance, and 25th/75th percentiles
```

**Hint:** np.var() is simply np.std() squared; np.percentile(arr, q) generalizes the median (which is the 50th percentile).

````python
import numpy as np

arr = np.array([10, 20, 30, 100, 200, 300])

mean = np.mean(arr)
median = np.median(arr)
std = np.std(arr)
variance = np.var(arr)
q25 = np.percentile(arr, 25)
q75 = np.percentile(arr, 75)

print(f"Mean              : {mean}")
print(f"Median            : {median}")
print(f"Standard Deviation: {std}")
print(f"Variance          : {variance}")
print(f"25th percentile   : {q25}")
print(f"75th percentile   : {q75}")
print(f"\nVariance == Std**2: {np.isclose(variance, std ** 2)}")
````
