# Accepting a List as Input From a User

`input()` always returns a single plain string — there's no built-in 'read a list' function. To get a list from user input, you need to PARSE that string yourself: split it into pieces, then (optionally) convert each piece to the type you actually want.

This notebook walks through several ways to do that — from the most manual (a loop with explicit conversion) to the most compact (`map()` in one line) — plus reading strings and nested lists.

**Note:** every `input()` call in this notebook is simulated (pre-supplied values, echoed to output) so the notebook runs unattended, exactly as it would with a real user typing those same answers.

```python
# Helper used throughout this notebook to simulate input() non-interactively

import builtins

def simulate_input(*responses):

    it = iter(responses)

    def fake_input(prompt=""):

        val = next(it)

        print(f"{prompt}{val}")

        return val

    builtins.input = fake_input



_real_input = input

def restore_input():

    builtins.input = _real_input
```

## Basic Approach: `input()` + `.split()` + Manual Conversion

1. Read one line with `input()`.
2. `.split()` breaks it into a list of STRING pieces (default: split on any whitespace).
3. Loop through and convert each piece with `int()` (or `float()`) if you need actual numbers.

```python
simulate_input("5 10 15 20 25 30")



input_string = input('Enter elements of a list separated by space \n')

user_list = input_string.split()

print('string list: ', user_list)



for i in range(len(user_list)):

    user_list[i] = int(user_list[i])



print('User list: ', user_list)

print("Sum = ", sum(user_list))

restore_input()
```

**Note:** for a list of FLOATS instead of ints, just use `float()` in place of `int()` in the conversion loop.

## Reading a List One Element at a Time

Instead of one space-separated line, ask for the list's SIZE first, then loop that many times, reading (and converting) one number per `input()` call — useful when you want a separate prompt for each element.

```python
simulate_input("5", "5", "10", "15", "20", "25")



number_list = []

n = int(input("Enter the list size "))



print()

for i in range(0, n):

    print("Enter number at index", i)

    item = int(input())

    number_list.append(item)

print("User list is ", number_list)

restore_input()
```

## Reading a List via List Comprehension

A more compact, single-expression version of the split-and-convert pattern: `.strip()` trims any stray leading/trailing whitespace, `.split()` breaks it into pieces, and the comprehension converts each piece to `int()` in the same line. `[:n]` at the end trims the result down to the requested size, in case extra numbers were typed.

```python
simulate_input("5", "2 4 6 8 10")



n = int(input("Enter the size of the list "))

print()

num_list = list(int(num) for num in input("Enter the list items separated by space ").strip().split())[:n]



print("User list: ", num_list)

restore_input()
```

## Reading a List via `map()`

`map(function, iterable)` applies a conversion function to EVERY item in one call, without writing an explicit loop or comprehension — arguably the most idiomatic, readable version of this pattern. Swapping `float` for `int` (as shown here) reads a list of decimal numbers instead.

```python
simulate_input("5", "2.5 5.5 7.5 10.5 12.5")



n = int(input("Enter the size of list : "))

print()

numList = list(map(float, input("Enter the list numbers separated by space ").strip().split()))[:n]

print("User List: ", numList)

restore_input()
```

## Reading a List of Strings

Simplest case of all — if you actually WANT strings (not numbers), just skip the conversion step entirely and keep whatever `.split()` produces.

```python
simulate_input("Jessa Emma Scott Kelly Tom")



input_string = input("Enter all family members name separated by space  ")

family_list = input_string.split(" ")



print()

print("Printing all family member names")

for name in family_list:

    print(name)

restore_input()
```

## Reading a Nested List

For a genuinely 2D structure (a list of lists, all the same size), nest a comprehension inside another — the outer comprehension builds each ROW, and the inner comprehension (reading one `input()` per number) fills that row.

```python
simulate_input("3", "10", "20", "30", "40", "50", "60", "70", "80", "90")



list_size = int(input("Enter the number of sub list "))

print()

final_list = [[int(input("Enter single number and press enter: ")) for _ in range(list_size)] for _ in range(list_size)]

print("List is", final_list)

restore_input()
```

## Summary

| Technique | Best for |
|---|---|
| `.split()` + manual loop conversion | Clear, step-by-step, easy to follow for beginners |
| `input()` inside a `range()` loop | One prompt per element, explicit size known upfront |
| List comprehension (`[int(x) for x in ...]`) | Compact, single-line, same-line conversion |
| `map()` | Most idiomatic; easiest to swap `int`/`float`/`str` |
| Plain `.split()`, no conversion | Reading a list of STRINGS |
| Nested comprehension | Reading a genuinely 2D (nested) list |

All of these ultimately rely on the same two building blocks: `input()` returns a raw string, and you decide how to split and convert it into the structure you actually want.

## Practice Exercises

1. Simulate a user entering 6 numbers on one line, and print the list sorted in descending order.
2. Using `map()`, read a space-separated list of floats and print their average.
3. Read a list of words (space-separated) and print only the words longer than 4 characters.
4. Read a 3x3 nested list of integers (one `input()` call per number) and print the sum of each row.
5. Write a function `read_list_safe(raw)` that takes a raw space-separated string (simulating what `input()` would return) and returns a list of ints, skipping any piece that isn't a valid integer instead of crashing.

### Exercise 1 solution

```python
simulate_input("7 2 9 4 1 8")

numbers = list(map(int, input("Enter 6 numbers separated by space: ").split()))

numbers.sort(reverse=True)

print(numbers)

restore_input()
```

### Exercise 2 solution

```python
simulate_input("2.5 3.5 4.0 1.5")

values = list(map(float, input("Enter numbers separated by space: ").split()))

print("Average:", sum(values) / len(values))

restore_input()
```

### Exercise 3 solution

```python
simulate_input("cat elephant dog hippopotamus ox")

words = input("Enter words separated by space: ").split()

long_words = [w for w in words if len(w) > 4]

print(long_words)

restore_input()
```

### Exercise 4 solution

```python
simulate_input(*[str(n) for n in range(1, 10)])

size = 3

grid = [[int(input("Enter number: ")) for _ in range(size)] for _ in range(size)]

for row in grid:

    print(row, "-> row sum:", sum(row))

restore_input()
```

### Exercise 5 solution

```python
def read_list_safe(raw):

    result = []

    for piece in raw.split():

        try:

            result.append(int(piece))

        except ValueError:

            print(f"Skipping invalid value: {piece!r}")

    return result



print(read_list_safe("10 20 abc 30 xyz 40"))
```
