# Python StatementsA **statement** is any instruction the Python interpreter can execute. In the simplest terms, every line of working Python code is a statement. A statement ends at a NEWLINE character, so — with a couple of exceptions covered below — one line is normally one statement.`a = 10` is an assignment statement: `a` is a variable name, `10` is the value being assigned. Python has many kinds of statements — assignment, `print`, `if`, `for`, `while`, and more — but they all share the same core idea: an instruction the interpreter carries out, one after another, top to bottom.This notebook covers:- Single-line and multi-line statements- Multiple statements on one line- Compound statements (`if`, `for`, `while`, `try`, `with`)- Simple statements (expression, `pass`, `del`, `return`, `import`, `break`/`continue`)- Practice exercises at the end

## A Basic Sequence of Statements

A script is usually just a sequence of statements executed in order. Nothing gets printed twice or reordered — each statement runs exactly once, in the order it's written (barring loops, functions, or conditionals that alter that flow later on).

```python
# statement 1
print('Hello')

# statement 2
x = 20

# statement 3
print(x)
```

## Multiple Statements on One Line

Normally each statement gets its own line, but you can pack more than one onto a single line by separating them with semicolons (`;`). This works, but is used sparingly in real code — it's handy for quick one-liners, less so for anything meant to stay readable.

```python
# two statements in a single line
l = 10; b = 5

# statement 3
print('Area of rectangle:', l * b)
```

## Multi-Line Statements

A statement ends at a newline by default, but sometimes an expression is naturally long enough that splitting it across several lines improves readability. Python gives you two ways to do this.

### Explicit line continuation with `\`
A backslash at the end of a line tells Python "this statement isn't finished yet — keep reading on the next line." This is called *explicit* continuation, since you're the one explicitly marking where the line breaks.

```python
addition = 10 + 20 + \
           30 + 40 + \
           50 + 60 + 70
print(addition)
```

### Implicit continuation with `()`, `[]`, `{}`
Wrapping an expression in parentheses `()`, square brackets `[]`, or curly braces `{}` lets you spread it across multiple lines with **no backslash needed at all** — Python already knows the statement isn't finished until the matching closing bracket appears. This is called *implicit* continuation, and it's the far more common style in real-world code, since it reads more cleanly and doesn't break if you accidentally leave trailing whitespace after the backslash (a classic and confusing bug with explicit continuation).

The same bracket trick is exactly why list literals and dictionary literals are so often written one item per line — it's not special syntax, it's just implicit continuation being used for readability.

```python
addition = (10 + 20 +
            30 + 40 +
            50 + 60 + 70)
print(addition)

# list of strings, one item per line for readability
names = ['Emma',
         'Kelly',
         'Jessa']
print(names)

# dictionary: name as key, mark as value
students = {'Emma': 70,
            'Kelly': 65,
            'Jessa': 75}
print(students)
```

## Compound Statements

A **compound statement** contains (and controls the execution of) a *group* of other statements — typically introduced by a header line ending in `:`, followed by an indented block. The indented block only runs based on the logic of the header.

| Statement | Role |
|---|---|
| `if` | Runs its block only when a condition is true — a *conditional* statement |
| `while` | Repeats its block as long as a condition stays true — a *looping* statement |
| `for` | Iterates over a sequence (string, list, dict, set, tuple, or any iterable) — a *looping* statement |
| `try` | Defines exception handlers around a block that might raise an error |
| `with` | Wraps a block with setup/cleanup code (like automatically closing a file) |

Each of these gets its own full tutorial later in this series — `if`/`for`/`while` next, `try` in the exceptions notebook.

## Simple Statements

Beyond declarations and calculations, Python has several single-purpose *simple statements* — each does exactly one specific thing, contained on a single logical line.

### Expression Statements

An **expression** is any combination of values, variables, and operators that Python can evaluate down to a single value — even a bare value like `5` counts as a (trivial) expression. Typed alone in an interactive shell, an expression's value is displayed automatically.

But inside a script, an expression by itself does nothing useful — its computed value has nowhere to go. That's why expressions are almost always the *right-hand side* of an assignment: `x = x + 10` is a complete **statement**, where `x + 10` is the expression being evaluated and assigned.

```python
x = 5
# x + 10 alone is just an expression - it computes nothing useful on its own

# x = x + 10 is a complete statement: the expression's value is assigned to x
x = x + 10
print(x)
```

### The `pass` Statement

`pass` is a genuine no-op — literally nothing happens when it runs. It exists purely as a placeholder wherever Python's syntax *requires* an indented block (after a `:`), but you don't have anything to put there yet.

The classic use case: stubbing out a function you plan to implement later, without breaking the rest of your program in the meantime.

```python
def fun1(arg):
    pass  # a function that does nothing (yet)

fun1(42)  # runs without error, just does nothing
print("Called fun1 successfully - it did nothing, as designed.")
```

### The `del` Statement

`del` removes a variable (or other object reference) entirely. Once deleted, trying to use that name again raises `NameError`, exactly as if it had never been defined.

**Syntax:** `del target_list` — a comma-separated list of names to delete.

```python
x = 10
y = 30
print(x, y)

# delete x and y
del x, y

# try to access it
try:
    print(x, y)
except NameError as e:
    print("NameError:", e)
```

### The `return` Statement

Functions exist to perform a task and often hand back a result. `return` is what sends a value out of a function back to whoever called it — execution of the function stops immediately at that point.

```python
# function accepts two numbers and returns their sum
def addition(num1, num2):
    return num1 + num2  # return the sum of two numbers

# result is the return value
result = addition(10, 20)
print(result)
```

### The `import` Statement

`import` brings a module's code into your current program so you can use it. Python ships with a large standard library of built-in modules ready to import — no installation needed — covering everything from dates (`datetime`) to file paths (`os`) to randomness (`random`).

```python
import datetime

# get current datetime
now = datetime.datetime.now()
print(now)
```

### The `break` and `continue` Statements

Both alter a loop's normal flow:
- **`break`** exits the loop immediately, skipping any remaining iterations entirely.
- **`continue`** skips just the *rest of the current* iteration and jumps straight to the next one — the loop keeps going.

Both only make sense inside a loop body (`for` or `while`) — using either outside a loop is a `SyntaxError`. They get a full dedicated notebook of their own later in this series.

```python
for i in range(10):
    if i == 3:
        continue  # skip printing 3, move to next iteration
    if i == 6:
        break     # stop the loop entirely once i reaches 6
    print(i)
```
