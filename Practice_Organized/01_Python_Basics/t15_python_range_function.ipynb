# Python `range()` Function

`range()` generates an immutable SEQUENCE of integers, most commonly paired with a `for` loop to repeat something a specific number of times. Crucially, `range()` doesn't build a full list of numbers in memory up front — it's a lazy object that computes each value only as it's needed, which is why it stays fast and memory-efficient no matter how large a range you ask for.

**Syntax:** `range(start, stop[, step])` — `stop` is the only REQUIRED argument; `start` defaults to 0, `step` defaults to 1.

| Parameter | Meaning | Default |
|---|---|---|
| `start` | first number in the sequence | 0 |
| `stop` | generate numbers up to (but NEVER including) this number | required |
| `step` | difference between consecutive numbers | 1 |

```python
for i in range(6):

    print(i)
```

`range(6)` produced 0 through 5 — six numbers, but never 6 itself. This exclusive upper bound is one of the most important things to internalize about `range()`.

## `type(range(...))` and the Three Call Forms

`range()` returns an object of its own dedicated `range` class, not a list.

```python
print(type(range(10)))
```

### `range(stop)` — one argument

```python
for i in range(10):

    print(i, end=' ')
```

If `stop` is 0 or negative, the result is simply an empty sequence — not an error.

```python
print(list(range(0)))

print(list(range(-5)))
```

### `range(start, stop)` — two arguments

```python
for i in range(10, 16):

    print(i, end=' ')
```

If `stop` is less than or equal to `start` (with the default positive step), the result is empty.

```python
print(list(range(16, 10)))
```

### `range(start, stop, step)` — three arguments

```python
for i in range(10, 50, 5):

    print(i, end=' ')
```

**Important:** `step` must never be `0` — Python raises `ValueError` if you try.

```python
try:

    list(range(1, 5, 0))

except ValueError as e:

    print("ValueError:", e)
```

**Practice problem:** generate numbers from 9 to 100 divisible by 3.

```python
for i in range(9, 100, 3):

    print(i, end=' ')
```

## Key Rules to Remember

- All three arguments must be plain integers — no floats (see the linked float-range workaround in the summary below).
- All three can be positive OR negative.
- `step` can never be `0`.

## `for` Loop with `range()`: How It Actually Works

`range()` is lazy — it doesn't generate 1, 3, 5, 7, 9 all at once. On EACH loop iteration, it computes just the next value as `previous + step`, hands it to the loop variable, and waits for the next request. This is why `range()` uses the same tiny, constant amount of memory whether it represents 10 numbers or 10 million.

```python
for i in range(1, 10, 2):

    print("Current value of i is:", i)
```

## Iterating a List Using `range()`

Combine `range()` with `len()` to get INDEX numbers while looping — useful specifically when you need to also MODIFY list items by position, not just read them.

```python
list1 = ['Jessa', 'Emma', 20, 30, 75.5]

for i in range(len(list1)):

    print(list1[i])
```

**Practice problem:** print a number pattern where row N repeats N, N times.

```python
for num in range(4):

    for i in range(num):

        print(num, end=" ")

    print()
```

## Reverse Range

Two ways to count backward.

### Using a negative `step`

```python
for i in range(5, -1, -1):

    print(i)
```

### Using `reversed()`

```python
for i in reversed(range(10, 21)):

    print(i, end=' ')
```

```python
for i in reversed(range(10, 21, 2)):

    print(i, end=' ')
```

`reversed(range(...))` returns a distinct `range_iterator` type, not a `range` — convert to `list()` if you need a concrete reversed list.

```python
print(type(range(0, 5)))

print(type(reversed(range(0, 5))))
```

### Using `range()` to Reverse a List by Index

```python
list1 = [10, 20, 30, 40, 50]

for i in range(len(list1) - 1, -1, -1):

    print(list1[i], end=" ")
```

## `range()` Step in More Depth

Decrementing (a negative step) requires `start` to be GREATER than `stop`, or you'll silently get an empty sequence — the reverse of the usual positive-step rule.

```python
for i in range(30, 20, -2):

    print(i, end=' ')
```

```python
# start < stop with a NEGATIVE step -> empty sequence, not an error

result = list(range(20, 30, -2))

print(result)
```

```python
# Using step to generate multiples of 7

for i in range(7, 77, 7):

    print(i, end=' ')
```

## Negative `range()`

`start` and `stop` can both be negative too, independent of the step.

```python
# negative range from -1 to -10

for i in range(-1, -11, -1):

    print(i, end=', ')
```

```python
# negative-to-zero range using a POSITIVE step

for i in range(-10, 0):

    print(i, end=', ')
```

```python
# combining negative and positive numbers in one range

for i in range(2, -5, -1):

    print(i, end=", ")
```

## Converting `range()` to a `list`

`range()` itself is not a list — wrap it in `list()` to get a genuine, concrete list object.

```python
sample_list = list(range(2, 10, 2))

print(type(sample_list))

print(sample_list)

for item in sample_list:

    print(item)
```

```python
# Using range() to access AND modify list items by index

sample_list = list(range(10, 100, 10))

for i in range(0, len(sample_list), 1):

    sample_list[i] = sample_list[i] * 2

print(sample_list)
```

## Inclusive Range

`range()` is exclusive of `stop` by default. To INCLUDE the stop value, simply add one `step` to whatever `stop` was going to be.

```python
start = 1

stop = 5

step = 1

stop += step   # the trick: push stop one step further out



for i in range(start, stop, step):

    print(i, end=' ')
```

```python
# Even inclusive range

step = 2

for i in range(2, 20 + step, step):

    print(i, end=' ')
```

## `range()` vs. the Old Python 2 `xrange()`

This distinction only matters for legacy Python 2 code: Python 2 had both `range()` (built the FULL list in memory immediately — slow, memory-heavy for large ranges) and `xrange()` (lazy, one value at a time — exactly like Python 3's `range()`). Python 3 removed the old eager `range()` entirely and renamed `xrange()` to `range()` — so Python 3's `range()` already has the efficient, lazy behavior Python 2 needed `xrange()` for.

## Concatenating Two Ranges

`range` objects don't support `+` directly — use `itertools.chain()` to lazily stitch multiple ranges (or any iterables) together into one sequence.

```python
from itertools import chain



new_range = chain(range(5), range(5, 10))

for num in new_range:

    print(num, end=' ')
```

## `range()` Indexing and Slicing

A `range` object supports the same indexing and slicing syntax as a list, computed on demand rather than by storing every value.

```python
range1 = range(0, 10)

print(range1.start)

print(range1.stop)

print(range1.step)
```

```python
range1 = range(0, 10)

print(range1[0])              # first number

print(range1[5])               # 6th number (0-indexed)

print(range1[range1.stop - 1]) # last number
```

```python
# negative indexing

print(range(10)[-1])   # last number

print(range(10)[-2])   # second-to-last number
```

```python
# slicing a range returns another range object

sliced = range(10)[3:8]

print(sliced, type(sliced))

for i in sliced:

    print(i, end=' ')
```

## `range()` Over Characters

`range()` itself only works with integers — there's no built-in way to write `range('a', 'z')` directly. The workaround: convert characters to their integer ASCII/Unicode code points with `ord()`, range over THOSE integers, then convert each result back to a character with `chr()`. A small custom generator wraps this cleanly.

```python
def character_range(char1, char2):

    # +1 makes this an INCLUSIVE range - without it, char2 itself would never be reached

    for char in range(ord(char1), ord(char2) + 1):

        yield char



for letter in character_range('a', 'z'):

    print(chr(letter), end=', ')
```

## Bonus: Membership Testing With `in`

You don't need to convert a `range` to a list to check whether a value falls inside it — `in` works directly and efficiently on the range object itself.

```python
x = 10

if x in range(0, 10):

    print("x is between 0 and 9")

else:

    print("x is outside [0..9]")
```

## Summary

| Operation | Description |
|---|---|
| `range(stop)` | 0 to stop-1 |
| `range(start, stop)` | start to stop-1 |
| `range(start, stop, step)` | start, incrementing by step, stopping before stop |
| `range(5, -1, -1)` | reverse range via negative step |
| `reversed(range(5))` | reverse range via `reversed()` |
| `range(-1, -11, -1)` | negative range |
| `list(range(2, 10, 2))` | convert to a real list |
| `range(start, stop+step, step)` | inclusive range |
| `range(0, 10)[5]` | indexing |
| `range(10)[3:8]` | slicing |
| `range1.start` / `.stop` / `.step` | inspecting a range's own parameters |

**FAQ recap:** `range()` starts at 0 by default; it returns a `range` object (not a list); `sum(range(10))` sums it directly, no conversion needed.

```python
def char_range_reversed(c1, c2):

    for code_point in range(ord(c2), ord(c1) - 1, -1):

        yield chr(code_point)



print(list(char_range_reversed('a', 'f')))
```
