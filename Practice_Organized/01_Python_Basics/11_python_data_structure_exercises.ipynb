## Concepts you'll need

This set is a mixed-practice round for the four core Python collections — **list, dictionary, set, and tuple** — combining them in small realistic tasks.

- **List slicing with a step** — `list[start::step]` picks elements at regular intervals (e.g. every other index); `slice(start, stop)` builds a reusable slice object.
- **List mutation trio** — `.pop(index)` removes and returns an item; `.insert(index, item)` inserts at a position; `.append(item)` adds to the end.
- **Counting with a dictionary** — looping and checking `if item in count_dict` (or `dict.get(item, 0) + 1`) tallies occurrences without a library.
- **`zip()` for pairing** — combines two same-length sequences into tuples position by position; wrapping the result in `set()` deduplicates pairs.
- **Set relationship methods** — `.intersection()`, `.issubset()`, `.issuperset()`, `.remove()`, `.clear()` for comparing and cleaning up two sets.
- **Filtering a list against a dictionary's values** — `[item for item in a_list if item in a_dict.values()]` keeps only list items that also appear somewhere in a dictionary.
- **De-duplication** — checking `if val not in a_list` while looping preserves order; `list(set(a_list))` is faster but does not preserve order.
- **Combining structures** — converting a de-duplicated list into a `tuple()`, then applying `min()`/`max()` to it, shows how the four structures interoperate in a single task.

Each exercise below gives a problem, a hint, a solution, and an explanation.

## Exercise 1. Build a Third List From Two Lists

**Concept:** list slicing with a step

**Problem:** From two lists, build a third list containing the odd-indexed elements of the first list followed by the even-indexed elements of the second.

**Given:**
```
l1 = [3, 6, 9, 12, 15, 18, 21], l2 = [4, 8, 12, 16, 20, 24, 28]
```

**Expected Output:**
```
Element at odd-index positions from list one
[6, 12, 18]
Element at even-index positions from list two
[4, 12, 20, 28]
Printing Final third list
[6, 12, 18, 4, 12, 20, 28]
```

**Hint:** list1[1::2] starts at index 1 and steps by 2 (odd indices); list2[0::2] starts at 0 and steps by 2 (even indices).

````python
list1 = [3, 6, 9, 12, 15, 18, 21]
list2 = [4, 8, 12, 16, 20, 24, 28]
res = list()

odd_elements = list1[1::2]
print("Element at odd-index positions from list one")
print(odd_elements)

even_elements = list2[0::2]
print("Element at even-index positions from list two")
print(even_elements)

print("Printing Final third list")
res.extend(odd_elements)
res.extend(even_elements)
print(res)
````

**Explanation:** list1[1::2] starts at index 1 and steps forward by 2, landing exactly on the odd indices (1, 3, 5). list2[0::2] does the same starting from index 0, landing on the even indices. .extend() then appends each list's individual elements onto res, in order.

## Exercise 2. Remove and Reinsert a List Item

**Concept:** pop(), insert(), append()

**Problem:** Remove the item at index 4 from a list, then insert that same value at index 2, and also append it to the end.

**Given:**
```
sample_list = [34, 54, 67, 89, 11, 43, 94]
```

**Expected Output:**
```
List After removing element at index 4  [34, 54, 67, 89, 43, 94]
List after Adding element at index 2  [34, 54, 11, 67, 89, 43, 94]
List after Adding element at last  [34, 54, 11, 67, 89, 43, 94, 11]
```

**Hint:** pop(index) both removes and returns the value at that position, making it easy to reuse.

````python
sample_list = [34, 54, 67, 89, 11, 43, 94]

print("Original list ", sample_list)
element = sample_list.pop(4)
print("List After removing element at index 4 ", sample_list)

sample_list.insert(2, element)
print("List after Adding element at index 2 ", sample_list)

sample_list.append(element)
print("List after Adding element at last ", sample_list)
````

**Explanation:** pop(4) removes the value 11 (at index 4) and hands it back in the same step, so it's ready to reuse without a separate lookup. insert(2, element) shifts everything from index 2 onward one slot to the right to make room, and append(element) simply adds it again at the very end.

## Exercise 3. Slice a List Into 3 Chunks and Reverse Each

**Concept:** slice(), reversed()

**Problem:** Divide a list into three equal-sized chunks and print each chunk reversed.

**Given:**
```
sample_list = [11, 45, 8, 23, 14, 12, 78, 45, 89]
```

**Expected Output:**
```
Chunk  0 [11, 45, 8]
After reversing it  [8, 45, 11]
Chunk  1 [23, 14, 12]
After reversing it  [12, 14, 23]
Chunk  2 [78, 45, 89]
After reversing it  [89, 45, 78]
```

**Hint:** Divide the list length by 3 for the chunk size, then loop 3 times, advancing start/end each pass.

````python
sample_list = [11, 45, 8, 23, 14, 12, 78, 45, 89]
print("Original list ", sample_list)

length = len(sample_list)
chunk_size = int(length / 3)
start = 0
end = chunk_size

for i in range(3):
    indexes = slice(start, end)

    list_chunk = sample_list[indexes]
    print("Chunk ", i, list_chunk)

    print("After reversing it ", list(reversed(list_chunk)))

    start = end
    end += chunk_size
````

**Explanation:** slice(start, end) builds a reusable slice object equivalent to writing sample_list[start:end] directly. Each loop pass captures one chunk, prints its reversed version via reversed() wrapped in list(), then advances start and end forward by chunk_size to line up with the next chunk.

## Exercise 4. Count Occurrences of Each List Element

**Concept:** dictionary-based counting

**Problem:** Count how many times each element appears in a list and store the results in a dictionary.

**Given:**
```
sample_list = [11, 45, 8, 11, 23, 45, 23, 45, 89]
```

**Expected Output:**
```
Printing count of each item   {11: 2, 45: 3, 8: 1, 23: 2, 89: 1}
```

**Hint:** Check membership in the dictionary to decide between starting a new count or incrementing an existing one.

````python
sample_list = [11, 45, 8, 11, 23, 45, 23, 45, 89]
print("Original list ", sample_list)

count_dict = dict()
for item in sample_list:
    if item in count_dict:
        count_dict[item] += 1
    else:
        count_dict[item] = 1

print("Printing count of each item  ", count_dict)
````

**Explanation:** Each value in the list becomes a dictionary key. The first time a value is seen, it's added with a count of 1; every subsequent appearance of the same value just increments its existing count by 1.

## Exercise 5. Pair Elements From Two Lists Into a Set

**Concept:** zip() + set()

**Problem:** Create a set of tuples pairing up corresponding elements from two lists.

**Given:**
```
first_list = [2, 3, 4, 5, 6, 7, 8], second_list = [4, 9, 16, 25, 36, 49, 64]
```

**Expected Output:**
```
Result is  {(6, 36), (8, 64), (4, 16), (5, 25), (3, 9), (7, 49), (2, 4)}
```

**Hint:** zip() pairs elements positionally into tuples; wrap that in set() to make a set of those tuples.

````python
first_list = [2, 3, 4, 5, 6, 7, 8]
print("First List ", first_list)

second_list = [4, 9, 16, 25, 36, 49, 64]
print("Second List ", second_list)

result = zip(first_list, second_list)
result_set = set(result)
print(result_set)
````

**Explanation:** zip(first_list, second_list) lazily produces (2, 4), (3, 9), (4, 16), and so on, pairing elements at matching positions. set(result) then collects all of those pairs into a set — since tuples are hashable and each pair here happens to be unique, all seven pairs survive.

## Exercise 6. Find and Remove Set Intersection

**Concept:** set.intersection() + set.remove()

**Problem:** Find the elements common to two sets, then remove those shared elements from the first set.

**Given:**
```
first_set = {23, 42, 65, 57, 78, 83, 29}, second_set = {57, 83, 29, 67, 73, 43, 48}
```

**Expected Output:**
```
Intersection is  {57, 83, 29}
First Set after removing common element  {65, 42, 78, 23}
```

**Hint:** Compute the intersection first, then loop through it calling .remove() on the first set.

````python
first_set = {23, 42, 65, 57, 78, 83, 29}
second_set = {57, 83, 29, 67, 73, 43, 48}

print("First Set ", first_set)
print("Second Set ", second_set)

intersection = first_set.intersection(second_set)
print("Intersection is ", intersection)
for item in intersection:
    first_set.remove(item)

print("First Set after removing common element ", first_set)
````

**Explanation:** intersection() finds every value shared by both sets without modifying either one. Looping over that intersection and calling .remove() on first_set for each shared value strips those specific elements out, leaving only what was unique to first_set. (A single call to first_set.difference_update(intersection) would achieve the same result in one step.)

## Exercise 7. Subset/Superset Check, Then Clear

**Concept:** issubset(), issuperset(), clear()

**Problem:** Check whether either of two sets is a subset or superset of the other, and clear whichever set turns out to be the subset.

**Given:**
```
first_set = {57, 83, 29}, second_set = {57, 83, 29, 67, 73, 43, 48}
```

**Expected Output:**
```
First set is subset of second set - True
Second set is subset of First set -  False
First set is Super set of second set -  False
Second set is Super set of First set -  True
First Set  set()
Second Set  {67, 73, 43, 48, 83, 57, 29}
```

**Hint:** issubset() and issuperset() are mirror checks; clear() empties whichever set satisfies the subset condition.

````python
first_set = {57, 83, 29}
second_set = {57, 83, 29, 67, 73, 43, 48}

print("First Set ", first_set)
print("Second Set ", second_set)

print("First set is subset of second set -", first_set.issubset(second_set))
print("Second set is subset of First set - ", second_set.issubset(first_set))

print("First set is Super set of second set - ", first_set.issuperset(second_set))
print("Second set is Super set of First set - ", second_set.issuperset(first_set))

if first_set.issubset(second_set):
    first_set.clear()

if second_set.issubset(first_set):
    second_set.clear()

print("First Set ", first_set)
print("Second Set ", second_set)
````

**Explanation:** Since every element of first_set is also in second_set, first_set.issubset(second_set) is True, which is exactly why the following if block empties it with .clear(). second_set.issubset(first_set) is False (second_set has extra elements), so second_set is left completely untouched.

## Exercise 8. Filter a List Against Dictionary Values

**Concept:** list comprehension checking dict.values()

**Problem:** Keep only the elements of a list that also appear somewhere among a dictionary's values.

**Given:**
```
roll_number = [47, 64, 69, 37, 76, 83, 95, 97], sample_dict = {'Jhon':47, 'Emma':69, 'Kelly':76, 'Jason':97}
```

**Expected Output:**
```
After removing unwanted elements from list [47, 69, 76, 97]
```

**Hint:** [item for item in roll_number if item in sample_dict.values()] filters in one line.

````python
roll_number = [47, 64, 69, 37, 76, 83, 95, 97]
sample_dict = {'Jhon': 47, 'Emma': 69, 'Kelly': 76, 'Jason': 97}

print("List:", roll_number)
print("Dictionary:", sample_dict)

roll_number[:] = [item for item in roll_number if item in sample_dict.values()]
print("after removing unwanted elements from list:", roll_number)
````

**Explanation:** sample_dict.values() gives the four values to check against; the comprehension keeps a roll number only if it appears among them. Assigning to roll_number[:] (a full-slice assignment) replaces the list's contents in place rather than rebinding the name to a brand-new list object — useful if other variables reference the same list elsewhere.

## Exercise 9. Extract Unique Dictionary Values Into a List

**Concept:** loop + membership check to preserve order

**Problem:** Collect a dictionary's values into a list without adding any duplicate values.

**Given:**
```
speed = {'jan': 47, 'feb': 52, 'march': 47, 'April': 44, 'May': 52, 'June': 53, 'july': 54, 'Aug': 44, 'Sept': 54}
```

**Expected Output:**
```
[47, 52, 44, 53, 54]
```

**Hint:** Loop speed.values(), appending a value to the result list only if it isn't already there.

````python
speed = {'jan': 47, 'feb': 52, 'march': 47, 'April': 44, 'May': 52, 'June': 53,
         'july': 54, 'Aug': 44, 'Sept': 54}

print("Dictionary's values - ", speed.values())

speed_list = list()

for val in speed.values():
    if val not in speed_list:
        speed_list.append(val)
print("unique list", speed_list)
````

**Explanation:** Checking `if val not in speed_list` before appending prevents duplicates from being added a second time. Because the loop moves through the dictionary's values in their original order, the resulting list preserves the order each distinct value was first encountered — unlike list(set(speed.values())), which would not guarantee that order.

## Exercise 10. Deduplicate a List, Then Find Min/Max via a Tuple

**Concept:** set() dedup, tuple(), min(), max()

**Problem:** Remove duplicate values from a list, convert the result to a tuple, and find its minimum and maximum values.

**Given:**
```
sample_list = [87, 45, 41, 65, 94, 41, 99, 94]
```

**Expected Output:**
```
unique items (order may vary): {87, 45, 41, 65, 94, 99}
tuple: (87, 45, 41, 65, 94, 99)
min: 41
max: 99
```

**Hint:** set(sample_list) removes all duplicates (both repeated values, 41 and 94, are each kept only once); convert that to a tuple, then apply min()/max().

````python
sample_list = [87, 45, 41, 65, 94, 41, 99, 94]
print("Original list", sample_list)

unique_list = list(set(sample_list))
print("unique list", unique_list)

t = tuple(unique_list)
print("tuple ", t)

print("Minimum number is: ", min(t))
print("Maximum number is: ", max(t))
````

**Explanation:** set(sample_list) drops every duplicate — both 41 and 94 appear twice in the original list, so each is kept only once, leaving 6 unique values (not 5, since 94 is also a repeat). tuple() then locks that deduplicated collection into an immutable sequence, and min()/max() scan it directly for the smallest and largest values. Note that converting through a set does not preserve the original order.
