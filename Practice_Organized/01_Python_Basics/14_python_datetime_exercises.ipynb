## Concepts you'll need

This set covers the `datetime`, `calendar`, and `zoneinfo` modules — creating, formatting, parsing, and doing arithmetic on dates and times.

- **The core types** — `date` (year/month/day only), `time` (hour/minute/second only), `datetime` (both combined). `datetime.now()` / `date.today()` get the current moment; `datetime.combine(d, t)` merges separate date and time objects.
- **Formatting and parsing** — `.strftime(format)` turns a datetime *into* a string; `datetime.strptime(string, format)` parses a string *into* a datetime (the format must describe the string's exact structure). Common codes: `%Y` `%m` `%d` (year/month/day), `%H` `%M` `%S` (24h time), `%I` `%p` (12h + AM/PM), `%A` `%B` (full weekday/month names), `%j` (day of year), `%V` (ISO week number), `%f` (microseconds), `%Z` (timezone name).
- **Reading components** — a datetime object exposes `.year`, `.month`, `.day`, `.hour`, `.minute`, `.second`, `.microsecond`, and `.weekday()` (0=Monday..6=Sunday) directly as attributes/methods.
- **`timedelta`** — represents a *duration* (`timedelta(days=7)`, `timedelta(hours=5, minutes=30)`, etc.). Add or subtract it from a date/datetime with `+`/`-`; subtracting two dates/datetimes *produces* a timedelta, whose `.days` and `.total_seconds()` give the difference in different units.
- **Immutability & `.replace()`** — date/datetime objects can't be changed in place; `.replace(day=1)` (etc.) returns a *new* object with just the given field(s) changed, which is the standard trick for "first of the month" or clamping to a specific day.
- **The `calendar` module** — `calendar.isleap(year)`, `calendar.monthrange(year, month)` (returns `(weekday_of_1st, days_in_month)`), and `calendar.month(year, month)` (pretty-printed text calendar).
- **Timezones** — `zoneinfo.ZoneInfo("Region/City")` (standard library, Python 3.9+) attaches real-world timezone rules to a datetime; `.astimezone(ZoneInfo("UTC"))` converts between zones, correctly handling DST.

**A note on output:** several exercises use `datetime.now()` / `date.today()`, so their actual printed output will differ from the examples below depending on when you run the notebook — that's expected, and each explanation focuses on the technique rather than the exact numbers.

## Exercise 21. Calculate the Date 4 Months From Today

**Concept:** manual month/year arithmetic + calendar.monthrange for clamping

**Problem:** Find the date exactly 4 months from today, safely handling months of different lengths.

**Given:**
```
no input — uses date.today()
```

**Expected Output:**
```
A date 4 months ahead, with the day clamped if the target month is shorter (varies by run)
```

**Hint:** timedelta can't add 'months' directly — compute the new year/month with zero-based arithmetic, then clamp the day.

````python
from datetime import date
import calendar

today = date.today()
months_to_add = 4

month = today.month - 1 + months_to_add
year = today.year + month // 12
month = month % 12 + 1

max_day = calendar.monthrange(year, month)[1]
day = min(today.day, max_day)

future_date = today.replace(year=year, month=month, day=day)
print("Date 4 months from today:", future_date)
````

**Explanation:** Converting the month to zero-based (today.month - 1) before adding makes the wraparound arithmetic (month // 12 for the year carry, month % 12 + 1 to restore 1-based numbering) work out correctly across year boundaries. calendar.monthrange(year, month)[1] gives the target month's actual day count, so min(today.day, max_day) prevents an invalid date like 'February 31st' when starting from a 31-day month.

## Exercise 24. Find the Date of the Next Monday

**Concept:** weekday() + modulo arithmetic + timedelta

**Problem:** Calculate the date of the upcoming Monday, regardless of what day it is today.

**Given:**
```
no input — uses date.today()
```

**Expected Output:**
```
Next Monday: a date within the next 7 days (varies by run)
```

**Hint:** (7 - today.weekday()) % 7 gives the days-ahead offset; if that's 0 (today IS Monday), use 7 instead.

````python
from datetime import date, timedelta

today = date.today()
days_ahead = (7 - today.weekday()) % 7
if days_ahead == 0:
    days_ahead = 7

next_monday = today + timedelta(days=days_ahead)
print("Next Monday:", next_monday)
````

**Explanation:** weekday() returns 0 for Monday through 6 for Sunday. ## Exercise 25. Round Time to the Nearest Hour


## Exercise 26. List All Sundays in a Year

**Concept:** date iteration with timedelta stepping

**Problem:** Generate a list of every Sunday's date in a given year.

**Given:**
```
year = 2026
```

**Expected Output:**
```
52 (or 53) Sunday dates, from early January through late December
```

**Hint:** Advance to the first Sunday day-by-day, then step 7 days at a time for every subsequent Sunday.

````python
from datetime import date, timedelta

year = 2026
sundays = []
current = date(year, 1, 1)
end = date(year, 12, 31)

while current.weekday() != 6:
    current += timedelta(days=1)

while current <= end:
    sundays.append(current)
    current += timedelta(weeks=1)

print(f"Total Sundays in {year}: {len(sundays)}")
for sunday in sundays[:5]:
    print(sunday)
print("...")
for sunday in sundays[-2:]:
    print(sunday)
````

**Explanation:** The first loop advances day-by-day only until the very first Sunday of the year is found (weekday() == 6). From there, the second loop steps a full 7 days at a time — since every date exactly a week after a Sunday is guaranteed to also be a Sunday — which is far more efficient than checking every single day of the year individually.

## Exercise 27. Calculate Business Days Between Two Dates

**Concept:** weekday filtering over a date range

**Problem:** Count the days between two dates, excluding weekends.

**Given:**
```
start = date(2025, 7, 1), end = date(2025, 7, 31)
```

**Expected Output:**
```
Business days between 2025-07-01 and 2025-07-31: 22
```

**Hint:** weekday() < 5 matches Monday through Friday only; loop with a strict < so the end date isn't counted.

````python
from datetime import date, timedelta

start = date(2025, 7, 1)
end = date(2025, 7, 31)
business_days = 0
current = start
while current < end:
    if current.weekday() < 5:
        business_days += 1
    current += timedelta(days=1)
print(f"Business days between {start} and {end}: {business_days}")
````

**Explanation:** weekday() < 5 keeps only Monday (0) through Friday (4), skipping Saturday (5) and Sunday (6) automatically. Using while current < end (strict less-than) matches the common convention of counting days between two dates exclusively, so July 31st itself isn't included in the tally. Note this doesn't account for public holidays — a holidays set with an additional `and current not in holidays` check would handle that.

## Exercise 28. Convert Local Time to UTC

**Concept:** zoneinfo.ZoneInfo + .astimezone()

**Problem:** Attach a real-world timezone to a naive datetime, then convert it to UTC.

**Given:**
```
naive_dt = datetime(2025, 7, 15, 10, 30, 0), representing IST (UTC+5:30)
```

**Expected Output:**
```
Local (IST): 2025-07-15 10:30:00+05:30
UTC: 2025-07-15 05:00:00+00:00
```

**Hint:** `.replace(tzinfo=...)` stamps a timezone onto a naive datetime; `.astimezone(...)` then converts it, recalculating the clock time.

````python
from datetime import datetime
from zoneinfo import ZoneInfo

naive_dt = datetime(2025, 7, 15, 10, 30, 0)

local_dt = naive_dt.replace(tzinfo=ZoneInfo("Asia/Kolkata"))

utc_dt = local_dt.astimezone(ZoneInfo("UTC"))

print("Local (IST):", local_dt)
print("UTC:        ", utc_dt)
````

**Explanation:** ZoneInfo("Asia/Kolkata") pulls real IANA timezone rules from the standard library (Python 3.9+), no third-party package required. .replace(tzinfo=...) only stamps the zone onto the existing clock value — appropriate when you already know which zone the naive datetime represents — while .astimezone() actually recalculates the clock time for the new zone, which is why 10:30 IST correctly becomes 05:00 UTC (IST is 5 hours 30 minutes ahead).

## Exercise 29. Get Current Time in a Specific City

**Concept:** datetime.now(tz=ZoneInfo(...))

**Problem:** Print the current time in two different cities/timezones.

**Given:**
```
"Asia/Tokyo" and "America/New_York"
```

**Expected Output:**
```
Current times in Tokyo and New York, with timezone abbreviations (varies by run)
```

**Hint:** Passing tz= directly to datetime.now() gives the current moment already expressed in that zone.

````python
from datetime import datetime
from zoneinfo import ZoneInfo

timezones = {
    "Tokyo":    "Asia/Tokyo",
    "New York": "America/New_York",
}

for city, tz_name in timezones.items():
    current_time = datetime.now(tz=ZoneInfo(tz_name))
    formatted = current_time.strftime("%Y-%m-%d %H:%M:%S %Z")
    print(f"Current time in {city}: {formatted}")
````

**Explanation:** datetime.now(tz=ZoneInfo(tz_name)) is preferred over converting from UTC after the fact — it correctly handles Daylight Saving Time transitions for whichever zone is requested. Storing city-to-zone mappings in a dictionary makes it trivial to add more cities without touching the loop logic; %Z in the format string appends the zone's abbreviation (like JST or EDT) so each line is self-explanatory.

## Exercise 30. Calculate Date After N Working Days

**Concept:** a counting loop that skips weekends

**Problem:** Given a start date, find the date exactly N working days later.

**Given:**
```
start_date = date(2025, 7, 1), n = 10
```

**Expected Output:**
```
Date after 10 working days from 2025-07-01: 2025-07-15
```

**Hint:** Advance one day at a time; only increment the working-day counter on weekdays; the start date itself is never counted.

````python
from datetime import date, timedelta

def add_working_days(start_date, n):
    current = start_date
    working_days_counted = 0
    while working_days_counted < n:
        current += timedelta(days=1)
        if current.weekday() < 5:
            working_days_counted += 1
    return current

start_date = date(2025, 7, 1)
n = 10
result = add_working_days(start_date, n)
print(f"Date after {n} working days from {start_date}: {result}")
````

**Explanation:** Advancing the date first and checking it second means the starting date itself is never counted as one of the N working days, matching the everyday meaning of 'N working days from today.' weekday() < 5 filters to Monday-Friday, so weekends are stepped over without incrementing the counter. Wrapping this in a function makes it reusable for any start date or day count; extending it to skip public holidays too would just mean adding a `holidays` set and an extra condition.
