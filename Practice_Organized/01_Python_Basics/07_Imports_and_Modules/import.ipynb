##### Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
import math

# List all attributes, including functions, in the math module
print(dir(math))
```

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
from math import pow

print(pow(2,3))
```

```python
import custmath

print(custmath.addition(1,2))
```

```python
import custmath as cm

import importlib
importlib.invalidate_caches() # sometime python will not create __pycache__ folder for compiled code so it clear cache but not guaranteed


print(cm.subtraction(10,5))
#print(custmath.subtraction(11,7)) # once you alias you can only call with alias not with module name
print(cm.addition(4,5))
```

```python
from mypackage import custstat as cs

cs.meanofnumbers()
```

```python
import mypackage.custstat as cs2

cs2.meanofnumbers()
```

```python
#if your module is not in the same directory as the main script you're running,\ 
# Python will need to be able to find the module via the import system. 

import sys
sys.path.append('/path/to/your/module_directory')
```

#### Conclusion
Importing modules and packages in Python allows you to organize your code, reuse functionalities, and keep your projects clean and manageable.
