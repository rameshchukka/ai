## Concepts you'll need

This set covers Python **object-oriented programming**, from a bare class through the four OOP pillars and several dunder (magic) methods.

- **Classes and instances** — `class Name:` defines a blueprint; `__init__(self, ...)` is the constructor, called automatically when you write `Name(...)`. `self` always refers to the specific instance a method is running on.
- **Instance vs. class attributes** — `self.x = x` inside `__init__` creates a per-object attribute; a plain `x = value` written directly in the class body (outside any method) is shared by every instance until an instance overrides it individually.
- **Encapsulation** — a double-underscore prefix (`self.__balance`) triggers name mangling, making an attribute hard to reach from outside the class. `@property` turns a method into an attribute-style getter; a paired `@x.setter` lets assignment (`obj.x = value`) run validation logic invisibly.
- **Inheritance** — `class Child(Parent):` gives the child every attribute and method the parent has. `super().__init__(...)` calls the parent's constructor so shared setup isn't duplicated; `super().method(...)` similarly delegates to the parent's version of any method.
- **Polymorphism** — different subclasses can override the same method name (e.g. `speak()`) with their own implementation; calling it on a mixed list of objects runs whichever version matches each object's actual class, automatically.
- **Type inspection** — `type(obj)` gives the exact class; `isinstance(obj, Cls)` also matches subclasses (inheritance-aware); `issubclass(Child, Parent)` checks the class relationship directly.
- **Dunder (magic) methods** — special double-underscore methods that hook into Python's own syntax: `__add__` powers `+`, `__len__` powers `len(obj)`, `__call__` lets an instance be invoked like `obj(...)`, `__repr__`/`__str__` control how `print(obj)` displays it.
- **Composition ("has-a")** — a class can hold a list or attribute of *other* objects (like `Zoo` holding `Animal`s, or `Flight` holding `Passenger`s) without inheriting from them — a different relationship from the "is-a" of inheritance.

## Exercise 2. Vehicle Class With Instance Attributes

**Concept:** __init__ and self

**Problem:** Create a class with instance attributes set via the constructor.

**Given:**
```
vehicle1 = Vehicle("Tesla Model S", 250, 18)
```

**Expected Output:**
```
Vehicle Name: Tesla Model S, Speed: 250, Mileage: 18
```

## Exercise 3. Rectangle Class With Area & Perimeter

**Concept:** instance methods using self

**Problem:** Add methods that compute derived values from an object's own attributes.

**Given:**
```
rect = Rectangle(10, 4)
```

**Expected Output:**
```
Area = 40
Perimeter = 28
```

**Hint:** area(self) and perimeter(self) read self.length and self.width to compute their formulas.

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle(10, 4)
print("Area =", rect.area())
print("Perimeter =", rect.perimeter())
````

**Explanation:** Both methods derive their result purely from the instance's own stored attributes via self. Calling rect.area() automatically passes rect in as self — you never pass self explicitly at the call site.

## Exercise 4. Student Class With Average Grade

**Concept:** an attribute holding a list, plus list-based computation

**Problem:** Store a list as an instance attribute and compute a derived statistic from it.

**Given:**
```
s1 = Student("Alice", [85, 90, 78, 92, 88])
```

**Expected Output:**
```
Alice's Average Grade: 86.6
```

**Hint:** sum(self.marks) / len(self.marks) computes the mean directly.

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

s1 = Student("Alice", [85, 90, 78, 92, 88])
print(f"{s1.name}'s Average Grade: {s1.average()}")
````

**Explanation:** Instance attributes aren't limited to simple values — self.marks holds an entire list, and each Student object gets its own independent copy. sum()/len() computes the mean without any manual loop, and Python 3's / always returns a float.

## Exercise 5. Product Class With Stock Value Calculator

**Concept:** a computed method modeling a business calculation

**Problem:** Compute total stock value (price times quantity) from instance attributes.

**Given:**
```
p1 = Product("Laptop", 899.99, 5)
```

**Expected Output:**
```
Total stock value of Laptop: $4499.95
```

**Hint:** total_value() multiplies self.price by self.quantity; format the result with :.2f for currency.

````python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_value(self):
        return self.price * self.quantity

p1 = Product("Laptop", 899.99, 5)
print(f"Total stock value of {p1.name}: ${p1.total_value():.2f}")
````

**Explanation:** total_value() needs no external input — everything it requires is already stored on the instance. The :.2f format specifier rounds the float display to exactly two decimal places, the standard convention for currency.

## Exercise 6. Bank Account With Deposit & Overdraw Protection

**Concept:** conditional validation inside a mutating method

**Problem:** Allow deposits freely, but block withdrawals that would overdraw the account.

**Given:**
```
starting balance 1000; deposit 500; withdraw 200; then attempt to withdraw 2000
```

**Expected Output:**
```
Balance after deposit: 1500
Balance after withdrawal: 1300
Insufficient funds. Current balance: 1300
```

**Hint:** Check `amount <= self.balance` before deducting; otherwise print an insufficient-funds message instead.

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print(f"Balance after deposit: {self.balance}")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Balance after withdrawal: {self.balance}")
        else:
            print(f"Insufficient funds. Current balance: {self.balance}")
account = BankAccount(1000)
account.deposit(500)
account.withdraw(200)
account.withdraw(2000)
````

**Explanation:** deposit() mutates self.balance directly, and that change persists across every subsequent method call on the same object. The if guard in withdraw() enforces the rule that balance can never go negative, printing a rejection message instead of silently going through.

## Exercise 7. Light Class With On/Off State Toggle

**Concept:** a boolean instance attribute representing state

**Problem:** Model a light that remembers whether it's on or off, and reports its status.

**Given:**
```
turn_on(), status(), turn_off(), status()
```

**Expected Output:**
```
Light is ON
Current status: ON
Light is OFF
Current status: OFF
```

**Hint:** self.is_on starts False; each method flips it and status() reports it via a ternary.

````python
class Light:
    def __init__(self):
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print("Light is ON")

    def turn_off(self):
        self.is_on = False
        print("Light is OFF")

    def status(self):
        state = "ON" if self.is_on else "OFF"
        print(f"Current status: {state}")

light = Light()
light.turn_on()
light.status()
light.turn_off()
light.status()
````

**Explanation:** self.is_on is the single source of truth for the light's state, initialized to False when the object is created. Each toggle method simply flips this boolean and stores it, and status() converts it into a readable string with a compact ternary expression rather than a full if/else block.

## Exercise 8. User Class With Password Validation

**Concept:** a method that checks rather than exposes sensitive data

**Problem:** Verify a candidate password against a stored one through a dedicated method.

**Given:**
```
u1 = User("alice", "secure123")
```

**Expected Output:**
```
True
False
```

**Hint:** check_password() compares the input to self.password with == and returns the boolean directly.

````python
class User:
    def __init__(self, username, password):
        self.username = username
        self.password = password

    def check_password(self, input_password):
        return self.password == input_password

u1 = User("alice", "secure123")
print(u1.check_password("secure123"))
print(u1.check_password("wrongpass"))
````

## Exercise 9. Temperature Class With Unit Converters

**Concept:** multiple methods deriving values from one shared attribute

**Problem:** Store a Celsius temperature and provide Fahrenheit and Kelvin conversions.

**Given:**
```
t = Temperature(100)
```

**Expected Output:**
```
Celsius: 100
Fahrenheit: 212.0
Kelvin: 373.15
```

**Hint:** to_fahrenheit() applies (C * 9/5) + 32; to_kelvin() applies C + 273.15.

````python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return (self.celsius * 9 / 5) + 32

    def to_kelvin(self):
        return self.celsius + 273.15

t = Temperature(100)
print("Celsius:", t.celsius)
print("Fahrenheit:", t.to_fahrenheit())
print("Kelvin:", t.to_kelvin())
````

**Explanation:** self.celsius is the single stored value both conversion methods derive from — updating it would automatically change what both conversions report. Since 9/5 and 273.15 involve floating-point arithmetic, both results come back as floats even though the input was an integer.

## Exercise 10. Notebook Class With Add & Display Notes

**Concept:** a growing list managed entirely inside the class

**Problem:** Maintain a list of notes that grows over the object's lifetime, with a numbered display.

**Given:**
```
add three notes: "Buy groceries", "Read a book", "Call the doctor"
```

**Expected Output:**
```
1. Buy groceries
2. Read a book
3. Call the doctor
```

**Hint:** self.notes = [] must live inside __init__, not at the class level, or all instances would share one list.

````python
class Notebook:
    def __init__(self):
        self.notes = []

    def add_note(self, note):
        self.notes.append(note)

    def show_notes(self):
        for i, note in enumerate(self.notes, start=1):
            print(f"{i}. {note}")

nb = Notebook()
nb.add_note("Buy groceries")
nb.add_note("Read a book")
nb.add_note("Call the doctor")
nb.show_notes()
````

**Explanation:** Initializing self.notes = [] inside __init__ is essential — defining it at the class level instead would make every Notebook instance share the exact same list, so notes added to one notebook would wrongly appear in all of them. enumerate(self.notes, start=1) numbers the notes starting at 1 without a manual counter variable.

## Exercise 11. Coffee Machine With Multi-Resource Tracking

**Concept:** checking multiple preconditions before mutating state

**Problem:** Check three resource levels at once before making a latte, only deducting if all three are sufficient.

**Given:**
```
CoffeeMachine(water=300, coffee=100, milk=200); a latte needs 200ml water, 20g coffee, 150ml milk
```

**Expected Output:**
```
Latte made! Remaining - Water: 100ml, Coffee: 80g, Milk: 50ml
Not enough resources to make a latte.
```

**Hint:** Combine all three resource checks with `and` in one if condition, so deduction only happens if every check passes.

````python
class CoffeeMachine:
    def __init__(self, water, coffee, milk):
        self.water = water
        self.coffee = coffee
        self.milk = milk

    def make_latte(self):
        water_needed = 200
        coffee_needed = 20
        milk_needed = 150

        if self.water >= water_needed and self.coffee >= coffee_needed and self.milk >= milk_needed:
            self.water -= water_needed
            self.coffee -= coffee_needed
            self.milk -= milk_needed
            print(f"Latte made! Remaining - Water: {self.water}ml, Coffee: {self.coffee}g, Milk: {self.milk}ml")
        else:
            print("Not enough resources to make a latte.")

machine = CoffeeMachine(water=300, coffee=100, milk=200)
machine.make_latte()
machine.make_latte()
````

**Explanation:** Combining all three checks with and means the deduction only ever happens when every resource is sufficient — preventing a scenario where water and milk get consumed but coffee runs out midway. Because the machine's state is genuinely mutated after a success, the second call correctly fails once water drops to 100ml (below the 200ml needed).

## Exercise 12. Shared Class Attribute Across Instances

**Concept:** class attributes vs. instance attributes

**Problem:** Demonstrate that a class-level attribute is shared until an instance overrides it, and that changing it via the class affects every instance still relying on the default.

**Given:**
```
v1 = Vehicle("Tesla", 250), v2 = Vehicle("BMW", 200)
```

**Expected Output:**
```
Tesla - Color: White, Speed: 250
BMW - Color: White, Speed: 200
Tesla - Color: Red, Speed: 250
BMW - Color: Red, Speed: 200
```

**Hint:** Vehicle.color = 'Red' updates the shared class attribute; v1.color = 'Red' would instead only affect v1.

````python
class Vehicle:
    color = "White"

    def __init__(self, name, max_speed):
        self.name = name
        self.max_speed = max_speed

v1 = Vehicle("Tesla", 250)
v2 = Vehicle("BMW", 200)

print(f"{v1.name} - Color: {v1.color}, Speed: {v1.max_speed}")
print(f"{v2.name} - Color: {v2.color}, Speed: {v2.max_speed}")

Vehicle.color = "Red"

print(f"{v1.name} - Color: {v1.color}, Speed: {v1.max_speed}")
print(f"{v2.name} - Color: {v2.color}, Speed: {v2.max_speed}")
````

**Explanation:** color declared directly in the class body (not inside __init__) belongs to the class itself, not to any one instance — when Python looks up v1.color and finds no instance attribute by that name, it walks up and finds the class attribute instead. Reassigning via Vehicle.color = "Red" updates that shared value, so every instance still relying on it (both v1 and v2) reflects the change immediately; assigning v1.color = "Red" directly would instead create a separate instance attribute on v1 alone, leaving v2 unaffected.

## Exercise 13. Bus Subclass Inheriting From Vehicle

**Concept:** basic inheritance with class Bus(Vehicle)

**Problem:** Create a subclass that adds nothing new, confirming it still has full access to the parent's method.

**Given:**
```
bus1 = Bus("School Bus", 120)
```

**Expected Output:**
```
Vehicle: School Bus, Max Speed: 120 km/h
```

**Hint:** class Bus(Vehicle): pass inherits everything automatically.

## Exercise 14. Override a Parent Method Using super()

**Concept:** method overriding + super() to reuse parent logic

**Problem:** Override a method to supply a default argument, while still delegating the actual work to the parent's version.

**Given:**
```
bus = Bus("School Bus", 120)
```

**Expected Output:**
```
School Bus seating capacity is: 50
```

**Hint:** super().seating_capacity(50) calls the parent's version, passing the bus-specific default of 50.

````python
class Vehicle:
    def __init__(self, name, max_speed):
        self.name = name
        self.max_speed = max_speed

    def seating_capacity(self, capacity):
        print(f"{self.name} seating capacity is: {capacity}")

class Bus(Vehicle):
    def seating_capacity(self):
        super().seating_capacity(50)

bus = Bus("School Bus", 120)
bus.seating_capacity()
````

**Explanation:** Bus overrides seating_capacity() with a no-argument version — this is the override, since calling it on a Bus instance runs this version instead of the parent's. super() returns a reference to the parent class, letting the child delegate the actual printing logic back to it, but supplying its own hardcoded default of 50 — combining a child-specific default with shared parent logic in one clean step.

## Exercise 15. Add a Maintenance Fee via super() in a Child Constructor

**Concept:** super().__init__() to extend, not replace, parent setup

**Problem:** Extend a parent's fare model with an additional child-specific fee, computed on top of inherited data.

**Given:**
```
base_fare = 500
```

**Expected Output:**
```
Total fare with maintenance fee: 550.0
```

**Hint:** super().__init__(base_fare) initializes the parent first, then the child adds its own maintenance_fee attribute.

````python
class Vehicle:
    def __init__(self, base_fare):
        self.base_fare = base_fare

class Taxi(Vehicle):
    def __init__(self, base_fare):
        super().__init__(base_fare)
        self.maintenance_fee = base_fare * 0.10

    def total_fare(self):
        return self.base_fare + self.maintenance_fee

taxi = Taxi(500)
print("Total fare with maintenance fee:", taxi.total_fare())
````

**Explanation:** super().__init__(base_fare) ensures self.base_fare gets set up exactly the way the parent defines it, before Taxi adds anything of its own. This lets Taxi extend Vehicle's behavior (adding a maintenance fee) without needing to duplicate or modify a single line of the parent's constructor.

## Exercise 16. Polymorphism With Dog & Cat speak()

**Concept:** overriding the same method name differently per subclass

**Problem:** Give two subclasses their own version of a shared method name.

**Given:**
```
Dog() and Cat() instances
```

**Expected Output:**
```
Dog says: Woof!
Cat says: Meow!
```

**Hint:** Each subclass overrides speak() with its own return value, replacing the parent's generic version.

````python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

dog = Dog()
cat = Cat()

print("Dog says:", dog.speak())
print("Cat says:", cat.speak())
````

## Exercise 17. Full-Time vs. Part-Time Employee Pay Logic

**Concept:** shared base class, divergent subclass calculations

**Problem:** Model two employee types that share a name attribute but calculate pay completely differently.

**Given:**
```
FullTimeEmployee("Alice", 60000), PartTimeEmployee("Bob", 500, 20)
```

**Expected Output:**
```
Alice's monthly pay: 5000.0
Bob's monthly pay: 10000
```

**Hint:** super().__init__(name) in both subclasses avoids duplicating the shared name assignment.

````python
class Employee:
    def __init__(self, name):
        self.name = name

    def calculate_pay(self):
        return 0

class FullTimeEmployee(Employee):
    def __init__(self, name, annual_salary):
        super().__init__(name)
        self.annual_salary = annual_salary

    def calculate_pay(self):
        return self.annual_salary / 12

class PartTimeEmployee(Employee):
    def __init__(self, name, hourly_rate, hours_worked):
        super().__init__(name)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_pay(self):
        return self.hourly_rate * self.hours_worked

ft = FullTimeEmployee("Alice", 60000)
pt = PartTimeEmployee("Bob", 500, 20)

print(f"{ft.name}'s monthly pay: {ft.calculate_pay()}")
print(f"{pt.name}'s monthly pay: {pt.calculate_pay()}")
````

**Explanation:** Employee holds only what's genuinely shared (name) plus a placeholder calculate_pay() that returns 0. Each subclass calls super().__init__(name) to avoid re-writing that assignment, then defines its own entirely different calculate_pay() formula — annual salary divided by 12 for full-time, hourly rate times hours for part-time.

## Exercise 18. Shape Subclasses With Custom area() Methods

**Concept:** polymorphism across three different geometric formulas

**Problem:** Give three shape subclasses their own area() implementation, then loop over mixed shape types uniformly.

**Given:**
```
Circle(7), Square(4), Triangle(6, 8)
```

**Expected Output:**
```
Circle area: 153.94
Square area: 16
Triangle area: 24.0
```

**Hint:** type(shape).__name__ retrieves the actual runtime class name for a reusable print loop.

````python
class Shape:
    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return round(3.14159 * self.radius ** 2, 2)

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

shapes = [Circle(7), Square(4), Triangle(6, 8)]
for shape in shapes:
    print(f"{type(shape).__name__} area: {shape.area()}")
````

**Explanation:** Each subclass overrides area() with a formula appropriate to its own shape — there's no way a single shared formula could work for all three. type(shape).__name__ dynamically retrieves each object's actual class name at runtime, letting one loop print correct labels for every shape type without hardcoding any names.

## Exercise 19. Media Subclasses With Type-Specific Attributes

**Concept:** shared base fields plus subclass-specific extras

**Problem:** Model different media types that share title/price but each carry their own unique extra attribute.

**Given:**
```
Book("Clean Code", 499, "Robert C. Martin"), Magazine("Wired", 150, "Monthly"), DVD("Inception", 299, 148)
```

**Expected Output:**
```
Book: Clean Code by Robert C. Martin - Rs.499
Magazine: Wired (Monthly) - Rs.150
DVD: Inception, 148 mins - Rs.299
```

**Hint:** Each subclass calls super().__init__(title, price) then adds one attribute unique to its own type.

````python
class Media:
    def __init__(self, title, price):
        self.title = title
        self.price = price

    def describe(self):
        return f"{self.title} - Rs.{self.price}"

class Book(Media):
    def __init__(self, title, price, author):
        super().__init__(title, price)
        self.author = author

    def describe(self):
        return f"Book: {self.title} by {self.author} - Rs.{self.price}"

class Magazine(Media):
    def __init__(self, title, price, frequency):
        super().__init__(title, price)
        self.frequency = frequency

    def describe(self):
        return f"Magazine: {self.title} ({self.frequency}) - Rs.{self.price}"

class DVD(Media):
    def __init__(self, title, price, duration):
        super().__init__(title, price)
        self.duration = duration

    def describe(self):
        return f"DVD: {self.title}, {self.duration} mins - Rs.{self.price}"

items = [
    Book("Clean Code", 499, "Robert C. Martin"),
    Magazine("Wired", 150, "Monthly"),
    DVD("Inception", 299, 148)
]

for item in items:
    print(item.describe())
````

**Explanation:** title and price genuinely belong on every media type, so they live in the shared Media base class; author, frequency, and duration are each specific to exactly one subclass and would make no sense sitting in the shared parent. Calling the same describe() method on three different object types and getting three differently-formatted results is polymorphism in direct action.

## Exercise 20. Discounted Order Subclass With 10% Off

**Concept:** overriding a calculated method without touching the parent

**Problem:** Extend an Order class so a subclass applies a discount, while the underlying stored total stays unchanged.

**Given:**
```
DiscountedOrder("ORD001", 1200)
```

**Expected Output:**
```
Order ID: ORD001
Original Total: 1200
Discounted Total: 1080.0
```

**Hint:** get_total() is overridden in the subclass to return self.total * 0.90, while self.total itself is left untouched.

````python
class Order:
    def __init__(self, order_id, total):
        self.order_id = order_id
        self.total = total

    def get_total(self):
        return self.total

class DiscountedOrder(Order):
    def __init__(self, order_id, total):
        super().__init__(order_id, total)

    def get_total(self):
        return self.total * 0.90

order = DiscountedOrder("ORD001", 1200)
print("Order ID:", order.order_id)
print("Original Total:", order.total)
print("Discounted Total:", order.get_total())
````

**Explanation:** Accessing order.total directly still reflects the original, undiscounted amount, while order.get_total() applies the 10% reduction — keeping the raw data and the derived value cleanly separate. The parent Order class needed zero modification to support this: it's closed for changes but open for exactly this kind of extension via subclassing.

## Exercise 21. Vehicle Class Hierarchy With Bike, Truck & Bus

**Concept:** subclasses hardcoding different values into a shared parent field

**Problem:** Give three vehicle subclasses their own fixed max_speed, using one shared describe() method from the parent.

**Given:**
```
Bike(), Truck(), Bus()
```

**Expected Output:**
```
Bike max speed: 120 km/h
Truck max speed: 90 km/h
Bus max speed: 100 km/h
```

**Hint:** Each subclass's __init__ calls super().__init__(<its own speed>) — no describe() override is needed at all.

````python
class Vehicle:
    def __init__(self, max_speed):
        self.max_speed = max_speed

    def describe(self):
        print(f"{type(self).__name__} max speed: {self.max_speed} km/h")

class Bike(Vehicle):
    def __init__(self):
        super().__init__(120)

class Truck(Vehicle):
    def __init__(self):
        super().__init__(90)

class Bus(Vehicle):
    def __init__(self):
        super().__init__(100)

vehicles = [Bike(), Truck(), Bus()]
for v in vehicles:
    v.describe()
````

**Explanation:** Each subclass's constructor hardcodes its own speed value and forwards it straight to the parent via super().__init__() — no subclass needs any attributes or methods beyond that single line. type(self).__name__ inside the shared describe() dynamically picks up whichever subclass is actually calling it, so one method serves all three vehicle types without needing to be overridden anywhere.

## Exercise 23. Type Checking With isinstance() & issubclass()

**Concept:** inheritance-aware type checks

**Problem:** Distinguish exact type checking from inheritance-aware checking.


**Expected Output:**
```
Is d an instance of Dog? True
Is d an instance of Animal? True
Is Dog a subclass of Animal? True

```

**Hint:** isinstance() considers the whole inheritance chain; issubclass() checks the class relationship itself, not an instance.

````python
class Animal:
    pass

class Dog(Animal):
    pass

d = Dog()

print("Is d an instance of Dog?", isinstance(d, Dog))
print("Is d an instance of Animal?", isinstance(d, Animal))
print("Is Dog a subclass of Animal?", issubclass(Dog, Animal))
print("Is Animal a subclass of Dog?", issubclass(Animal, Dog))
````

## Exercise 24. Vector Addition Using __add__ Overloading

**Concept:** operator overloading with __add__ and __repr__

**Problem:** Make a custom class support the + operator between two instances.

**Given:**
```

```

**Expected Output:**
```

```

**Hint:** __add__(self, other) is what Python calls automatically when it sees self + other.

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 1)

result = v1 + v2
print(result)
````

**Explanation:** Writing v1 + v2 automatically calls v1.__add__(v2) behind the scenes, with self as the left operand and other as the right. Without __repr__, print(result) would instead show an unhelpful default like <__main__.Vector object at 0x...>.

## Exercise 26. Private Balance With Property Getter & Setter

**Concept:** name mangling + @property + @x.setter

**Problem:** Store a balance privately and only allow non-negative values through a validated setter.

**Given:**
```
BankAccount(1000), deposit(500), then attempt to set balance to -200
```

**Expected Output:**
```
Current balance: 1000
Current balance: 1500
Invalid balance. Must be non-negative.
```

**Hint:** self.__balance triggers name mangling; @property/@balance.setter make account.balance read/write with validation.

````python
class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            print("Invalid balance. Must be non-negative.")
        else:
            self.__balance = amount

    def deposit(self, amount):
        self.balance = self.__balance + amount

account = BankAccount(1000)
print("Current balance:", account.balance)

account.deposit(500)
print("Current balance:", account.balance)

account.balance = -200
````

**Explanation:** The double-underscore prefix on __balance triggers name mangling, internally renaming it to _BankAccount__balance and making it awkward to reach directly from outside the class. @property turns the balance method into something read like a plain attribute (account.balance, no parentheses), while @balance.setter intercepts any assignment to account.balance and runs the non-negative check before actually updating __balance. Notably, deposit() writes through self.balance = ... rather than self.__balance = ... directly, so every balance change — including from deposit() — passes through the same validation.

## Exercise 27. Callable Object Class Using __call__

**Concept:** making instances invocable like functions

**Problem:** Give a class stateful, function-like behavior via __call__.

**Given:**
```
Multiplier(3) called with 10, Multiplier(5) called with 7
```

**Expected Output:**
```
30
35
```

**Hint:** __call__(self, value) runs whenever the instance itself is invoked with parentheses, like triple(10).

````python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return self.factor * value

triple = Multiplier(3)
penta = Multiplier(5)

print(triple(10))
print(penta(7))
````

## Exercise 28. Flight Class With Passenger Capacity Check

**Concept:** object composition + a boundary/capacity check

**Problem:** Model a Flight that owns a growing list of Passenger objects and enforces a hard capacity limit.

**Given:**
```
a Flight with capacity 2, then three booking attempts
```

**Expected Output:**
```
Alice booked on Flight AI202.
Bob booked on Flight AI202.
Sorry, Flight AI202 is fully booked.
```

**Hint:** book() checks len(self.passengers) < self.capacity before appending — the enforcement point for the business rule.

````python
class Passenger:
    def __init__(self, name):
        self.name = name

class Flight:
    def __init__(self, flight_number, capacity):
        self.flight_number = flight_number
        self.capacity = capacity
        self.passengers = []

    def book(self, passenger):
        if len(self.passengers) < self.capacity:
            self.passengers.append(passenger)
            print(f"{passenger.name} booked on Flight {self.flight_number}.")
        else:
            print(f"Sorry, Flight {self.flight_number} is fully booked.")

flight = Flight("AI202", 2)
flight.book(Passenger("Alice"))
flight.book(Passenger("Bob"))
flight.book(Passenger("Charlie"))
````

**Explanation:** Flight owns a list of Passenger objects — a 'has-a' relationship known as composition, distinct from inheritance. The guard condition in book() enforces the capacity limit right at the point new data would be added, which is exactly where a business rule like this belongs; the third booking attempt correctly fails once the list has reached its capacity of 2.

## Exercise 29. Zoo Class That Feeds All Animals

**Concept:** composition combined with polymorphism

**Problem:** Build a container class that calls a shared method on every object it holds, regardless of each object's specific subtype.

**Given:**
```
a zoo containing a Lion, an Elephant, and a Parrot
```

**Expected Output:**
```
Lion eats meat.
Elephant eats grass.
Parrot eats seeds.
```

**Hint:** feed_all() loops self.animals and calls .eat() on each — it never needs to know which specific subclass it's dealing with.

````python
class Animal:
    def eat(self):
        return "eating."

class Lion(Animal):
    def eat(self):
        return "Lion eats meat."

class Elephant(Animal):
    def eat(self):
        return "Elephant eats grass."

class Parrot(Animal):
    def eat(self):
        return "Parrot eats seeds."

class Zoo:
    def __init__(self):
        self.animals = []

    def add_animal(self, animal):
        self.animals.append(animal)

    def feed_all(self):
        for animal in self.animals:
            print(animal.eat())

zoo = Zoo()
zoo.add_animal(Lion())
zoo.add_animal(Elephant())
zoo.add_animal(Parrot())

zoo.feed_all()
````

**Explanation:** Zoo does not inherit from Animal at all — it simply contains a list of Animal objects, another example of composition ('has-a') rather than inheritance ('is-a'). feed_all() calls .eat() uniformly on every stored animal, and because each subclass overrides eat() differently, Python automatically dispatches to the correct version for each object — this is runtime polymorphism in action. Adding a brand-new animal type later would require creating just one new subclass; Zoo and feed_all() would need zero changes.

## Exercise 30. Character Class With Auto Level-Up Logic

**Concept:** stateful game logic with overflow handling

**Problem:** Automatically level up a character and carry over excess experience once a threshold is reached.

**Given:**
```
Character("Aria", health=100), then gain_exp(60) twice
```

**Expected Output:**
```
Aria gained 60 exp. (Total: 60)
Aria gained 60 exp. Level up! Now Level 2. (Remaining exp: 20)
```

**Hint:** Subtract 100 (not reset to 0) from exp on level-up, so any excess correctly carries forward.

````python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health
        self.exp = 0
        self.level = 1

    def gain_exp(self, amount):
        self.exp += amount
        if self.exp >= 100:
            self.exp -= 100
            self.level += 1
            print(f"{self.name} gained {amount} exp. Level up! Now Level {self.level}. (Remaining exp: {self.exp})")
        else:
            print(f"{self.name} gained {amount} exp. (Total: {self.exp})")

hero = Character("Aria", health=100)
hero.gain_exp(60)
hero.gain_exp(60)
````

**Explanation:** exp and level default to 0 and 1 in __init__, since every new character reasonably starts at those values. self.exp += amount accumulates across multiple calls, and the >= 100 check runs fresh after every single gain. Subtracting exactly 100 (rather than resetting to 0) on a level-up correctly carries over any excess — 60 + 60 = 120 total, so after leveling up, 20 exp remains rather than being discarded. (Using a while loop instead of if would handle multiple level-ups from one very large exp gain.)

## Exercise 31. Playlist Class With Add, Remove & Shuffle

**Concept:** composition + list filtering + the random module

**Problem:** Manage a collection of Song objects supporting add, remove-by-title, and randomized shuffling.

**Given:**
```
a playlist with three songs, then a removal and a shuffle
```

**Expected Output:**
```
Playlist: Blinding Lights, Levitating, Peaches
Removed: Levitating
After shuffle: (order will vary)
```

**Hint:** A list comprehension filters out a matching title for removal; random.shuffle() reorders the list in place.

````python
import random

class Song:
    def __init__(self, title, artist):
        self.title = title
        self.artist = artist

class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)

    def remove_song(self, title):
        original_count = len(self.songs)
        self.songs = [s for s in self.songs if s.title != title]
        if len(self.songs) < original_count:
            print(f"Removed: {title}")
        else:
            print(f"Song '{title}' not found in playlist.")

    def shuffle(self):
        random.shuffle(self.songs)

    def display(self):
        titles = [s.title for s in self.songs]
        print(f"Playlist: {', '.join(titles)}")

playlist = Playlist("My Mix")
playlist.add_song(Song("Blinding Lights", "The Weeknd"))
playlist.add_song(Song("Levitating", "Dua Lipa"))
playlist.add_song(Song("Peaches", "Justin Bieber"))

playlist.display()
playlist.remove_song("Levitating")
playlist.shuffle()
print("After shuffle:", end=" ")
playlist.display()
````

**Explanation:** Playlist stores Song objects rather than plain strings, making it easy to filter or display by either title or artist later. remove_song() builds a new list that excludes the matching title via a comprehension, then compares the before/after length to report whether a removal actually happened. random.shuffle() reorders self.songs in place with no new list created, so the displayed order will differ on every run of the notebook.
