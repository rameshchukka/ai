# Python Data Types

A **data type** determines what kind of value a variable can hold and what operations make sense on it. Python is dynamically typed, so you never declare a type up front — whatever value you assign determines the type automatically.

Python's built-in types fall into four broad families:
- **Numeric** — `int`, `float`, `complex`
- **Sequence** — `str`, `list`, `tuple` (plus `range`, `bytes`, `bytearray`)
- **Set** — `set`, `frozenset`
- **Mapping** — `dict`

Plus `bool` (technically a subtype of `int`) and a couple of memory-level types (`memoryview`). Use `type(x)` to check an object's exact type, or `isinstance(x, SomeType)` to check whether it belongs to a type (including subclasses) — `isinstance()` gets its own dedicated notebook later in this series.

| Type | Purpose | Example |
|---|---|---|
| `int` | whole numbers | `n = 20` |
| `float` | decimal numbers | `n = 20.75` |
| `complex` | real + imaginary parts | `n = 10+20j` |
| `str` | text | `name = 'Jessa'` |
| `bool` | True/False | `flag = True` |
| `list` | mutable ordered sequence | `l = [3, 'a', 2.5]` |
| `tuple` | immutable ordered sequence | `t = (2, 'b', 6.4)` |
| `dict` | key-value pairs | `d = {1:'J', 2:'E'}` |
| `set` | unordered unique values | `s = {1, 3, 5}` |
| `frozenset` | immutable set | `f = frozenset({5,7})` |
| `range` | a sequence of numbers | `r = range(10)` |
| `bytes` | immutable byte sequence | `b = bytes([5,10,15])` |

## `str` — Text

A sequence of characters in single or double quotes. Strings are **immutable** — once created, they can never be changed in place; any 'modification' actually builds a brand-new string.

```python
platform = "PYnative"

print(type(platform))

print(platform)

print(platform[1])   # index 1 - the second character
```

```python
platform = "PYnative"

try:

    platform[0] = 'p'

except TypeError as e:

    print("TypeError:", e)
```

## `int` — Whole Numbers

Positive or negative whole numbers of any length (Python's ints have no fixed maximum size, unlike many other languages). Create one directly, or explicitly via `int()`.

```python
roll_no = 33

print("Roll number is:", roll_no)

print(type(roll_no))



student_id = int(25)

print(student_id, type(student_id))
```

### Non-Decimal Integer Literals

Python also lets you write integer literals directly in binary, octal, or hexadecimal — all three still just become ordinary `int` objects once parsed; the prefix only affects how the literal is *written*, not how it's *stored*.

```python
octal_num = 0o20        # base 8 -> decimal 16

print(octal_num, type(octal_num))



hexadecimal_num = 0x10  # base 16 -> decimal 16

print(hexadecimal_num, type(hexadecimal_num))



binary_num = 0b10000    # base 2 -> decimal 16

print(binary_num, type(binary_num))



print("All three equal 16:", octal_num == hexadecimal_num == binary_num == 16)
```

## `float` — Decimal Numbers

Numbers with a fractional (decimal) part. Create directly, or via `float()`.

```python
salary = 8000.456

print("Salary is:", salary)

print(type(salary))



num = float(54.75)

print(num, type(num))
```

### Scientific (Exponential) Notation

`1.22e4` means 1.22 x 10^4 — a compact way to write very large or very small floats without typing out every digit.

```python
num1 = 1.22e4

print(num1)

print(type(num1))



tiny = 5e-3   # 5 x 10^-3

print(tiny)
```

## `complex` — Real + Imaginary

Written as `a+bj`, where `a` is the real part and `b` is the imaginary part. The imaginary part must be written in plain decimal form — you can't write it as binary/hex/octal.

```python
x = 9 + 8j        # both int

y = 10 + 4.5j      # one int, one float

z = 11.2 + 1.2j    # both float



print(type(x))

print(x, y, z)

print("real:", x.real, "| imag:", x.imag)
```

## `list` — Mutable, Ordered, Allows Duplicates

Represents a group of elements as one entity, preserving insertion order. Lists allow mixed data types and duplicate values, and can be modified after creation. Build one with `[]` or `list()`.

```python
my_list = ["Jessa", "Kelly", 20, 35.75]

print(my_list)

print(type(my_list))



print(my_list[0])         # first element

print(my_list[1:4])       # slice: elements at index 1, 2, 3



my_list[1] = "Emma"       # lists are mutable

print(my_list[1])



my_list2 = list(["Jessa", "Kelly", 20, 35.75])  # via the list() class

print(my_list2)
```

## `tuple` — Immutable, Ordered, Allows Duplicates

Just like a list, except **read-only** once created — think of it as a locked-down list. Build one with `()` or `tuple()`.

```python
my_tuple = (11, 24, 56, 88, 78)

print(my_tuple)

print(type(my_tuple))



print(my_tuple[2])        # 3rd element

print(my_tuple[2:5])      # slice



my_tuple2 = tuple((10, 20, 30, 40))

print(my_tuple2)
```

```python
my_tuple = (11, 24, 56, 88, 78)

try:

    my_tuple[1] = 35

except TypeError as e:

    print("TypeError:", e)
```

## `dict` — Key-Value Pairs

Stores data as key-to-value mappings. Keys must be unique and hashable (immutable types like str, int, or tuple qualify); values can be anything, including duplicates. Assigning to an existing key overwrites its value. Build one with `{}` or `dict()`.

```python
my_dict = {1: "Smith", 2: "Emma", 3: "Jessa"}

print(my_dict)

print(type(my_dict))



my_dict_alt = dict({1: "Smith", 2: "Emma", 3: "Jessa"})

print(my_dict_alt)



print(my_dict[1])       # look up by key



my_dict[1] = "Kelly"    # overwrite the value for an existing key

print(my_dict[1])
```

## `set` — Unordered, Unique Values

A collection of distinct elements with no guaranteed order — duplicates are automatically dropped, and you can't index into a set by position. Mutable: elements can be added/removed after creation. Build one with `{}` (non-empty only — `{}` alone means an empty *dict*, not a set) or `set()`.

```python
my_set = {100, 25.75, "Jessa"}

print(my_set)

print(type(my_set))



my_set = set({100, 25.75, "Jessa"})

print(my_set)



my_set.add(300)

print(my_set)



my_set.remove(100)

print(my_set)
```

### `frozenset` — an Immutable Set

Same semantics as a set, but locked once created — no `.add()`/`.remove()` allowed. Useful whenever you need a set-like object that's hashable (and therefore usable as a dict key, or nested inside another set).

```python
my_set = {11, 44, 75, 89, 56}

print(type(my_set))



f_set = frozenset(my_set)

print(type(f_set))



try:

    f_set.add(99)

except AttributeError as e:

    print("AttributeError:", e)
```

## `bool` — True / False

Represents the two truth values used throughout conditional logic. Any comparison expression evaluates to one of these.

```python
x = 25

y = 20



z = x > y

print(z)

print(type(z))
```

## `bytes` — Immutable Binary Data

Represents a fixed sequence of raw byte values — used for binary data like images, audio, or network payloads. Each value must be an integer from 0 to 255 (256 possible byte values, indices 0-255) — anything outside that range raises `ValueError`.

```python
a = [9, 14, 17, 11, 78]

b = bytes(a)

print(type(b))

print(b[0])

print(b[-1])
```

```python
a = [999, 314, 17, 11, 78]  # 999 and 314 are both out of the valid 0-255 range

try:

    b = bytes(a)

except ValueError as e:

    print("ValueError:", e)
```

### `bytearray` — the Mutable Version of `bytes`

Identical to `bytes`, except its individual elements CAN be modified after creation.

```python
list1 = [9, 17, 11, 78]

b_array = bytearray(list1)

print(b_array)

print(type(b_array))



b_array[1] = 99   # bytearray is mutable, unlike bytes

print(b_array[1])



for i in b_array:

    print(i, end=" ")
```

## `range` — a Lazily-Generated Sequence of Numbers

`range(start, stop, step)` describes a sequence of numbers without actually storing every one of them in memory — it computes each value on demand, which is why iterating a `range` of a billion numbers still uses almost no memory. Very commonly paired with `for` loops.

```python
numbers = range(10, 15, 1)

print(type(numbers))



for i in numbers:

    print(i, end=" ")
```

## `memoryview` — a Zero-Copy View Into Binary Data

`memoryview(obj)` gives you a view onto an object's raw internal buffer WITHOUT copying that data — for large binary blobs, this can save significant memory compared to slicing/copying repeatedly. It works on any object supporting the **buffer protocol**, which `bytes` and `bytearray` both do.

```python
b_array = b'PYnative'

b_array_view = memoryview(b_array)

print("view object:", b_array_view)
```

```python
byte_array = b'PYnative'

byte_array_view = memoryview(byte_array)



# Collect into a container to see the underlying (ordinal/ASCII) values

container = tuple(byte_array_view)

for char in container:

    print(char, end=" ")
```

```python
byte_array = b'PYnative'

byte_array_view = memoryview(byte_array)



print("byte_array_view[0]:", byte_array_view[0])                       # indexing

print("Slicing byte_array_view[6:8]:", bytes(byte_array_view[6:8]))     # slicing, converted back to bytes for display
```

## Summary

- Numeric: `int` (whole numbers, any base), `float` (decimals, supports scientific notation), `complex` (real + imaginary).
- Sequence: `str` (immutable text), `list` (mutable, ordered), `tuple` (immutable, ordered), `range` (lazy number sequence), `bytes`/`bytearray` (binary data, immutable/mutable respectively).
- Set: `set` (mutable, unique, unordered), `frozenset` (immutable version).
- Mapping: `dict` (unique keys, mutable, unordered).
- `bool` for True/False.
- `memoryview` gives a zero-copy window into binary buffer objects.
- `type(x)` reveals the exact type; `isinstance(x, T)` checks membership including subclasses.
