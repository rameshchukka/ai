# Python Variables

A **variable** is a name that refers to a value stored in memory. Once you have a variable, you can read or change what it refers to using that name, rather than juggling raw memory addresses yourself. Since everything in Python is an object, a variable is really just a name pointing at an object.

An object can be **mutable** (its contents can change after creation — like a list) or **immutable** (it can't — like an int, string, or tuple). This distinction matters a lot once you start passing objects around, and comes up again and again through the rest of Python.

This notebook covers creating and reassigning variables, the built-in types you can store, naming rules, multiple assignment, scope, object identity, and unpacking.

## Creating a Variable

Python is **dynamically typed** — you never declare a variable's type up front. Writing `name = value` both creates the variable and assigns its value in one step; the `=` operator's left side is the name, the right side is the value.

```python
name = "John"       # string assignment

age = 25             # integer assignment

salary = 25800.60    # float assignment



print(name)

print(age)

print(salary)
```

## Changing a Variable's Value (and Type)

Unlike statically-typed languages, a Python variable isn't locked to one type for its whole lifetime. Reassigning it to a value of a completely different type is perfectly legal — the name simply starts pointing at a new object.

```python
var = 10

print(var, type(var))



var = 55

print(var, type(var))



var = "Now I'm a string"

print(var, type(var))



var = 35.69

print(var, type(var))
```

## Number Variables: int, float, complex

Python 3 has three built-in numeric types.

```python
# int: whole numbers, positive or negative, no decimal point

age = 28

print(age, type(age))
```

```python
# float: numbers with a decimal point

salary = 10800.55

print(salary, type(salary))
```

```python
# complex: has a real and an imaginary part, written as a+bj

a = 3 + 5j

print(a, type(a))

print("real part:", a.real, "| imaginary part:", a.imag)
```

## String Variables

A string is a sequence of characters, written inside single or double quotes (Python treats them identically). Strings support indexing (`[i]`), slicing (`[start:stop]`), `len()`, and concatenation with `+`.

```python
text = 'PYnative'

print(text)                # full string

print(text[0])              # first character

print(text[2:5])            # characters at index 2, 3, 4

print(len(text))            # length

print(text + "TEST")        # concatenation
```

## List Variables

A list groups multiple values into one ordered, changeable collection, written with square brackets `[]`. Lists preserve insertion order, allow mixed types and duplicates, and can grow or shrink after creation (they're **mutable**).

```python
my_list = ['Jessa', 10, 20, 'Kelly', 50, 10.5]

print(my_list)



print(my_list[0])       # first element

print(my_list[-1])      # last element (negative indexing)

print(my_list[1:3])     # a slice: elements at index 1 and 2



my_list[0] = 'Emma'     # lists are mutable - elements can be reassigned

print(my_list[0])



my_list.append(100)     # grow the list

print(my_list)
```

## Checking a Variable's Type

`type(x)` returns the class an object belongs to. `type(x).__name__` gives just the plain string name of that class, which is often more convenient for display purposes.

```python
a = 100

print(type(a))



b = 100.568

print(type(b))



str1 = "PYnative"

print(type(str1))



my_list = [10, 20, 20.5, 100]

print(type(my_list))



my_set = {'Emma', 'Jessa', 'Kelly'}

print(type(my_set))



my_tuple = (5, 10, 15, 20, 25)

print(type(my_tuple))



my_dict = {1: 'William', 2: 'John'}

print(type(my_dict))
```

```python
my_list = [10, 20, 20.5, 'Python', 100]

# __name__ gives just the plain type name, without the 'class ...' wrapper

print(type(my_list).__name__)
```

## Deleting a Variable

`del` removes the name-to-object binding entirely — once deleted, the name is undefined, exactly as if it had never been assigned.

```python
var1 = 100

print(var1)



del var1

try:

    print(var1)

except NameError as e:

    print("NameError:", e)
```

## Variables Are Case-Sensitive

`a`, `A`, and everything in between are all completely distinct names to Python.

```python
a = 100

A = 200



print(a)

print(A)



a = a + A

print(a)
```

## Constants (by Convention Only)

Python has no true, enforced constant — nothing stops you from reassigning any variable. By strong convention, an ALL_CAPS name signals 'please treat this as a constant, don't reassign it', but Python won't actually stop you if you do. It's purely a social contract between you and other readers of your code.

```python
MAX_VALUE = 500

print(MAX_VALUE)



# Nothing technically stops this reassignment - it's convention, not enforcement:

MAX_VALUE = 999

print(MAX_VALUE, "<- still 'worked', which is exactly the point: Python won't stop you")
```

```python
# A common real pattern: put constants in their own module and import them,

# so every file that needs PI or TOTAL_AREA shares one single source of truth.

import sys, os

constants_code = 'PI = 3.14\nTOTAL_AREA = 205\n'

with open("constants.py", "w") as f:

    f.write(constants_code)

if os.getcwd() not in sys.path:

    sys.path.insert(0, os.getcwd())



import constants

print(constants.PI)

print(constants.TOTAL_AREA)
```

## Naming Rules and Conventions

A name used for a variable, function, class, or module is called an **identifier**. Rules:

1. Made only of letters (a-z, A-Z), digits (0-9), and underscores (`_`).
2. Should be meaningful — `student_name` beats `x`.
3. No special symbols (`$ # @ % ! ~` etc.) — these raise a `SyntaxError`.
4. Can't START with a digit (`1student` is invalid; `student_1` is fine).
5. Case-sensitive — `total`, `Total`, and `TOTAL` are three different names.
6. Constants use ALL_CAPS by convention.
7. Multi-word names typically use an underscore to separate words (`current_temperature`) — this style is called *snake_case*.

```python
try:

    exec("ca$h = 1000")

except SyntaxError as e:

    print("SyntaxError:", e)
```

```python
try:

    exec("1student = 'Jessa'")

except SyntaxError as e:

    print("SyntaxError:", e)
```

```python
total = 120

Total = 130

TOTAL = 150

print(total)

print(Total)

print(TOTAL)
```

## Multiple Assignment

### One value, many variables
Chain the `=` operator to assign the same value to several names at once.

```python
a = b = c = 10

print(a, b, c)
```

### Many values, many variables

Assign several distinct values in one line — the count on each side must match exactly.

```python
roll_no, marks, name = 10, 70, "Jessa"

print(roll_no, marks, name)
```

## Variable Scope

**Scope** determines where in your code a variable is visible/accessible.

- A **local** variable is created inside a function and only exists within that function — trying to read it from outside (even from another function) raises `NameError`.
- A **global** variable is defined outside any function, at the top level of a file — it's readable from anywhere in that file, including inside functions.

```python
def test1():

    price = 900  # local to test1 only

    print("Value of price in test1 function:", price)



def test2():

    try:

        print("Value price in test2 function:", price)

    except NameError as e:

        print("NameError:", e)



test1()

test2()
```

```python
price = 900  # global variable - defined outside any function



def test1():

    print("price in 1st function:", price)



def test2():

    print("price in 2nd function:", price)



test1()

test2()
```

## Object Identity and References

Every object Python creates gets a unique **identity** — internally, its memory address — retrievable with `id()`. Two names that refer to the exact same object return the exact same `id()`.

When you write `n = 300` then `m = n`, no new object is created for `m` — it's simply a second name pointing at the SAME object `n` already refers to.

```python
n = 300

m = n

print(id(n))

print(id(m))

print(id(n) == id(m))
```

```python
m = 500

n = 400

print("Memory address of m:", id(m))

print("Memory address of n:", id(n))

print("Same object?", id(m) == id(n))
```

### Small-Integer Caching

As a CPython implementation detail (not a language guarantee), small integers (roughly -5 to 256) are pre-created once at startup and reused everywhere, rather than a fresh object being built every time. This is why several variables holding the same small int often report the identical `id()` — Python is quietly reusing one shared object instead of creating duplicates.

```python
a = 10

b = 10

print(id(a))

print(id(b))

print("a is b:", a is b)



c = 20

d = 20

e = 20

print(id(c))

print(id(d))

print(id(e))

print("c is d is e:", c is d is e)
```

### A Note on `sys.getrefcount()`

`sys.getrefcount(obj)` reports how many references currently point to an object. It's a genuinely useful diagnostic tool, but a common point of confusion: the number it reports is almost never the small number you'd naively expect from counting just your own variable names.

Two separate reasons for that:
1. **Calling `getrefcount(a)` itself temporarily creates one more reference** (as `a` gets passed into the function as an argument).
2. **Since Python 3.12, frequently-used objects like small integers, `None`, and `True`/`False` are "immortalized"** (PEP 683) — instead of tracking a real reference count that goes up and down, CPython gives them a special sentinel value (`2**32 - 1`) meaning 'never garbage-collect this, ever'. That's exactly the huge number you'll see below if you're running Python 3.12 or later — it isn't a bug or an error, it's a real, intentional optimization: these objects are so commonly created and destroyed that CPython simply stopped bothering to count references to them at all.

Either way, don't expect a clean '2' just because two of your own variables point to the same small int — the exact number is CPython-version-dependent internal detail, not something to rely on in portable code.

```python
import sys



a = 10

b = 10

print(f"Python version: {sys.version_info.major}.{sys.version_info.minor}")

print("Reported refcount for a:", sys.getrefcount(a))

print("(A huge sentinel value here means this int is 'immortal' - see note above. " \

      "On Python < 3.12 you'd instead see a smaller, ordinary-looking number.)")
```

## Unpacking a Collection Into Variables

**Packing**: writing several values separated by commas (with or without visible parentheses) bundles them into a tuple automatically.

**Unpacking**: the reverse — spreading a tuple's (or list's) values out into separate variables in one assignment. The number of variables on the left must exactly match the number of values on the right, or you get a `ValueError`.

```python
a = 10

b = 20

c = 20

d = 40

tuple1 = a, b, c, d   # packing

print(tuple1)
```

```python
tuple1 = (10, 20, 30, 40)

a, b, c, d = tuple1   # unpacking

print(a, b, c, d)
```

```python
try:

    a, b = 1, 2, 3

except ValueError as e:

    print("ValueError:", e)
```

## Summary

- Variables are created by simple assignment (`x = value`) — no separate declaration step, and no fixed type for the variable's whole lifetime.
- `type(x)` reports an object's type; `del x` removes the name entirely.
- Names are case-sensitive, must start with a letter or underscore, and can't contain special symbols.
- Constants are convention-only (ALL_CAPS) — Python enforces nothing.
- Multiple assignment supports both 'one value, many names' (`a = b = c = 10`) and 'many values, many names' (`a, b = 1, 2`).
- Scope: local variables only exist inside the function that created them; global variables are visible everywhere in the file.
- `id()` reveals an object's identity; equal small integers are often (but not guaranteedly, outside CPython) the exact same cached object.
- Packing/unpacking lets you bundle and spread values via tuples in one clean line.
