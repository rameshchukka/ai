## Concepts you'll need

This set of beginner exercises draws on a handful of core Python building blocks. A quick refresher before you dive in:

- **Variables & arithmetic** — storing values and combining them with `+ - * / // % **`.
- **Conditional logic** — `if` / `elif` / `else` branch the program based on a condition; comparisons like `<=`, `==`, `!=` produce `True`/`False`.
- **`for` and `while` loops** — `for` walks over a known sequence (e.g. `range(n)`); `while` repeats as long as a condition holds. Loops can be nested (a loop inside another loop) to build grids or patterns.
- **Strings as sequences** — strings support indexing (`s[0]`), slicing (`s[start:stop:step]`), and methods like `.lower()`, `.replace()`, `.split()`, `.count()`, `.isdigit()`.
- **Lists** — ordered, mutable collections; `.append()`, `.pop()`, `.count()`, indexing, and slicing all apply.
- **Dictionaries** — key/value stores; look up or set a value with `d[key]`, merge two dicts with `|` (Python 3.9+) or `.update()`.
- **Sets** — unordered collections of unique values, useful for de-duplication and operations like intersection (`&`).
- **Functions** — `def name(params):` bundles reusable logic; `return` sends a value back to the caller.
- **File I/O** — `with open(path, mode) as f:` opens a file safely and closes it automatically; `"w"` writes, `"r"` reads.
- **Classes (OOP)** — `class Name:` defines a blueprint; `__init__` sets up a new object's attributes, and `self` refers to the specific object a method is running on.

## Exercise 1. Product or Sum Based on a Threshold

**Concept:** if/else conditional branching, functions

**Problem:** Write a function taking two integers. If their product is 1000 or less, return the product; otherwise return their sum.

**Given:**
```
number1 = 20, number2 = 30  |  number1 = 40, number2 = 30
```

**Expected Output:**
```
The result is 600
The result is 70
```

**Hint:** Compute the product first, then use an if statement to decide whether to return the product or the sum.

````python
def multiplication_or_sum(num1, num2):
    product = num1 * num2
    if product <= 1000:
        return product
    else:
        return num1 + num2

result = multiplication_or_sum(20, 30)
print("The result is", result)

result = multiplication_or_sum(40, 30)
print("The result is", result)
````

**Explanation:** The if statement compares the computed product to the threshold and picks which value to return. Returning (rather than printing) inside the function lets the caller decide what to do with the result.

## Exercise 2. Track the Previous Value While Looping

**Concept:** state tracking across loop iterations

**Problem:** Loop through numbers 0-9. On each pass, show the current number, the previous number, and their sum.

**Given:**
```
range(10)
```

**Expected Output:**
```
Current Number 0 Previous Number 0 Sum: 0
Current Number 1 Previous Number 0 Sum: 1
...
Current Number 9 Previous Number 8 Sum: 17
```

**Hint:** Keep a variable outside the loop that remembers the previous value, and update it at the end of every iteration.

````python
previous_num = 0

for i in range(10):
    total = previous_num + i
    print(f"Current Number {i} Previous Number {previous_num} Sum: {total}")
    previous_num = i
````

**Explanation:** previous_num is declared outside the loop so its value survives between iterations. Updating it as the last step of each pass 'carries' the current value forward to be used as the previous value next time.

## Exercise 4. Remove Leading Characters

**Concept:** string slicing, functions

**Problem:** Write a function that strips off the first n characters of a string and returns what remains.

```

**Expected Output:**
```

**Hint:** Slicing a string as string[n:] keeps everything from index n onward.

````python
def remove_chars(word, n):
    print("Original string:", word)
    return word[n:]

print(remove_chars("pynative", 4))
print(remove_chars("pynative", 2))
````

## Exercise 12. Compare First and Last List Elements

**Concept:** negative indexing, boolean logic

**Problem:** Write a function that returns True if a list's first and last elements are equal, and False otherwise.

**Given:**
```
numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]
```

**Expected Output:**
```
Given list: [75, 65, 35, 75, 30] | result is False
Given list: [10, 20, 30, 40, 10] | result is True
```

**Hint:** Index 0 gives the first item; index -1 always gives the last item, no matter the list's length.

````python
def first_last_same(number_list):
    print("Given list:", number_list)
    first_num = number_list[0]
    last_num = number_list[-1]
    return first_num == last_num

numbers_x = [10, 20, 30, 40, 10]
print("result is", first_last_same(numbers_x))

numbers_y = [75, 65, 35, 75, 30]
print("result is", first_last_same(numbers_y))
````

**Explanation:** Negative indices count from the end, so [-1] is a safe shortcut for 'last item' without needing len(list) - 1. == compares the two values and yields a boolean directly.

## Exercise 16. Check if a Number Is a Palindrome

**Concept:** type casting, string reversal

**Problem:** Determine whether a number reads the same forwards and backwards.

**Given:**
```
number = 121; number = 125
```

**Expected Output:**
```
Number 125 is not palindrome number
Number 121 is palindrome number
```

**Hint:** Convert the number to a string, reverse it with slicing, and compare the two strings.

````python
def check_palindrome(number):
    print("original number", number)
    original_str = str(number)
    reversed_str = original_str[::-1]
    if original_str == reversed_str:
        print("Yes. given number is palindrome number")
    else:
        print("No. given number is not palindrome number")

check_palindrome(121)
check_palindrome(125)
````

**Explanation:** Converting the number to a string lets us reuse the familiar [::-1] reversal trick. Comparing the original and reversed strings with == tells us if the digits are symmetric.

## Exercise 17. Merge Lists by Parity

**Concept:** loop filtering, list building

**Problem:** Build a new list containing the odd numbers from one list and the even numbers from another.

**Given:**
```
list1 = [10, 20, 25, 30, 35]
list2 = [40, 45, 60, 75, 90]
```

**Expected Output:**
```
[25, 35, 40, 60, 90]
```

**Hint:** Loop through the first list keeping odd numbers, then loop through the second keeping even numbers, appending both into one result list.

````python
def merge_list(list1, list2):
    result_list = []
    for num in list1:
        if num % 2 != 0:
            result_list.append(num)
    for num in list2:
        if num % 2 == 0:
            result_list.append(num)
    return result_list

list1 = [10, 20, 25, 30, 35]
list2 = [40, 45, 60, 75, 90]
print("result list:", merge_list(list1, list2))
````

**Explanation:** Two separate loops apply two different filters — 'not divisible by 2' for oddness on the first list, and 'divisible by 2' for evenness on the second — appending matches into a shared result list in order.

## Exercise 19. Tiered Income Tax Calculation

**Concept:** cumulative conditional logic

**Problem:** Calculate tax owed given tiered brackets: 0% on the first 10,000, 10% on the next 10,000, and 20% on anything beyond that.

**Given:**
```
income = 45000
```

**Expected Output:**
```
Total income tax to pay is 6000
```

**Hint:** Work through the brackets in order, taxing only the portion of income that falls inside each one.

````python
income = 45000
tax_payable = 0
print("Given income:", income)

if income <= 10000:
    tax_payable = 0
elif income <= 20000:
    tax_payable = (income - 10000) * 10 / 100
else:
    tax_payable = 0 + (10000 * 10 / 100)
    tax_payable += (income - 20000) * 20 / 100

print("Total income tax to pay is", tax_payable)
````

**Explanation:** Instead of applying one flat rate to the whole income, the code taxes each bracket separately — the first 10,000 stays untaxed, the next 10,000 is taxed at 10%, and everything above 20,000 is taxed at 20%.

## Exercise 22. Custom Power Function

**Concept:** accumulator pattern with while loop

**Problem:** Write your own exponent(base, exp) function that computes base raised to exp without using **.

**Given:**
```
base = 2, exp = 5
```

**Expected Output:**
```
2 raises to the power of 5 is: 32
```

**Hint:** Start a result at 1 and multiply it by base, exp times.

````python
def exponent(base, exp):
    num = exp
    result = 1
    while num > 0:
        result = result * base
        num = num - 1
    print(base, "raises to the power of", exp, "is:", result)

exponent(2, 5)
exponent(5, 4)
````

**Explanation:** result starts at 1 (the multiplicative identity), and the while loop multiplies it by base once for every unit of exp, counting down num until the loop naturally ends.

## Exercise 24. Fibonacci Series

## Exercise 25. Leap Year Check

**Concept:** compound conditional logic

**Problem:** Determine whether a given year is a leap year, following the divisible-by-4-except-century-years-unless-divisible-by-400 rule.

**Given:**
```
year = 2024
```

**Expected Output:**
```
2024 is a leap year
```

**Hint:** Combine three conditions: divisible by 4 AND not divisible by 100, OR divisible by 400.

````python
def is_leap(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(f"{year} is a leap year")
    else:
        print(f"{year} is not a leap year")

is_leap(2024)
is_leap(2100)
````

**Explanation:** The first clause handles the ordinary 4-year rule while excluding century years like 1900 or 2100. The or (year % 400 == 0) clause restores the exception for years like 2000, which are leap years despite being divisible by 100.

## Exercise 30. Word Frequency Counter

**Concept:** dictionaries for counting

**Problem:** Count how many times each word appears in a piece of text.

```

```
{'apple': 3, 'banana': 2, 'cherry': 1}

**Hint:** Split the text into words, then use a dictionary to track how many times each word has been seen.

## Exercise 31. Every Other Prime Number

**Concept:** nested loops with for/else, slicing

**Problem:** Find the primes up to 20, then print only every second one.

**Given:**
```
limit = 20
```

**Expected Output:**
```
[2, 5, 11, 17]
```

**Hint:** Build the full list of primes first, then slice it with a step of 2.

````python
primes = []

for num in range(2, 21):
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            break
    else:
        primes.append(num)

alternate_primes = primes[::2]
print(alternate_primes)
````

**Explanation:** The inner loop's else clause runs only if no divisor broke the loop, which is the classic way to detect a prime. Checking factors only up to the square root keeps this efficient. The final [::2] then keeps every other prime.

## Exercise 34. Descending Number Pattern

**Concept:** nested loops with reverse ranges

**Problem:** Print a pattern where each row counts down from a shrinking starting point.

**Given:**
```
rows = 5
```

**Expected Output:**
```
5 4 3 2 1
4 3 2 1
3 2 1
2 1
1
```

**Hint:** The outer loop runs from 5 down to 1; the inner loop counts down from the outer loop's current value to 1.

````python
rows = 5
for i in range(rows, 0, -1):
    for j in range(i, 0, -1):
        print(j, end=' ')
    print("")
````

**Explanation:** Both loops count backwards, using a step of -1. The inner loop's starting point tracks the outer loop's current row number, which is why each row is one number shorter than the last.

**Hint:** Split the sentence into words, capitalize each one, then rejoin them with spaces.

## Exercise 37. Countdown Timer

**Concept:** while loop, time.sleep

**Problem:** Count down from a starting number to zero, pausing between each number.

```

```

**Expected Output:**
```
5
4
3
2
1
Blast off!
```

**Hint:** Loop while the counter is above 0, printing and then decrementing it each time.

````python
import time

count = 5

while count > 0:
    print(count)
    time.sleep(1)
    count -= 1

print("Blast off!")
````

**Explanation:** The while loop keeps running exactly as long as count > 0 holds. time.sleep(1) pauses execution for a second between prints, and count -= 1 is what eventually makes the loop's condition false.

## Exercise 38. Write and Read a Text File

**Concept:** file I/O with context managers

**Problem:** Create a text file, write three lines to it, then read the file back and print its contents.

**Given:**
```
3 lines of text
```

**Expected Output:**
```
Hello, this is my first note.
Python file handling is simple.
End of file.
```

**Hint:** Use `with open(path, mode) as f:` for both writing ("w") and reading ("r").

````python
with open("notes.txt", "w") as file:
    file.write("Hello, this is my first note.\n")
    file.write("Python file handling is simple.\n")
    file.write("End of file.\n")

print("Reading file contents:")
with open("notes.txt", "r") as file:
    content = file.read()
    print(content)
````

**Explanation:** The with statement is a context manager — it guarantees the file gets closed automatically, even if an error occurs. \n inserts line breaks so the three sentences don't run together. "w" mode writes (overwriting any existing file), while "r" mode reads.

## Exercise 39. Count Words in a File

**Concept:** file reading, split, error handling

**Problem:** Open a text file and count how many words it contains.

**Given:**
```
sample.txt containing "Coding is the language of the future."
```

**Expected Output:**
```
The file contains 7 words.
```

**Hint:** Read the file's contents into a string, split it into words, and take the length of the resulting list.

````python
try:
    with open("sample.txt", "r") as file:
        data = file.read()
        words = data.split()
        word_count = len(words)
        print(f"The file contains {word_count} words.")
except FileNotFoundError:
    print("Error: The file 'sample.txt' was not found.")
````

**Explanation:** .split() breaks the text apart wherever there's whitespace, producing one list entry per word; its length is the word count. Wrapping the file access in try/except means a missing file produces a friendly message instead of crashing the program.

## Exercise 40. Define a Simple Class

**Concept:** object-oriented programming: classes, __init__, self

**Problem:** Create a Car class with make, model, and year attributes, plus a method that announces the engine starting.

**Given:**
```
make="Toyota", model="Camry", year=2022
```

**Expected Output:**
```
The 2022 Toyota Camry's engine is now running!
```

**Hint:** Use __init__ to store the attributes on creation, and self to refer to the specific object inside methods.

````python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print(f"The {self.year} {self.make} {self.model}'s engine is now running!")

my_car = Car("Toyota", "Camry", 2022)
my_car.start_engine()
````

**Explanation:** __init__ runs automatically whenever a new Car is created, storing the given values as attributes on that specific object. self inside a method always refers back to that same object, so start_engine() can read the attributes that were set when it was built.
