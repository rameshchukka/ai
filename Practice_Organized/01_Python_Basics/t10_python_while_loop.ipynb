# Python `while` Loop

A `while` loop repeats a block of code for as long as a condition stays `True`, checking that condition fresh at the START of every single pass. The moment the condition becomes `False`, the loop stops immediately and execution continues at the next statement after it.

```
while condition:
    # block of statement(s)
```

Use `while` specifically when you DON'T know ahead of time how many iterations you'll need — this is called **indefinite iteration**, in contrast to `for`'s **definite iteration** (a known, fixed count). Classic example: repeatedly asking a user for input until they enter something valid — you genuinely don't know if that will take 1 try or 20.

while loop flowchart

````python
count = 1

# run loop till count is less than 3

while count < 3:

    print(count)

    count = count + 1
````

## A Second, Slightly Bigger Example

Print numbers less than 5. The loop body must eventually change something the condition depends on — here, incrementing `count` — or the loop would run forever (see 'Infinite Loops' below).

````python
count = 1

while count < 5:

    print(count)

    count = count + 1
````

### Why `while` Fits Better Here Than `for`

Check how many times a number can be divided by 3 before it drops to 10 or below. The exact iteration count genuinely depends on the starting number — you couldn't easily write this with `range()` up front.

````python
count = 0

number = 180

while number > 10:

    number = number / 3

    count = count + 1

print('Total iteration required', count)
````

## Why and When to Use `while`

- **Automates repetition** — same benefit as `for`.
- **Indefinite iteration** — the natural choice whenever the iteration count truly can't be known ahead of time.
- **Reduces complexity** — one block of logic instead of manual repetition.
- **Watch out for infinite loops** — if nothing inside the body ever changes the condition, the loop runs forever.

````python
# Simulates repeatedly asking: number = int(input('Enter any number between 100 and 500 '))

# until a valid value is entered.

import builtins

_inputs = iter(["700", "98", "300"])

_orig_input = builtins.input

def _fake_input(prompt=""):

    val = next(_inputs)

    print(f"{prompt}{val}")

    return val

builtins.input = _fake_input



number = int(input('Enter any number between 100 and 500 '))

while number < 100 or number > 500:

    print('Incorrect number, Please enter correct number:')

    number = int(input('Enter a Number between 100 and 500 '))

else:

    print("Given Number is correct", number)



builtins.input = _orig_input
````

### Infinite Loops

`while True:` never becomes false on its own — a deliberate infinite loop, normally paired with an internal `break` to exit under some condition. Shown here with a hard iteration cap so this cell terminates safely inside a notebook, since an unbounded infinite loop would freeze execution.

````python
# Illustrating the idea of `while True:` safely, capped at 3 iterations for this notebook

# (a genuine `while True: print('Hello')` with no break would never stop on its own)

counter = 0

while True:

    print('Hello')

    counter += 1

    if counter >= 3:

        break
````

## `if-else` Inside a `while` Loop

Combine a condition-driven loop with per-iteration branching — checked fresh every pass, just like inside a `for` loop.

````python
# Simulates: n = int(input('Please Enter Number '))

n = 7

while n > 0:

    if n % 2 == 0:

        print(n, 'is a even number')

    else:

        print(n, 'is a odd number')

    n = n - 1
````

## Transfer Statements in `while` Loops

### `break` — stop the loop entirely

````python
name = 'Jesaa29Roy'

size = len(name)

i = 0

while i < size:

    if name[i].isdecimal():

        break

    print(name[i], end=' ')

    i = i + 1
````

### `continue` — skip to the next iteration

````python
name = 'Jesaa29Roy'

size = len(name)

i = -1

# starting at -1 (not 0) because the FIRST statement inside the loop increments i

# to 0 before it's ever used - this pattern lets the increment happen once, up front,

# on every pass, including the very first one

while i < size - 1:

    i = i + 1

    if not name[i].isalpha():

        continue

    print(name[i], end=' ')
````

### `pass` — a deliberate no-op

````python
n = 4

while n > 0:

    n = n - 1

    pass

print("Done, n is now", n)
````

## Nested `while` Loops

A `while` loop inside another `while` loop — same principle as nested `for` loops: the inner loop runs to completion for every single pass of the outer loop.

````python
i = 1

# outer while loop - 4 rows

while i < 5:

    j = 0

    # nested (inner) while loop

    while j < i:

        print('*', end=' ')

        j = j + 1

    print('')

    i = i + 1
````

### `for` Loop Inside a `while` Loop

Loop types nest freely with each other, not just with themselves.

````python
i = 1

while i < 5:

    for j in range(1, i + 1):

        print("*", end=" ")

    print('')

    i = i + 1
````

## `else` Block in a `while` Loop

Just like `for`, a `while` loop can have an `else` clause — it runs only if the loop finished NORMALLY (the condition simply became false), and is SKIPPED if the loop was cut short by `break`.

````python
i = 1

while i <= 5:

    print(i)

    i = i + 1

else:

    print("Done. while loop executed normally")
````

````python
i = 1

while i <= 5:

    print(i)

    if i == 3:

        break

    i = i + 1

else:

    print("Done. while loop executed normally")  # skipped - break fired
````

## Reverse `while` Loop

Simply start high and decrement, instead of starting low and incrementing.

````python
i = 10

while i >= 0:

    print(i, end=' ')

    i = i - 1
````

## Iterating a String with `while`

Less idiomatic than `for char in string:`, but useful when you need explicit control over the index (like stepping by more than one, or stopping/restarting based on complex conditions).

````python
name = "Jessa"

i = 0

res = len(name) - 1

while i <= res:

    print(name[i])

    i = i + 1
````

## Iterating a List with `while`

Same idea — manual index tracking, useful mainly when the iteration logic is more complex than a plain `for` can express cleanly.

````python
numbers = [1, 2, 4, 5, 7]

size = len(numbers)

i = 0

while i < size:

    print(numbers[i])

    i = i + 1
````

## `for` vs. `while`: When to Use Which

| | `for` | `while` |
|---|---|---|
| Best for | a KNOWN or determinable number of iterations | an UNKNOWN number of iterations |
| Typical source | a sequence's own length | an external condition (user input, a search, a threshold) |
| Risk | rare — sequences are always finite | infinite loop if the condition never changes |
| Example | print every item in a list | keep asking until the user enters valid input |

## Summary

- `while condition:` repeats as long as the condition is `True`, re-checked at the start of every pass.
- Use `while` for INDEFINITE iteration (unknown count ahead of time); use `for` for DEFINITE iteration (known count).
- The loop body must eventually make the condition false, or you get an infinite loop — `while True:` is a deliberate version of this, normally paired with an internal `break`.
- `break`, `continue`, `pass`, nesting, and `else` all work in `while` loops exactly as they do in `for` loops.
- Manual index-based iteration (`while i < len(x):`) works for strings/lists but is usually less idiomatic than `for` — reach for `while` when the STOPPING CONDITION, not the sequence, is what actually drives the loop.

## Practice Exercises

1. Write a `while` loop that keeps doubling a starting value of 1 until it exceeds 1000, printing each doubled value and the final count of doublings.
2. Using `while` and `break`, search a list for the first negative number and print it (or print `"no negative numbers"` if none exist, using `while...else`).
3. Write a `while` loop that sums digits of a number repeatedly until a single digit remains (the 'digital root'), e.g. 9875 -> 9+8+7+5=29 -> 2+9=11 -> 1+1=2.
4. Simulate a simple countdown timer: starting from 5, print each number then `"Liftoff!"` at 0, using `while`.
5. Write a nested `while` loop that prints a multiplication table (1 through 5) as a grid, five rows by five columns.

### Exercise 1 solution

````python
value = 1

count = 0

while value <= 1000:

    print(value, end=" ")

    value *= 2

    count += 1

print(f"\nDoubled {count} times to exceed 1000 (final value: {value})")
````

### Exercise 2 solution

````python
def find_first_negative(items):

    i = 0

    while i < len(items):

        if items[i] < 0:

            print("First negative:", items[i])

            break

        i += 1

    else:

        print("no negative numbers")



find_first_negative([4, 8, -3, 10])

find_first_negative([4, 8, 3, 10])
````

### Exercise 3 solution

````python
def digital_root(n):

    while n >= 10:

        total = 0

        while n > 0:

            total += n % 10

            n //= 10

        n = total

    return n



print(digital_root(9875))
````

### Exercise 4 solution

````python
count = 5

while count >= 0:

    if count == 0:

        print("Liftoff!")

    else:

        print(count)

    count -= 1
````

### Exercise 5 solution

````python
i = 1

while i <= 5:

    j = 1

    while j <= 5:

        print(i * j, end="\t")

        j += 1

    print()

    i += 1
````
