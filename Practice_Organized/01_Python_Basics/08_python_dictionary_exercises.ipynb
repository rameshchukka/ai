## Concepts you'll need

This set covers **dictionaries** — Python's key-value data structure — from basics through nesting, sorting, merging, and inversion.

- **Basic operations** — `d[key] = value` adds or updates; `d.pop(key)` removes and returns a value; `d.get(key, default)` reads safely without risking a `KeyError`.
- **Views** — `.keys()`, `.values()`, and `.items()` return live "view" objects, not copies; `.keys()` and `.items()` even support set operations (`& | -`) for comparing two dictionaries.
- **Construction shortcuts** — `dict(zip(keys, values))` pairs two lists; `dict.fromkeys(keys, default)` gives every key the same starting value; `dict(list_of_tuples)` converts pairs directly.
- **Comprehensions** — `{k: v for k, v in d.items() if condition}` filters; `{v: k for k, v in d.items()}` inverts; these are the standard, Pythonic way to transform a dictionary into a new one.
- **Sorting** — dictionaries don't sort themselves, but `sorted(d.items(), key=lambda item: ...)` produces sorted tuples that can be fed back into `dict()` to rebuild an ordered dictionary (Python 3.7+ preserves insertion order).
- **Nesting** — chained bracket access (`d["a"]["b"]`) reads or writes into dictionaries stored inside other dictionaries; recursion is the natural tool for arbitrarily deep structures (like flattening).
- **`.setdefault(key, default)`** — a one-line way to "insert the default if missing, then use whatever's there," commonly used to build grouped lists.
- **Copying** — `.copy()` makes a *shallow* copy (nested mutable objects are still shared); `copy.deepcopy()` makes every nested level independent.

Each exercise below gives a problem, a hint, a solution, and an explanation.

## Exercise 1. Add, Modify, and Access

**Concept:** basic dict operations

**Problem:** Add a new key, modify an existing key's value, and read a value from a dictionary.

**Given:**
```
student = {"name": "Alice", "age": 20, "grade": "B"}
```

**Expected Output:**
```
{'name': 'Alice', 'age': 21, 'grade': 'B', 'city': 'New York'}
Name: Alice
```

**Hint:** Assigning to a new key (dict['new'] = value) adds it; assigning to an existing key overwrites it.

````python
student = {"name": "Alice", "age": 20, "grade": "B"}

student["city"] = "New York"
student["age"] = 21

print(student)
print("Name:", student["name"])
````

**Explanation:** Assigning to a bracketed key that doesn't exist yet creates it (student["city"]); assigning to one that already exists overwrites its value in place, since dictionary keys are unique.

## Exercise 2. Remove, List Pairs, Check Existence

**Concept:** pop(), items(), the in operator

**Problem:** Remove a key, retrieve all key-value pairs, and check whether specific keys exist.

**Given:**
```
car = {"brand": "Toyota", "model": "Camry", "year": 2022, "color": "blue"}
```

**Expected Output:**
```
{'brand': 'Toyota', 'model': 'Camry', 'year': 2022}
'brand' exists: True
'color' exists: False
```

**Hint:** pop() removes and returns a value; items() gives (key, value) tuples; `in` checks keys by default.

````python
car = {"brand": "Toyota", "model": "Camry", "year": 2022, "color": "blue"}

car.pop("color")
print(car)

print(car.items())

print("'brand' exists:", "brand" in car)
print("'color' exists:", "color" in car)
````

**Explanation:** pop() deletes the given key and hands back its value in one step. items() returns a live view of (key, value) tuples. The `in` operator checks against the dictionary's keys, not its values.

## Exercise 3. Build a Dict From Two Lists

**Concept:** zip() + dict()

**Hint:** zip(keys, values) pairs them positionally; pass that straight into dict().

## Exercise 5. Merge Two Dictionaries

**Concept:** update(), unpacking, and the | operator

**Problem:** Combine two dictionaries, with the second dictionary's values winning on any shared key.

**Given:**
```
dict1 = {"a": 1, "b": 2}, dict2 = {"b": 3, "c": 4}
```

**Expected Output:**
```
{'a': 1, 'b': 3, 'c': 4}
```

**Hint:** dict1.update(dict2) merges in place; {**dict1, **dict2} or dict1 | dict2 both build a new merged dictionary.

````python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

dict1.update(dict2)
print(dict1)

dict1 = {"a": 1, "b": 2}
merged = {**dict1, **dict2}
print(merged)

dict1 = {"a": 1, "b": 2}
merged = dict1 | dict2
print(merged)
````

**Explanation:** All three approaches resolve overlapping keys the same way: whichever dictionary's value is applied 'later' (dict2 in every case here) wins. update() mutates dict1 directly, while unpacking and | both build fresh dictionaries, leaving the originals untouched.

## Exercise 10. Delete a List of Keys

**Concept:** looping pop(key, None) for safe bulk removal

**Problem:** Remove several specific keys from a dictionary at once.

**Given:**
```
product = {"id": 101, "name": "Laptop", "price": 999, "stock": 50, "warehouse": "A3"}, remove ["stock", "warehouse"]
```

**Expected Output:**
```
{'id': 101, 'name': 'Laptop', 'price': 999}
```

**Hint:** pop(key, None) removes a key if present and does nothing harmful if it's missing.

````python
product = {"id": 101, "name": "Laptop", "price": 999, "stock": 50, "warehouse": "A3"}
keys_to_remove = ["stock", "warehouse"]

for key in keys_to_remove:
    product.pop(key, None)

print(product)
````

**Explanation:** Looping the removal list lets the same logic handle any number of keys. Using pop(key, None) instead of del means a key that's already missing from the dictionary won't raise a KeyError and crash the loop.

## Exercise 15. Character Frequency Count

**Concept:** dict.get() as a safe counter

**Problem:** Count how many times each character (including spaces) appears in a string.

**Given:**
```
text = "hello world"
```

**Expected Output:**
```
{'h': 1, 'e': 1, 'l': 3, 'o': 2, ' ': 1, 'w': 1, 'r': 1, 'd': 1}
```

**Hint:** freq.get(char, 0) + 1 increments safely whether or not the character has been seen before.

````python
text = "hello world"

freq = {}
for char in text:
    freq[char] = freq.get(char, 0) + 1

print(freq)
````

**Explanation:** Every character in the string is counted, including the space — nothing is filtered out. .get(char, 0) returns the running count if the character has appeared before, or 0 for a brand-new character, so + 1 always works safely.

## Exercise 21. Key With the Maximum Value (and Ties)

**Concept:** max(dict, key=dict.get), then finding all ties

**Problem:** Find the top-scoring key, and separately find every key that's tied for the top score.

**Given:**
```
scores = {"Alice": 88, "Bob": 95, "Carol": 72, "Dave": 95, "Eve": 84}
```

**Expected Output:**
```
Top scorer: Bob
All top scorers: ['Bob', 'Dave']
```

**Hint:** max(dict, key=dict.get) only returns the first tied key; a list comprehension against the max value finds them all.

````python
scores = {"Alice": 88, "Bob": 95, "Carol": 72, "Dave": 95, "Eve": 84}

top = max(scores, key=scores.get)
print("Top scorer:", top)

best_score = max(scores.values())
all_top = [k for k, v in scores.items() if v == best_score]
print("All top scorers:", all_top)
````

**Explanation:** max(scores, key=scores.get) returns only the first key it finds with the top value — here 'Bob' beats 'Dave' just by appearing first. To catch every tied key, the second approach first computes the actual maximum value, then filters scores.items() for every pair matching it.

## Exercise 25. Dictionary Intersection by Value

**Concept:** combining key intersection with a value-match filter

**Problem:** Build a new dictionary containing only the entries where both dictionaries agree on the key AND the value.

**Given:**
```
d1 = {"a": 1, "b": 2, "c": 3}, d2 = {"a": 1, "b": 99, "c": 3}
```

**Expected Output:**
```
Intersection: {'a': 1, 'c': 3}
```

**Hint:** Get the shared keys first with &, then filter to only those where d1[k] == d2[k].

````python
d1 = {"a": 1, "b": 2, "c": 3}
d2 = {"a": 1, "b": 99, "c": 3}

intersection = {k: d1[k] for k in d1.keys() & d2.keys() if d1[k] == d2[k]}

print("Intersection:", intersection)
````

## Exercise 26. Case-Insensitive Word Count

**Concept:** lower() + split() + dict.get()

**Problem:** Count word occurrences in a sentence, treating different capitalizations as the same word.

**Given:**
```
text = "the cat sat on the mat the cat"
```

**Expected Output:**
```
{'the': 3, 'cat': 2, 'sat': 1, 'on': 1, 'mat': 1}
```

**Hint:** Lowercase the whole string before splitting, so 'The' and 'the' become identical tokens.

````python
text = "the cat sat on the mat the cat"

word_count = {}
for word in text.lower().split():
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
````

**Explanation:** .lower() normalizes case before .split() breaks the sentence into individual words. The same get(word, 0) + 1 counting pattern seen in earlier character-counting exercises applies here at the word level instead.

## Exercise 29. Sort a Dictionary by Its Values

## Exercise 30. Check if All Values Are Unique

**Concept:** comparing lengths of a list and its set

**Problem:** Determine whether a dictionary's values contain any duplicates.

**Given:**
```
data = {"a": 1, "b": 2, "c": 3, "d": 2}
```

**Expected Output:**
```
False
```

**Hint:** Converting values to a set drops duplicates; compare its length to the original count.

````python
data = {"a": 1, "b": 2, "c": 3, "d": 2}

values = list(data.values())
is_unique = len(values) == len(set(values))

print(is_unique)
````

**Explanation:** set(values) keeps only distinct values, so if any value repeated, the set will be shorter than the original list. Here [1, 2, 3, 2] becomes {1, 2, 3} — a length mismatch (4 vs 3) reveals a duplicate exists, so the result is False.

## Exercise 31. Check if One Dictionary Is a Subset of Another

**Concept:** the <= operator on dict.items() views

**Problem:** Verify that every key-value pair in a smaller dictionary also exists in a larger one.

**Given:**
```
main = {"a": 1, "b": 2, "c": 3, "d": 4}, subset = {"a": 1, "c": 3}
```

**Expected Output:**
```
True
```

**Hint:** subset.items() <= main.items() checks the subset relationship directly on dict item views.

````python
main = {"a": 1, "b": 2, "c": 3, "d": 4}
subset = {"a": 1, "c": 3}
is_subset = subset.items() <= main.items()
print(is_subset)
````

**Explanation:** dict.items() views behave like sets of (key, value) pairs, so the <= operator can check whether every pair in subset.items() is also present in main.items(). Both the key and the value must match exactly for a pair to count.

## Exercise 32. Sort a Dictionary by Value Length

**Concept:** sorted key = len(value)

**Problem:** Order a dictionary's entries by how long their string values are.

**Given:**
```
words = {"a": "banana", "b": "kiwi", "c": "strawberry", "d": "fig"}
```

**Expected Output:**
```
{'d': 'fig', 'b': 'kiwi', 'a': 'banana', 'c': 'strawberry'}
```

**Hint:** key=lambda item: len(item[1]) sorts by the character count of each value.

````python
words = {"a": "banana", "b": "kiwi", "c": "strawberry", "d": "fig"}

sorted_words = dict(sorted(words.items(), key=lambda item: len(item[1])))

print(sorted_words)
````

**Explanation:** The lambda extracts each pair's value (item[1]) and measures its length, which sorted() then uses as the comparison basis instead of comparing the raw strings or keys.

## Exercise 33. Key With the Longest List Value

**Concept:** max() with a length-based key, then indexing [0]

**Problem:** Find which key holds the list with the most elements.

**Given:**
```
data = {"fruits": ["apple", "banana", "cherry"], "vegs": ["carrot"], "grains": ["rice", "wheat"]}
```

**Expected Output:**
```
fruits
```

**Hint:** max(data.items(), key=lambda item: len(item[1]))[0] gets the winning tuple, then its key.

````python
data = {"fruits": ["apple", "banana", "cherry"], "vegs": ["carrot"], "grains": ["rice", "wheat"]}

longest_key = max(data.items(), key=lambda item: len(item[1]))[0]

print(longest_key)
````

## Exercise 35. Invert a Dictionary (Unique Values)

**Concept:** {v: k for k, v in d.items()}

**Problem:** Swap a dictionary's keys and values, assuming all values are unique.

**Given:**
```
original = {"a": 1, "b": 2, "c": 3}
```

**Expected Output:**
```
{1: 'a', 2: 'b', 3: 'c'}
```

**Hint:** {v: k for k, v in original.items()} flips each pair around.

````python
original = {"a": 1, "b": 2, "c": 3}

inverted = {v: k for k, v in original.items()}

print(inverted)
````

## Exercise 36. Invert a Dictionary With Duplicate Values

**Concept:** dict.setdefault() to build grouped lists

**Problem:** Invert a dictionary where multiple keys share a value, grouping the original keys into a list instead of overwriting.

**Given:**
```
original = {"a": 1, "b": 2, "c": 1, "d": 3, "e": 2}
```

**Expected Output:**
```
{1: ['a', 'c'], 2: ['b', 'e'], 3: ['d']}
```

**Hint:** inverted.setdefault(v, []).append(k) creates the list on first sight, then appends on every sighting.

````python
original = {"a": 1, "b": 2, "c": 1, "d": 3, "e": 2}

inverted = {}
for k, v in original.items():
    inverted.setdefault(v, []).append(k)

print(inverted)
````

**Explanation:** setdefault(v, []) checks whether v is already a key in inverted; if not, it inserts an empty list for it. Either way it returns that list, so .append(k) always has something valid to add to — the net effect is that every original key ends up grouped under its shared value.

## Exercise 37. Flatten a Nested Dictionary

**Concept:** recursion with a dotted-key prefix

**Problem:** Collapse a multi-level nested dictionary into one flat dictionary, joining nested key paths with dots.

**Given:**
```
nested = {"a": 1, "b": {"c": 2, "d": {"e": 3, "f": 4}}}
```

**Expected Output:**
```
{'a': 1, 'b.c': 2, 'b.d.e': 3, 'b.d.f': 4}
```

**Hint:** Recurse whenever a value is itself a dict, building up a dotted key path as you go deeper.

````python
nested = {"a": 1, "b": {"c": 2, "d": {"e": 3, "f": 4}}}

def flatten(d, prefix=""):
    result = {}
    for k, v in d.items():
        new_key = f"{prefix}.{k}" if prefix else k
        if isinstance(v, dict):
            result.update(flatten(v, new_key))
        else:
            result[new_key] = v
    return result

print(flatten(nested))
````

**Explanation:** prefix tracks the dotted path accumulated so far; it's empty at the top level, so top-level keys stay bare. isinstance(v, dict) detects when there's another level to descend into — the function calls itself with that nested dict and the updated prefix, merging results back with .update(). A non-dict value is a leaf, stored directly under its fully-built key.

## Exercise 38. Group Words by First Letter

**Concept:** setdefault() for building grouped lists

**Problem:** Organize a list of words into groups keyed by their starting letter.

**Given:**
```
words = ["apple", "avocado", "banana", "blueberry", "cherry", "apricot"]
```

**Expected Output:**
```
{'a': ['apple', 'avocado', 'apricot'], 'b': ['banana', 'blueberry'], 'c': ['cherry']}
```

**Hint:** word[0] gives the grouping key; setdefault(letter, []).append(word) builds each group.

````python
words = ["apple", "avocado", "banana", "blueberry", "cherry", "apricot"]

grouped = {}
for word in words:
    letter = word[0]
    grouped.setdefault(letter, []).append(word)

print(grouped)
````

**Explanation:** word[0] extracts the first character to use as a grouping key. setdefault ensures a list exists for that letter (creating an empty one on first use) and always returns it, so .append(word) can add the current word to the correct group every time.

## Exercise 39. Merge Two Dictionaries, Summing Shared Keys

**Concept:** copy() + get() to add rather than overwrite

**Problem:** Merge two dictionaries so that shared keys have their numeric values added together.

**Given:**
```
dict1 = {"a": 10, "b": 20, "c": 30}, dict2 = {"b": 5, "c": 15, "d": 25}
```

**Expected Output:**
```
{'a': 10, 'b': 25, 'c': 45, 'd': 25}
```

**Hint:** Start from a copy of dict1, then merged[k] = merged.get(k, 0) + v while looping dict2.

````python
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 5, "c": 15, "d": 25}

merged = dict1.copy()
for k, v in dict2.items():
    merged[k] = merged.get(k, 0) + v

print(merged)
````

**Explanation:** Starting from dict1.copy() means dict1 itself is never modified. merged.get(k, 0) safely retrieves an existing value (or 0 if the key is new), so adding v on top always works — for a shared key like 'b' this computes 20 + 5 = 25, and for a brand-new key like 'd' it computes 0 + 25 = 25.

## Exercise 40. Shallow Copy vs. Deep Copy

**Concept:** copy.deepcopy() vs. .copy()

**Problem:** Demonstrate how mutating a nested list differs between a shallow copy and a deep copy.

**Given:**
```
original = {"name": "Alice", "scores": [90, 85, 92]}
```

**Expected Output:**
```
Shallow mutation affects the original too; deep mutation does not.
```

**Hint:** A shallow copy shares nested mutable objects with the original; copy.deepcopy() duplicates every level independently.

````python
import copy

original = {"name": "Alice", "scores": [90, 85, 92]}

shallow = original.copy()
shallow["scores"].append(100)
print("Shallow copy scores:", shallow["scores"])
print("Original scores after shallow mutation:", original["scores"])

print()

original = {"name": "Alice", "scores": [90, 85, 92, 100]}

deep = copy.deepcopy(original)
deep["scores"].append(99)
print("Deep copy scores:", deep["scores"])
print("Original scores after deep mutation:", original["scores"])
````

**Explanation:** original.copy() duplicates the outer dictionary, but the "scores" list inside it is the very same list object shared by both — appending to shallow["scores"] therefore also changes original["scores"]. copy.deepcopy() instead recursively duplicates every nested object, so deep["scores"] is a completely independent list and appending to it leaves original untouched.
