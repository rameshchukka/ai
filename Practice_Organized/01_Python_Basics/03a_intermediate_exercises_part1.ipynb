## Exercise 1. Filter and Transform in One Comprehension

**Concept:** list comprehension with filter + transform

**Problem:** Using one list comprehension, keep only strings of length 4+ and convert them to uppercase.

**Given:**
```
words = ["apple", "bat", "cherry", "dog", "elderberry"]
```

**Expected Output:**
```
['APPLE', 'CHERRY', 'ELDERBERRY']
```

**Hint:** The pattern is [expression for item in iterable if condition] — combine .upper() with a length check.

````python
words = ["apple", "bat", "cherry", "dog", "elderberry"]

filtered_words = [w.upper() for w in words if len(w) >= 4]

print(f"Original: {words}")
print(f"Result:   {filtered_words}")
````

**Explanation:** The if clause filters out short strings before the expression phase runs; .upper() is only ever applied to words that already passed the length check. The result is a new list — the original is untouched.

## Exercise 2. Merge Two Dictionaries, Summing Shared Keys

**Concept:** dict.get() with a default

**Problem:** Merge two dictionaries so that shared keys have their values added together, and unique keys are kept as-is.

**Given:**
```
dict_a = {'a': 10, 'b': 20}, dict_b = {'b': 5, 'c': 15}
```

**Expected Output:**
```
Merged Dictionary: {'a': 10, 'b': 25, 'c': 15}
```

**Hint:** Copy the first dictionary, then loop the second, using .get(key, 0) to safely add to existing or missing keys.

````python
def merge_dicts(d1, d2):
    result = d1.copy()
    for key, value in d2.items():
        result[key] = result.get(key, 0) + value
    return result

dict_a = {'a': 10, 'b': 20}
dict_b = {'b': 5, 'c': 15}

merged = merge_dicts(dict_a, dict_b)
print(f"Merged Dictionary: {merged}")
````

**Explanation:** .copy() keeps the original dict_a untouched. result.get(key, 0) returns the existing value if present, or 0 if it's the first time the key appears — so + value always works without a separate if/else.

## Exercise 4. Anagram Checker

**Concept:** sorting as a canonical form

**Problem:** Determine whether two strings are anagrams of each other.

**Given:**
```
word1 = "listen", word2 = "silent"
```

**Expected Output:**
```
Is "listen" an anagram of "silent"? True
```

**Hint:** Two strings are anagrams exactly when their sorted character lists are identical.

````python
def is_anagram(str1, str2):
    s1 = sorted(str1.lower().replace(" ", ""))
    s2 = sorted(str2.lower().replace(" ", ""))
    return s1 == s2

w1, w2 = "listen", "silent"
result = is_anagram(w1, w2)

print(f'Is "{w1}" an anagram of "{w2}"? {result}')
````

**Explanation:** sorted() turns each string into a canonical, order-independent list of characters. If the two sorted lists match exactly, the originals must contain the same characters, just rearranged.

## Exercise 5. Flatten a Nested List (Recursive)

**Concept:** recursion, isinstance()

**Problem:** Flatten a list containing other lists, of any nesting depth, into one flat list.

**Given:**
```
nested = [1, [2, 3], [4, [5, 6]], 7]
```

**Expected Output:**
```
Flattened: [1, 2, 3, 4, 5, 6, 7]
```

**Hint:** If an item is itself a list, recursively flatten it; otherwise append it directly.

````python
def flatten(lst):
    flat_list = []
    for item in lst:
        if isinstance(item, list):
            flat_list.extend(flatten(item))
        else:
            flat_list.append(item)
    return flat_list

nested_data = [1, [2, 3], [4, [5, 6]], 7]
result = flatten(nested_data)

print(f"Original:  {nested_data}")
print(f"Flattened: {result}")
````

**Explanation:** isinstance(item, list) detects when there's more nesting to unwrap, triggering a recursive call. extend() (not append()) folds the recursive result's individual items into the outer list rather than nesting it further.

## Exercise 6. Reverse Each Word, Keep Word Order

**Concept:** split/join with per-word slicing

**Problem:** Reverse the letters within each word of a sentence while keeping the words in their original order.

**Given:**
```
"Python is awesome"
```

**Expected Output:**
```
"nohtyP si emosewa"
```

**Hint:** Split into words, reverse each one with slicing, then join them back with spaces.

````python
def reverse_individual_words(sentence):
    words = sentence.split()
    reversed_words = [word[::-1] for word in words]
    return " ".join(reversed_words)

text = "Python is awesome"
result = reverse_individual_words(text)
print(f"Original: {text}")
print(f"Result:   {result}")
````

**Explanation:** split() breaks the sentence into a list of words. Each word is reversed independently via [::-1], and " ".join() glues the reversed words back together with spaces, preserving their original order.

## Exercise 7. Palindrome Sentence (Ignoring Punctuation)

**Concept:** str.isalnum() for sanitizing input

**Problem:** Check whether a full sentence is a palindrome, ignoring case, spaces, and punctuation.

**Given:**
```
"A man, a plan, a canal: Panama"
```

**Expected Output:**
```
True
```

**Hint:** Keep only alphanumeric characters (lowercased), then compare that cleaned string to its reverse.

````python
def is_palindrome_sentence(sentence):
    clean_chars = [char.lower() for char in sentence if char.isalnum()]
    clean_str = "".join(clean_chars)
    return clean_str == clean_str[::-1]

test_s = "A man, a plan, a canal: Panama"
print(f"Is palindrome? {is_palindrome_sentence(test_s)}")
````

**Explanation:** .isalnum() strips out commas, colons, and spaces, leaving only letters and digits. Lowercasing first means 'A' and 'a' are treated as the same character before the symmetry check runs.

## Exercise 9. Remove Duplicates, Preserve Order

**Concept:** a 'seen' set alongside a result list

**Problem:** Remove duplicates from a list while keeping the original element order, without converting to a set directly.

**Given:**
```
[1, 2, 2, 3, 1, 4, 2]
```

**Expected Output:**
```
[1, 2, 3, 4]
```

**Hint:** Track what's been seen in a set for fast lookups, while building the ordered output in a separate list.

````python
def remove_duplicates_ordered(items):
    seen = set()
    result = []
    for x in items:
        if x not in seen:
            result.append(x)
            seen.add(x)
    return result

nums = [1, 2, 2, 3, 1, 4, 2]
print(f"Cleaned List: {remove_duplicates_ordered(nums)}")
````

**Explanation:** result preserves insertion order, while seen (a set) makes the membership check close to instant regardless of list size — much faster than checking `x not in result` directly, which gets slower as the list grows.

## Exercise 10. Rotate a List Left or Right

**Concept:** slicing with modulo for safe shift amounts

**Problem:** Write a function that rotates a list by n positions in a given direction ('left' or 'right').

**Given:**
```
[1, 2, 3, 4, 5], shift 2, direction 'right'
```

**Expected Output:**
```
[4, 5, 1, 2, 3]
```

**Hint:** For a right shift of n, the answer is lst[-n:] + lst[:-n]; reduce n with modulo first to handle oversized shifts.

````python
def rotate_list(lst, n, direction='right'):
    if not lst:
        return lst
    n = n % len(lst)
    if direction == 'right':
        return lst[-n:] + lst[:-n]
    else:
        return lst[n:] + lst[:n]

data = [1, 2, 3, 4, 5]
print(f"Right Shift 2: {rotate_list(data, 2, 'right')}")
print(f"Left Shift 1:  {rotate_list(data, 1, 'left')}")
````

**Explanation:** n % len(lst) keeps the shift amount within range, so a shift of 102 on a 5-item list behaves the same as a shift of 2. Slicing then splits the list at the rotation point and swaps the two halves' order.

## Exercise 11. Merge Dictionaries Into Grouped Lists

**Concept:** set union of keys, conditional value collection

**Problem:** Merge two dictionaries so a shared key maps to a list of both values, instead of one overwriting the other.

**Given:**
```
d1 = {"a": 1, "b": 2}, d2 = {"b": 3, "c": 4}
```

**Expected Output:**
```
{'a': [1], 'b': [2, 3], 'c': [4]}
```

**Hint:** Union the two dictionaries' keys with |, then for each key collect whichever values exist.

````python
def merge_and_group(dict1, dict2):
    combined = {}
    all_keys = set(dict1.keys()) | set(dict2.keys())
    for key in all_keys:
        values = []
        if key in dict1:
            values.append(dict1[key])
        if key in dict2:
            values.append(dict2[key])
        combined[key] = values
    return combined

d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
print(f"Grouped Merge: {merge_and_group(d1, d2)}")
````

**Explanation:** set(d1) | set(d2) unions the keys from both dictionaries with no duplicates. Every value ends up wrapped in a list — even single values — which keeps the result's shape consistent regardless of how many sources contributed.

## Exercise 12. Invert an Author-to-Books Mapping

**Concept:** nested loops building a reverse lookup

**Problem:** Convert a dictionary of author -> list of books into book -> author.

**Given:**
```
{"Orwell": ["1984", "Animal Farm"], "Huxley": ["Brave New World"]}
```

**Expected Output:**
```
{'1984': 'Orwell', 'Animal Farm': 'Orwell', 'Brave New World': 'Huxley'}
```

**Hint:** Loop through authors and their book lists; for each book, make it a key pointing back to the author.

````python
def invert_bibliography(data):
    inverted = {}
    for author, books in data.items():
        for book in books:
            inverted[book] = author
    return inverted

authors = {
    "Orwell": ["1984", "Animal Farm"],
    "Huxley": ["Brave New World"]
}
print(f"Inverted Index: {invert_bibliography(authors)}")
````

**Explanation:** The outer loop picks an author and their book list; the inner loop unpacks that list one title at a time. Each book becomes a key pointing back to its author — this only works cleanly if book titles are unique across the whole dataset.

## Exercise 14. Subset / Superset / Disjoint Check

**Concept:** set relational methods

**Problem:** Determine whether one set is a subset, superset, or disjoint from another.

**Given:**
```
[1, 2] vs [1, 2, 3, 4]
```

**Expected Output:**
```
Set A is a subset of Set B.
```

**Hint:** Use .issubset(), .issuperset(), and .isdisjoint() on two sets.

````python
def validate_relationships(list1, list2):
    a = set(list1)
    b = set(list2)
    if a.issubset(b):
        print("Set A is a subset of Set B.")
    elif a.issuperset(b):
        print("Set A is a superset of Set B.")
    if a.isdisjoint(b):
        print("The sets are disjoint (they share no elements).")
    else:
        print(f"The sets share these elements: {a & b}")

validate_relationships([1, 2], [1, 2, 3, 4])
````

**Explanation:** issubset() checks that every element of A is also in B; issuperset() checks the reverse. isdisjoint() reports True only if the two sets have zero elements in common — a quick way to test for no overlap at all.

## Exercise 15. Symmetric Difference of Two ID Lists

## Exercise 16. Generate a Power Set

**Concept:** itertools.combinations

**Problem:** Generate every possible subset of a set, including the empty set and the full set.

**Given:**
```
{1, 2, 3}
```

**Expected Output:**
```
[(), (1,), (2,), (3,), (1, 2), (1, 3), (2, 3), (1, 2, 3)]
```

**Hint:** Loop the subset size from 0 to the set's length, using itertools.combinations for each size.

````python
from itertools import combinations

def get_power_set(s):
    elements = list(s)
    power_set = []
    for r in range(len(elements) + 1):
        for combo in combinations(elements, r):
            power_set.append(combo)
    return power_set

my_set = {1, 2, 3}
print(f"Power Set: {get_power_set(my_set)}")
````

**Explanation:** combinations(elements, r) generates every way to choose r items, order-independent and without repeats. Sweeping r from 0 up to the full length covers everything from the empty subset to the complete set — 2**n subsets in total.

## Exercise 17. Exact Age Calculator

**Concept:** datetime.strptime, dateutil.relativedelta

**Problem:** Given a birthdate string, calculate the exact age in years, months, and days.

**Given:**
```
"1995-05-15"
```

**Expected Output:**
```
30 years, 7 months, 18 days (relative to today)
```

**Hint:** Parse the string with strptime, then use relativedelta to get a calendar-aware difference.

````python
from datetime import datetime
from dateutil.relativedelta import relativedelta

def calculate_exact_age(birthdate_str):
    try:
        birthdate = datetime.strptime(birthdate_str, "%Y-%m-%d").date()
        today = datetime.now().date()
        diff = relativedelta(today, birthdate)
        return f"{diff.years} years, {diff.months} months, {diff.days} days"
    except ValueError:
        return "Invalid date format. Please use YYYY-MM-DD."

user_age = calculate_exact_age("1995-05-15")
print(f"Exact Age: {user_age}")
````

**Explanation:** strptime converts the text into an actual date object using a format code. Unlike a plain subtraction, relativedelta understands calendar rules (leap years, varying month lengths) and can break the difference down into whole years, months, and days.

## Exercise 18. Countdown to New Year

**Concept:** timedelta arithmetic, divmod

**Problem:** Calculate how many days, hours, and minutes remain until the next New Year's Day.

**Given:**
```
current date
```

**Expected Output:**
```
e.g. 363 days, 16 hours, 50 minutes
```

**Hint:** Subtract now from a datetime for next Jan 1, then use divmod to convert seconds into hours/minutes.

````python
from datetime import datetime

def countdown_to_new_year():
    now = datetime.now()
    next_year = now.year + 1
    target = datetime(next_year, 1, 1)
    diff = target - now
    days = diff.days
    hours, remainder = divmod(diff.seconds, 3600)
    minutes, _ = divmod(remainder, 60)
    return f"{days} days, {hours} hours, {minutes} minutes"

print(f"Time remaining: {countdown_to_new_year()} until New Year!")
````

**Explanation:** Subtracting two datetime objects produces a timedelta with a whole-days count and a leftover-seconds count. divmod(seconds, 3600) splits those seconds into hours and a remainder in one step, and a second divmod converts that remainder into minutes.

## Exercise 19. Add N Business Days

**Concept:** while loop skipping weekends

**Problem:** Find the date exactly N weekdays (Mon-Fri) after a given start date.

**Given:**
```
Friday 2026-01-02, 5 business days
```

**Expected Output:**
```
Friday, 2026-01-09
```

**Hint:** Step forward one day at a time, only counting a step if weekday() is less than 5 (i.e. not Sat/Sun).

````python
from datetime import timedelta, datetime

def add_business_days(start_date, n):
    current_date = start_date
    days_added = 0
    while days_added < n:
        current_date += timedelta(days=1)
        if current_date.weekday() < 5:
            days_added += 1
    return current_date

start = datetime(2026, 1, 2)
result = add_business_days(start, 5)
print(f"Start: {start.strftime('%A, %Y-%m-%d')}")
print(f"Business Target: {result.strftime('%A, %Y-%m-%d')}")
````

**Explanation:** weekday() returns 0 for Monday through 6 for Sunday, so values 5 and 6 are the weekend. The loop steps the date forward one day at a time but only increments the business-day counter on weekdays, effectively skipping weekends.

## Exercise 20. Custom Iterator Class

**Concept:** __iter__ / __next__ / StopIteration

**Problem:** Build a class that iterates through powers of 2 up to a maximum exponent.

**Given:**
```
PowerOfTwo(3)
```

**Expected Output:**
```
1
2
4
8
```

**Hint:** Track state in the instance; __next__ should raise StopIteration once the exponent exceeds the max.

````python
class PowerOfTwo:
    def __init__(self, max_exponent):
        self.max = max_exponent
        self.n = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.n <= self.max:
            result = 2 ** self.n
            self.n += 1
            return result
        else:
            raise StopIteration

for val in PowerOfTwo(3):
    print(val)
````

**Explanation:** __iter__ tells Python the object can be looped over; __next__ is called on every step. Raising StopIteration is the built-in signal that tells a for loop to stop — without it the loop would never end.

## Exercise 21. Find Duplicates in Linear Time

**Concept:** set as an O(1) lookup table

**Problem:** Identify duplicate values in a list by scanning it only once.

**Given:**
```
[1, 2, 3, 2, 4, 5, 1, 6]
```

**Expected Output:**
```
Duplicates found: {1, 2}
```

**Hint:** Track already-seen values in a set; a value seen again is a duplicate.

````python
def find_duplicates(nums):
    seen = set()
    duplicates = set()
    for x in nums:
        if x in seen:
            duplicates.add(x)
        else:
            seen.add(x)
    return duplicates

numbers = [1, 2, 3, 2, 4, 5, 1, 6]
print(f"Duplicates found: {find_duplicates(numbers)}")
````

**Explanation:** Checking membership in a set is close to instant regardless of its size, unlike a list. A single pass suffices because each value is compared only against everything seen before it, giving O(n) overall instead of the O(n^2) a nested-loop approach would cost.

## Exercise 22. Singly Linked List

**Concept:** Node/next pointers, class-based data structure

**Problem:** Build a Node class and a LinkedList class with append and display methods.

**Given:**
```
append 10, 20, 30
```

**Expected Output:**
```
10 -> 20 -> 30 -> None
```

**Hint:** Each Node stores data and a reference to the next Node; LinkedList tracks the head and walks .next to append.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last = self.head
        while last.next:
            last = last.next
        last.next = new_node

    def display(self):
        current = self.head
        while current:
            print(f"{current.data} -> ", end="")
            current = current.next
        print("None")

ll = LinkedList()
ll.append(10)
ll.append(20)
ll.append(30)
ll.display()
````

**Explanation:** None marks the end of the chain, like a null pointer. append() walks from the head via .next until it finds the node with no successor, then attaches the new node there. Nodes are linked purely by reference, unlike a list's contiguous memory layout.

## Exercise 23. Stack (LIFO) via a List

**Concept:** push/pop/peek encapsulation

**Problem:** Implement a Stack with push, pop, and peek, backed by a plain Python list.

**Given:**
```
push "google.com", "pynative.com"; then pop
```

**Expected Output:**
```
Current Top: pynative.com
Popped: pynative.com
New Top: google.com
```

**Hint:** list.append() and list.pop() (no args) are already optimized for adding/removing from the end.

````python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return "Stack is empty"

    def peek(self):
        return self.items[-1] if not self.is_empty() else None

    def is_empty(self):
        return len(self.items) == 0

browser_history = Stack()
browser_history.push("google.com")
browser_history.push("pynative.com")

print(f"Current Top: {browser_history.peek()}")
print(f"Popped: {browser_history.pop()}")
print(f"New Top: {browser_history.peek()}")
````

**Explanation:** pop() with no arguments removes the most recently added item, giving Last-In-First-Out behavior. Wrapping the list in a class hides direct list access, so callers can only interact with the stack through push/pop/peek.

## Exercise 24. Queue (FIFO) via deque

**Concept:** collections.deque, popleft()

**Problem:** Implement a Queue with enqueue and dequeue, backed by collections.deque.

**Given:**
```
enqueue "Customer A", "Customer B"; then dequeue
```

**Expected Output:**
```
Serving: Customer A
Next in line: Customer B
```

**Hint:** append() adds to the back; popleft() removes from the front in O(1) time.

````python
from collections import deque

class Queue:
    def __init__(self):
        self.buffer = deque()

    def enqueue(self, val):
        self.buffer.append(val)

    def dequeue(self):
        if len(self.buffer) == 0:
            return "Queue is empty"
        return self.buffer.popleft()

    def peek(self):
        return self.buffer[0] if self.buffer else None

ticket_line = Queue()
ticket_line.enqueue("Customer A")
ticket_line.enqueue("Customer B")

print(f"Serving: {ticket_line.dequeue()}")
print(f"Next in line: {ticket_line.peek()}")
````

**Explanation:** deque supports fast additions and removals from both ends, unlike a plain list where removing index 0 requires shifting every other element. popleft() removes the earliest-added item, giving First-In-First-Out order.

## Exercise 25. Recursive Binary Search

**Concept:** divide and conquer, call stack

**Problem:** Find a target's index in a sorted list using recursive binary search.

**Given:**
```
[10, 20, 30, 40, 50, 60], target 50
```

**Expected Output:**
```
Target found at index: 4
```

**Hint:** Compare the target to the middle element, then recurse into the left or right half accordingly.

````python
def binary_search_recursive(arr, low, high, x):
    if high < low:
        return -1
    mid = (low + high) // 2
    if arr[mid] == x:
        return mid
    elif arr[mid] > x:
        return binary_search_recursive(arr, low, mid - 1, x)
    else:
        return binary_search_recursive(arr, mid + 1, high, x)

data = [10, 20, 30, 40, 50, 60]
target = 50
result = binary_search_recursive(data, 0, len(data) - 1, target)

print(f"Target found at index: {result}")
````

**Explanation:** Each call halves the search space by comparing the target to the midpoint and recursing into only the relevant half. The base case (high < low) signals the target isn't present; otherwise the matching index eventually bubbles back up through the chain of calls, giving O(log n) performance.

## Exercise 27. Combine map() and filter()

```
[1, 2, 3, 4, 5, 6]
```

**Expected Output:**
```
[4, 16, 36]
```

**Hint:** filter() keeps the evens first; wrap that result in map() to square each survivor.

````python
nums = [1, 2, 3, 4, 5, 6]

result = list(map(lambda x: x**2, filter(lambda x: x % 2 == 0, nums)))

print(f"Original: {nums}")
print(f"Squares of evens: {result}")
````

## Exercise 28. Custom @timer Decorator

**Concept:** decorators, *args/**kwargs, time.perf_counter

**Problem:** Write a decorator that prints how long a wrapped function took to run.

**Given:**
```
a function that sleeps 1.5 seconds
```

**Expected Output:**
```
Function 'waste_time' took ~1.5000 seconds.
```

**Hint:** A decorator is a function that takes a function, defines an inner wrapper, and returns that wrapper.

````python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()
        result = func(*args, **kwargs)
        end_time = time.perf_counter()
        duration = end_time - start_time
        print(f"Function '{func.__name__}' took {duration:.4f} seconds.")
        return result
    return wrapper

@timer
def waste_time():
    time.sleep(1.5)

waste_time()
````

**Explanation:** *args, **kwargs let the wrapper forward any combination of arguments to the wrapped function, so the decorator works on any function signature. @timer is shorthand for waste_time = timer(waste_time), swapping in the wrapped version.

## Exercise 29. Memory-Efficient Fibonacci Generator

**Concept:** yield, lazy generation

**Problem:** Write a generator function that yields Fibonacci numbers one at a time instead of building a full list.

**Given:**
```
8 terms
```

**Expected Output:**
```
0 1 1 2 3 5 8 13
```

**Hint:** Use a while loop with yield, updating both tracked values each pass with a, b = b, a + b.

````python
def fibonacci_gen(limit):
    a, b = 0, 1
    count = 0
    while count < limit:
        yield a
        a, b = b, a + b
        count += 1

fib = fibonacci_gen(8)

print("First 8 Fibonacci numbers:")
for num in fib:
    print(num, end=" ")
````

**Explanation:** yield hands back one value and pauses the function's execution in place, resuming exactly where it left off the next time a value is requested. This keeps memory use constant no matter how many numbers are generated, unlike building a full list up front.

## Exercise 30. Custom Context Manager

**Concept:** __enter__ / __exit__, guaranteed cleanup

**Problem:** Write a class-based context manager that prints connect/disconnect messages, ensuring disconnect happens even if an error occurs inside the with block.

**Given:**
```
a with block that raises an exception
```

**Expected Output:**
```
Connecting to Database...
Processing data...
Error: something went wrong
Closing Database Connection safely.
```

**Hint:** __enter__ sets up the resource; __exit__ is guaranteed to run at the end of the with block, error or not.

````python
class DatabaseConnection:
    def __enter__(self):
        print("Connecting to Database...")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Closing Database Connection safely.")
        return False

try:
    with DatabaseConnection() as db:
        print("Processing data...")
        raise Exception("something went wrong")
except Exception as e:
    print(f"Error: {e}")
````

**Explanation:** __enter__ runs when the with block starts, and __exit__ runs when it ends, no matter whether that's a normal exit or an exception. This is exactly why with open(...) is preferred for files — cleanup happens even when something goes wrong.

## Exercise 33. Memoized Fibonacci with lru_cache

**Concept:** functools.lru_cache, memoization

**Problem:** Speed up a recursive Fibonacci function by caching previously computed results.

**Given:**
```
fibonacci(50)
```

**Expected Output:**
```
Fibonacci(50) = 12586269025
```

**Hint:** Decorate the recursive function with @lru_cache(maxsize=None).

````python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

n = 50
print(f"Fibonacci({n}) = {fibonacci(n)}")
print(f"Cache Info: {fibonacci.cache_info()}")
````

**Explanation:** Without caching, recursive Fibonacci recomputes the same sub-values over and over, an exponential blowup. lru_cache stores each result the first time it's computed, so future calls with the same argument return instantly from the cache instead of recalculating.

## Exercise 34. Set Operations for Comparing Two Groups

**Concept:** &, -, ^ set operators together

**Problem:** Given two groups (trial users and paying users), find who upgraded, who's still just a lead, and who's exclusively in one group.

**Given:**
```
trial = {1,2,3,4,5}, paid = {4,5,6,7,8}
```

**Expected Output:**
```
Upgraded: {4, 5}, Leads: {1, 2, 3}, Unique: {1, 2, 3, 6, 7, 8}
```

**Hint:** & gives the overlap, - gives what's only in the first set, ^ gives what's in exactly one set.

````python
trial = {1, 2, 3, 4, 5}
paid = {4, 5, 6, 7, 8}

upgraded = trial & paid
leads = trial - paid
unique = trial ^ paid

print(f"Upgraded (Both): {upgraded}")
print(f"Leads (Trial only): {leads}")
print(f"Unique Status (Not both): {unique}")
````

**Explanation:** & (intersection) keeps only shared elements, - (difference) keeps what's in the first set but not the second, and ^ (symmetric difference) keeps everything that's in exactly one set. All three run in optimized C code, far faster than manual loops on large datasets.

## Exercise 36. Encapsulation With a Private Balance

**Concept:** name-mangled private attributes

**Problem:** Build a BankAccount where the balance can't be accessed directly from outside, only through deposit/withdraw methods.

**Given:**
```
account = BankAccount(100); deposit(50); withdraw(200)
```

**Expected Output:**
```
Deposited 50. New Balance: 150
Insufficient funds!
```

**Hint:** A double-underscore prefix (self.__balance) makes an attribute private via name mangling.

````python
class BankAccount:
    def __init__(self, initial_balance):
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New Balance: {self.__balance}")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New Balance: {self.__balance}")
        else:
            print("Insufficient funds!")

    def get_balance(self):
        return self.__balance

acc = BankAccount(100)
acc.deposit(50)
acc.withdraw(200)
print(f"Final Balance: {acc.get_balance()}")
````

**Explanation:** The double-underscore prefix hides __balance from direct outside access. Every change has to go through deposit() or withdraw(), both of which validate the amount first — that validation is the entire point of encapsulation.

## Exercise 37. Validated Attribute With @property

**Concept:** @property getter and @x.setter

**Problem:** Create a Student class whose score can only be set between 0 and 100, enforced through a property setter.

**Given:**
```
s.score = 105
```

**Expected Output:**
```
ValueError: Score must be between 0 and 100.
```

**Hint:** Store the real value in a protected _score, and validate inside a @score.setter.

````python
class Student:
    def __init__(self, name):
        self.name = name
        self._score = 0

    @property
    def score(self):
        return self._score

    @score.setter
    def score(self, value):
        if 0 <= value <= 100:
            self._score = value
        else:
            raise ValueError("Score must be between 0 and 100.")

s = Student("Kevin")
try:
    s.score = 95
    print(f"{s.name}'s score: {s.score}")
    s.score = 105
except ValueError as e:
    print(f"Error: {e}")
````

**Explanation:** @property turns score() into a getter accessed like a plain attribute (s.score, no parentheses). @score.setter intercepts assignment (s.score = 105), letting validation logic run invisibly behind what looks like ordinary attribute access.

## Exercise 38. Class Methods vs Static Methods

**Concept:** @classmethod (factory) vs @staticmethod (utility)

**Problem:** Add a classmethod that builds a preset Pizza object, and a staticmethod that validates a topping independent of any instance.

**Given:**
```
Pizza.margherita(); Pizza.validate_topping("pineapple")
```

**Expected Output:**
```
Pizza ordered: Margherita
Is topping valid? True
```

**Hint:** @classmethod receives cls (the class itself); @staticmethod receives neither cls nor self.

````python
class Pizza:
    def __init__(self, name, toppings):
        self.name = name
        self.toppings = toppings

    @classmethod
    def margherita(cls):
        return cls("Margherita", ["cheese", "tomato"])

    @staticmethod
    def validate_topping(topping):
        prohibited = ["plastic", "metal"]
        return topping not in prohibited

order = Pizza.margherita()
print(f"Pizza ordered: {order.name}")
print(f"Is 'mushrooms' valid? {Pizza.validate_topping('mushrooms')}")
````

**Explanation:** cls inside a classmethod refers to the class itself, so cls(...) is equivalent to calling Pizza(...) directly — handy for 'factory' presets. A staticmethod is just an ordinary function that happens to live inside the class for organizational purposes; it needs no access to instance or class state.

## Exercise 39. Magic Methods: __str__ and __add__

**Concept:** operator overloading, dunder methods

**Problem:** Give a Point class a readable string form and support adding two points with +.

**Given:**
```
Point(1,2) + Point(3,4)
```

**Expected Output:**
```
(4, 6)
```

**Hint:** __str__ controls what print() shows; __add__ controls what the + operator does between two Points.

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __add__(self, other):
        new_x = self.x + other.x
        new_y = self.y + other.y
        return Point(new_x, new_y)

p1 = Point(1, 2)
p2 = Point(3, 4)
combined = p1 + p2

print(f"Point 1: {p1}")
print(f"Point 2: {p2}")
print(f"Combined Result: {combined}")
````

**Explanation:** Without __str__, printing an object would show its raw memory address instead of useful data. __add__ is what Python calls when it sees p1 + p2 on custom objects — self is the left operand, other is the right, and the method returns a brand-new Point.

## Exercise 40. Abstract Base Classes

**Concept:** abc.ABC, @abstractmethod, polymorphism

**Problem:** Define an abstract Shape class with an area() method that every subclass must implement.

**Given:**
```
shapes = [Circle(5), Square(4)]
```

**Expected Output:**
```
Circle Area: 78.54
Square Area: 16
```

**Hint:** Inherit from ABC and mark area() with @abstractmethod so subclasses are forced to override it.

````python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

shapes = [Circle(5), Square(4)]
for s in shapes:
    print(f"{type(s).__name__} Area: {s.area():.2f}")
````

**Explanation:** Inheriting from ABC blocks Shape() from being instantiated directly. @abstractmethod forces every concrete subclass to supply its own area(). The loop calls .area() on each shape without caring which subclass it actually is — that's polymorphism in action.
