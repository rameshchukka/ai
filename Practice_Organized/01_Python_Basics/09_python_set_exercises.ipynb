## Concepts you'll need

This set covers Python **sets** — unordered collections of unique, hashable values — and the operations built around set theory.

- **Mutation methods** — `.add(x)` inserts one item; `.remove(x)` deletes and raises `KeyError` if missing; `.discard(x)` deletes safely with no error; `.pop()` removes and returns an arbitrary element; `.clear()` empties the set in place.
- **The big four set-theory operations**, each with a method and an operator:
  - Union (`|` / `.union()`) — everything from both sets
  - Intersection (`&` / `.intersection()`) — only what's shared
  - Difference (`-` / `.difference()`) — in the first but not the second (order matters!)
  - Symmetric difference (`^` / `.symmetric_difference()`) — in exactly one, not both
- **In-place versions** — `.difference_update()`, `.intersection_update()`, `.symmetric_difference_update()` modify a set directly instead of returning a new one.
- **Relationship checks** — `.issubset()`/`<=`, `.issuperset()`/`>=`, proper versions `<`/`>` (strict, excludes equality), and `.isdisjoint()` (no overlap at all).
- **Set comprehensions** — `{expr for x in iterable if condition}`, just like list comprehensions but automatically deduplicated.
- **Hashability** — only immutable types (numbers, strings, tuples) can live inside a set, because sets rely on a stable hash value for fast lookups; lists can't be set elements for exactly this reason.
- **`frozenset`** — an immutable set, usable as a dictionary key or nested inside another set.
- **Why sets matter for performance** — membership testing (`in`) is O(1) average on a set versus O(n) on a list, which is why sets are the standard tool for deduplication and fast lookups.

## Exercise 1. Add, Remove, and Discard

**Concept:** add(), remove() vs discard()

**Problem:** Add an element to a set, remove one with .remove(), and safely discard a non-existent one.

**Given:**
```
fruits = {"apple", "banana", "cherry"}
```

**Expected Output:**
```
After add: {..., 'mango'}
After remove: (without 'banana')
After discard (no error)
```

**Hint:** .remove() raises KeyError if the element is missing; .discard() does not.

````python
fruits = {"apple", "banana", "cherry"}

fruits.add("mango")
print("After add:", fruits)

fruits.remove("banana")
print("After remove:", fruits)

fruits.discard("grape")
print("After discard (element not present):", fruits)
````

**Explanation:** .add() inserts a new element (or does nothing if it's already there). .remove() deletes an element but raises KeyError if it's absent — use it only when you're sure the element exists. .discard() attempts the same removal but silently does nothing if the element isn't found, making it the safer choice for uncertain removals.

## Exercise 3. Count Elements Manually

**Concept:** a for loop counter, without len()

**Problem:** Count how many elements are in a set without using len().

**Given:**
```

```

**Expected Output:**
```

```

**Hint:** Loop over the set with a counter starting at 0, incrementing once per element.

````python
animals = {"cat", "dog", "bird", "fish"}

count = 0
for _ in animals:
    count += 1

print("Length of set:", count)
````

**Explanation:** The underscore _ is a conventional placeholder name signaling that the loop variable's actual value isn't needed — only the fact that one more element was visited. In practice, len(animals) achieves the same result instantly; this loop is purely for illustration.

## Exercise 7. Difference of Two Sets

**Concept:** difference() / the - operator (order matters)

**Problem:** Find elements in one set but not the other.

**Given:**
```
set_a = {1, 2, 3, 4}, set_b = {3, 4, 5, 6}
```

**Expected Output:**
```
Difference (A - B): {1, 2}
```

**Hint:** set_a - set_b keeps only what's unique to set_a; the reverse order gives a different answer.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

result = set_a.difference(set_b)
print("Difference (A - B):", result)
````

**Explanation:** difference() removes anything from set_a that also appears in set_b, leaving only set_a's exclusive elements. Unlike union and intersection, this operation isn't symmetric — set_b.difference(set_a) would instead return {5, 6}.

## Exercise 8. Symmetric Difference

**Concept:** symmetric_difference() / the ^ operator

**Problem:** Find elements that belong to exactly one of two sets, not both.

**Given:**
```
set_a = {1, 2, 3, 4}, set_b = {3, 4, 5, 6}
```

**Expected Output:**
```
{1, 2, 5, 6}
```

**Hint:** Think of it as the union minus the intersection.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

result = set_a.symmetric_difference(set_b)
print("Symmetric Difference:", result)
````

**Explanation:** Elements shared by both sets (3 and 4) are excluded, leaving only what's exclusive to each side. Unlike plain difference, symmetric difference is commutative — set_a ^ set_b always equals set_b ^ set_a.

## Exercise 13. Subset and Superset Checks

**Concept:** issubset() / issuperset()

**Problem:** Check whether one set is fully contained within another, and vice versa.

**Given:**
```
set_a = {1, 2, 3}, set_b = {1, 2, 3, 4, 5}
```

**Expected Output:**
```
Is set_a a subset of set_b? True
Is set_b a superset of set_a? True
```

**Hint:** a.issubset(b) checks every element of a is in b; b.issuperset(a) is the mirror check.

````python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}

print("Is set_a a subset of set_b?", set_a.issubset(set_b))
print("Is set_b a superset of set_a?", set_b.issuperset(set_a))
````

## Exercise 15. Remove Overlapping Elements In Place

**Concept:** difference_update()

**Problem:** Strip out of a set every element that also appears in another set, modifying the first set directly.

## Exercise 24. Proper Subset and Superset

**Concept:** the strict < and > operators

**Problem:** Distinguish a proper (strict) subset/superset relationship from a regular one.

**Given:**


**Expected Output:**
```
a < b: True
b > a: True

```

**Hint:** A set is never a proper subset of itself, even though it is always a (non-strict) subset of itself.

## Exercise 25. Frozen Sets

**Concept:** frozenset immutability

**Problem:** Create a frozenset, use it in read-only set operations, and confirm mutation attempts fail.

**Given:**
```
fs = frozenset([1, 2, 3, 4, 5])
```

**Expected Output:**
```
Intersection works normally; fs.add(6) raises AttributeError
```

**Hint:** Non-mutating methods like .intersection() work fine; mutating methods like .add() simply don't exist on frozenset.

````python
fs = frozenset([1, 2, 3, 4, 5])

print("frozenset:", fs)
print("Intersection with {3, 4, 5, 6}:", fs.intersection({3, 4, 5, 6}))

try:
    fs.add(6)
except AttributeError as e:
    print("Error:", e)
````

But it implements no mutation methods at all, so calling .add() raises an AttributeError rather than modifying anything. Because it's immutable, a frozenset is hashable and can be used as a dictionary key or nested inside another set.

## Exercise 27. Remove Duplicates While Preserving Order

**Concept:** a 'seen' set alongside an ordered result list

**Problem:** Deduplicate a list while keeping the original order of first appearances.

**Given:**
```
items = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
```

**Expected Output:**
```
[3, 1, 4, 5, 9, 2, 6]
```

**Hint:** Converting straight to a set loses order; track 'seen' items in a set while building an ordered result list.

````python
items = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

seen = set()
result = []
for item in items:
    if item not in seen:
        result.append(item)
        seen.add(item)

print(result)
````

**Explanation:** seen provides O(1) average membership checks, keeping the whole loop efficient even for large inputs. Because the loop moves through items left to right, result naturally ends up holding each value in the order it was first encountered — unlike list(set(items)), which would scramble that order.

## Exercise 30. Assignment vs. .copy()

**Concept:** reference sharing vs. an independent copy

**Problem:** Show the difference between assigning a set to a new name and making an independent copy.

**Given:**
```
original = {1, 2, 3, 4, 5}
```

**Expected Output:**
```
Assignment: mutating ref also changes original.
Copy: mutating copied leaves original untouched.
```

**Hint:** ref = original creates a second name for the same object; original.copy() creates a genuinely separate object.

````python
original = {1, 2, 3, 4, 5}

ref = original
ref.add(99)
print("--- Assignment (=) ---")
print("original:", original)
print("ref:", ref)

original = {1, 2, 3, 4, 5}

copied = original.copy()
copied.add(42)
print("\n--- Shallow copy (.copy()) ---")
print("original:", original)
print("copied:", copied)
````

**Explanation:** ref = original doesn't create anything new — both names simply point to the exact same set object, so mutating one is indistinguishable from mutating the other. original.copy() instead creates a brand-new, independent set that starts with the same elements but has no ongoing connection to the original, so changes to one never affect the other.

## Exercise 31. Membership Testing: List vs. Set Speed

**Concept:** O(n) list search vs. O(1) average set lookup

**Problem:** Measure and compare how long a membership check takes on a large list versus a large set.

**Given:**
```
a range of 1,000,000 integers, searching for the last one
```

**Expected Output:**
```
List lookup is dramatically slower than set lookup
```

**Hint:** `in` on a list scans linearly; `in` on a set uses hashing for near-instant lookup.

````python
import time

large_list = list(range(1_000_000))
large_set = set(range(1_000_000))
target = 999999

start = time.time()
result = target in large_list
list_time = time.time() - start

start = time.time()
result = target in large_set
set_time = time.time() - start

print(f"List lookup time:  {list_time:.6f} seconds")
print(f"Set lookup time:   {set_time:.6f} seconds")
````

**Explanation:** Checking membership in a list means scanning from the start until a match is found (or the list runs out) — for a target near the end, that's close to a million comparisons. A set instead computes the target's hash and jumps almost directly to where it would be stored, taking roughly the same tiny amount of time no matter how large the set is. The practical takeaway: convert a collection to a set once if you'll be checking membership against it repeatedly.
