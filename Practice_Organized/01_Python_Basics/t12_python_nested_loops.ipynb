# Nested Loops in Python

A **nested loop** is a loop placed inside the body of another loop. Either level can be `for` or `while`, in any combination — a `for` inside a `while`, a `while` inside a `for`, or same-type nesting — and there's no hard limit on how many levels deep you can go.

The key rule: for EVERY single iteration of the outer loop, the inner loop restarts from scratch and runs through ALL of its own iterations before the outer loop is allowed to advance. So the total number of times the inner body runs equals `(outer iterations) x (inner iterations)`.

Nested loops are the natural tool for anything grid- or table-shaped: 2D data, pattern printing, and combinations/pairings between two collections.

## Nested `for` Loop: Multiplication Table

The outer loop picks a row number (1-10); for EACH row, the inner loop runs a full 1-10 pass computing that row's products.

````python
for i in range(1, 11):

    for j in range(1, 11):

        print(i * j, end=' ')

    print()
````

## Nested Loop for Pattern Printing

Here the inner loop's own range depends on the OUTER loop's current value — this is what makes each row a different length.

````python
rows = 5

for i in range(1, rows + 1):

    for j in range(1, i + 1):

        print("*", end=" ")

    print('')
````

## `while` Loop Inside a `for` Loop

Repeat each name from a list exactly 5 times, using an inner `while` loop as the repetition counter.

````python
names = ['Kelly', 'Jessa', 'Emma']

for name in names:

    count = 0

    while count < 5:

        print(name, end=' ')

        count = count + 1

    print()
````

**Practice problem:** print a 5-row, 3-column rectangle of stars.

````python
for row in range(5):

    for col in range(3):

        print('*', end='')

    print()
````

## `break` in a Nested Loop

`break` inside the inner loop terminates ONLY that inner loop — for each outer value, once `j == i`, the current row stops early, but the outer loop still moves on to try the next row.

````python
for i in range(4):

    for j in range(4):

        if j == i:

            break

        print(i, j)
````

## `continue` in a Nested Loop

Skip a specific pairing (where both loop's current values match) without stopping either loop.

````python
first = [2, 4, 6]

second = [2, 4, 6]

for i in first:

    for j in second:

        if i == j:

            continue

        print(i, '*', j, '= ', i * j)
````

## Single-Line Nested Loops via List Comprehension

A nested `for` loop pair can often collapse into one comprehension. Start with the explicit two-loop version:

````python
first = [2, 3, 4]

second = [20, 30, 40]

final = []

for i in first:

    for j in second:

        final.append(i + j)

print(final)
````

The equivalent comprehension: the FIRST `for` clause is the outer loop, the SECOND is the inner loop — same left-to-right order as writing them out explicitly.

````python
first = [2, 3, 4]

second = [20, 30, 40]

final = [i + j for i in first for j in second]

print(final)
````

A comprehension can also carry a filter condition, dropping certain combinations entirely — here, excluding any pairing where both values happen to be equal.

````python
final = [[x, y] for x in [10, 20, 30] for y in [30, 10, 50] if x != y]

print(final)
````

## Nested `while` Loop

```
while expression:
    while expression:
        statement(s)
    statement(s)
```

````python
i = 1

while i <= 5:

    j = 1

    while j <= 10:

        print(j, end='')

        j = j + 1

    i = i + 1

    print()
````

### `for` Loop Inside a `while` Loop: Finding Perfect Numbers

A [perfect number](https://en.wikipedia.org/wiki/Perfect_number) equals the sum of its own proper divisors (e.g. 6 = 1+2+3). The outer `while` walks candidate numbers 2 through 100; for EACH candidate, an inner `for` loop checks every smaller number to see if it divides evenly, accumulating a running sum of divisors.

````python
print('Show Perfect number from 1 to 100')

n = 2

while n <= 100:

    x_sum = 0

    for i in range(1, n):

        if n % i == 0:

            x_sum += i

    if x_sum == n:

        print('Perfect number:', n)

    n += 1
````

## When to Use a Nested Loop

- Naturally 2D-shaped data — nested lists, grids, tables.
- Pattern printing (star/number patterns) where row and column both matter.
- Generating combinations/pairings between two collections.

### A Word on Time Complexity

A single loop over `n` items is O(n) — n total steps. A nested loop, with BOTH the outer and inner loop running their FULL range unconditionally, does n outer passes x n inner passes = **O(n^2)** total steps — growing quadratically, not linearly, as the input size increases. For large inputs this can matter a great deal: doubling the input roughly QUADRUPLES the work for an O(n^2) nested loop, whereas it only doubles the work for an O(n) single loop.

If a nested loop can be short-circuited early (via a well-placed `break`, or restructured to avoid the nesting entirely — e.g. using a set for O(1) lookups instead of an inner loop that searches linearly), the real-world cost can end up far below the theoretical worst case. As a general principle: reach for a nested loop when there's no simpler alternative, and keep an eye on whether the nesting is really necessary once your data gets large.

````python
numbers = [[1, 2, 3], [4, 5, 6]]



cnt = 0

for i in numbers:

    for j in i:

        print('iteration', cnt, end=': ')

        print(j)

        cnt = cnt + 1
````

## Summary

- A nested loop restarts and fully completes the inner loop for EVERY single outer-loop iteration.
- Total inner-body executions = outer iterations x inner iterations.
- `break`/`continue` inside a nested loop only ever affect the INNERMOST loop they're directly inside.
- List comprehensions can often express a simple nested `for` loop pair — and an optional filter — in one compact line.
- Any loop TYPE can nest inside any other: for-in-for, while-in-while, for-in-while, while-in-for.
- Unconditional nested loops cost O(n^2) — quadratic growth — so use nesting only when genuinely needed, and look for opportunities (early `break`, set-based lookups) to avoid the full cost in practice.

2. Rewrite exercise 1 as a one-line list comprehension.
3. Use a nested loop to find all Pythagorean triples `(a, b, c)` with `a < b < c <= 20` where `a**2 + b**2 == c**2`.
4. Print a hollow square of size 5x5 (stars on the border only, spaces inside) using nested `for` loops.

### Exercise 1 solution

````python
colors = ['red', 'blue']

sizes = ['S', 'M', 'L']

for color in colors:

    for size in sizes:

        print(f"{color}-{size}")
````

### Exercise 2 solution

````python
colors = ['red', 'blue']

sizes = ['S', 'M', 'L']

combos = [f"{color}-{size}" for color in colors for size in sizes]

print(combos)
````

### Exercise 3 solution

````python
triples = []

for a in range(1, 21):

    for b in range(a + 1, 21):

        for c in range(b + 1, 21):

            if a**2 + b**2 == c**2:

                triples.append((a, b, c))

print(triples)
````

### Exercise 4 solution

````python
n = 5

for i in range(n):

    for j in range(n):

        if i == 0 or i == n - 1 or j == 0 or j == n - 1:

            print("*", end="")

        else:

            print(" ", end="")

    print()
````

### Exercise 5 solution

````python
matrix = [[1, 2], [3, 4], [5, 6]]

found = False

for row_idx, row in enumerate(matrix):

    for col_idx, value in enumerate(row):

        if value > 3:

            print(f"First value > 3 is {value} at row {row_idx}, col {col_idx}")

            found = True

            break

    if found:

        break
````
