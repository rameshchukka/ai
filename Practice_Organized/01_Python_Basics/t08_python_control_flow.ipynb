# Python Control Flow Statements and Loops

**Control flow** is the order in which a program's statements actually execute at runtime — normally top to bottom, but conditionals and loops let you branch, repeat, or skip based on runtime conditions. Python's control flow tools fall into three categories:

1. **Conditional statements** — `if`, `if-else`, `if-elif-else`, nested `if` — run different code depending on whether a condition is true or false.
2. **Iterative statements** — `for`, `while` — repeat a block of code.
3. **Transfer statements** — `break`, `continue`, `pass` — alter a loop's normal flow.

This notebook covers all three with runnable examples. `for`, `while`, and `break`/`continue`/`pass` each get a much deeper dedicated notebook later in this series — this one gives the full overview plus the details of `if` in particular.

## The `if` Statement

The simplest control flow tool: evaluate a condition, and run the indented block only if that condition is `True`. If it's `False`, the block is skipped entirely and execution continues at the next line after the block.

```
if condition:
    statement 1
    statement 2
```

if statement flowchart

````python
number = 6

if number > 5:

    # Calculate square

    print(number * number)

print('Next lines of code')
````

## The `if-else` Statement

Adds an alternative: run one block if the condition is `True`, a different block if it's `False`. Exactly one of the two blocks ever runs.

## Chaining Multiple Conditions: `if-elif-else`

When you need to check several mutually-exclusive conditions in sequence, `elif` (short for 'else if') chains them together. Python checks each condition in order and runs the FIRST one that's true — everything after that is skipped, even if a later condition would also have been true.

```
if condition_1:
    statement 1
elif condition_2:
    statement 2
elif condition_3:
    statement 3
else:
    statement (fallback)
```

````python
def user_check(choice):

    if choice == 1:

        print("Admin")

    elif choice == 2:

        print("Editor")

    elif choice == 3:

        print("Guest")

    else:

        print("Wrong entry")



user_check(1)

user_check(2)

user_check(3)

user_check(4)
````

## Nested `if-else`

An `if` (or `if-else`) can contain another `if`/`if-else` inside its own block — indentation alone shows which condition each nested block belongs to. Useful when a decision genuinely depends on first narrowing down a broader case.

```
if outer_condition:
    if inner_condition:
        ...
    else:
        ...
else:
    ...
```

````python
# Simulates: num1 = int(input('Enter first number ')); num2 = int(input('Enter second number '))

num1 = 56

num2 = 15



if num1 >= num2:

    if num1 == num2:

        print(num1, 'and', num2, 'are equal')

    else:

        print(num1, 'is greater than', num2)

else:

    print(num1, 'is smaller than', num2)
````

## Single-Statement Suites

When an `if` (or loop) block is just ONE statement, Python lets you write it directly after the colon on the same line, instead of indenting it on the next line. This is legal but should be used sparingly — it reads well only for very short, simple blocks; anything more complex is clearer written the normal indented way.

````python
number = 56

if number > 0: print("positive")

else: print("negative")
````

````python
# Multiple statements on the same single-line suite, separated by semicolons

x = 1

while x <= 5: print(x, end=" "); x = x + 1
````

## `for` Loop (Overview)

Iterates over any sequence or iterable — string, list, dict, set, tuple, or a `range()`. Covered in full depth in its own dedicated notebook.

````python
for i in range(1, 11):

    print(i, end=" ")
````

## `while` Loop (Overview)

Repeats a block of code as long as a condition stays `True`, checked fresh at the START of every pass. Also covered in full depth in its own dedicated notebook.

````python
num = 10

total = 0

i = 1

while i <= num:

    total = total + i

    i = i + 1

print("Sum of first 10 numbers is:", total)
````

## `break` (Overview)

Exits the enclosing loop immediately, skipping any remaining iterations — useful for stopping as soon as you've found what you're looking for, instead of needlessly checking everything else.

````python
for num in range(10):

    if num > 5:

        print("stop processing.")

        break

    print(num)
````

## `continue` (Overview)

Skips just the REST of the current iteration and jumps straight to the next one — the loop itself keeps going.

````python
for num in range(3, 8):

    if num == 5:

        continue

    else:

        print(num)
````

## `pass` (Overview)

A genuine no-op — used as a placeholder wherever Python's syntax requires an indented block but you have nothing to put there yet (or, as here, deliberately want to do nothing).

````python
months = ['January', 'June', 'March', 'April']

for mon in months:

    pass

print(months)
````

## Summary

- **Conditional**: `if` runs a block only when true; `if-else` picks exactly one of two blocks; `elif` chains multiple mutually-exclusive checks; nesting lets you narrow decisions step by step.
- **Iterative**: `for` walks a known sequence; `while` repeats based on a condition — both get their own deep-dive notebooks.
- **Transfer**: `break` exits a loop early; `continue` skips to the next iteration; `pass` does nothing, as a syntactic placeholder.
- A single-statement block can go on the same line as its `:` — legal, but best kept for very short blocks only.

1. Write an `if-elif-else` chain that classifies a test score `s` into `"A"` (90+), `"B"` (80-89), `"C"` (70-79), or `"F"` (below 70). Test it with scores 95, 82, 71, and 50.
2. Write a nested `if` that checks whether a year is a leap year (divisible by 4, but if divisible by 100 must also be divisible by 400), printing a clear message for each case. Test with 2024, 1900, and 2000.

### Exercise 1 solution

````python
def grade(s):

    if s >= 90:

        return "A"

    elif s >= 80:

        return "B"

    elif s >= 70:

        return "C"

    else:

        return "F"



for score in [95, 82, 71, 50]:

    print(score, "->", grade(score))
````

### Exercise 2 solution

````python
def check_leap(year):

    if year % 4 == 0:

        if year % 100 == 0:

            if year % 400 == 0:

                print(year, "is a leap year (century, divisible by 400)")

            else:

                print(year, "is NOT a leap year (century, not divisible by 400)")

        else:

            print(year, "is a leap year")

    else:

        print(year, "is not a leap year")



check_leap(2024)

check_leap(1900)

check_leap(2000)
````

### Exercise 3 solution

````python
for num in range(1, 100):

    if num % 7 == 0 and num % 5 == 0:

        print("First number divisible by both 7 and 5:", num)

        break
````

### Exercise 4 solution

````python
for num in range(1, 21):

    if num % 3 == 0:

        continue

    print(num, end=" ")
````
