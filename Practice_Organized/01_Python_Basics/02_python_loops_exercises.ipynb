## Concepts you'll need

This set is entirely about **branching and looping** — deciding what runs, and how many times.

- **`if` / `elif` / `else`** — branch execution based on a condition.
- **`for` loops** — iterate over a known sequence, most often `range(start, stop, step)`. Remember `stop` is always exclusive, and a negative `step` counts backwards.
- **`while` loops** — repeat as long as a condition holds; you're responsible for updating whatever makes that condition eventually false.
- **`break` and `continue`** — `break` exits a loop immediately; `continue` skips the rest of the current iteration and moves to the next.
- **The `for...else` / `while...else` clause** — the `else` block runs only if the loop finishes naturally, without hitting a `break`. Handy for "searched everything, found nothing" logic.
- **Nested loops** — a loop inside another loop, typically an outer loop for rows and an inner loop for columns/repetitions, used heavily here for pattern printing and grid/matrix work.
- **Accumulator and state-tracking patterns** — a variable declared outside the loop (a running sum, a "previous value", a flag) that's updated on every pass to carry information from one iteration to the next.
- **Digit math** — `% 10` grabs the last digit of a number, `// 10` (floor division) drops it, letting you process a number digit-by-digit without converting it to a string.

Each exercise names the loop concept it's practicing, then gives a problem, a hint, a solution, and an explanation.

## Exercise 1. Print First 10 Natural Numbers

**Given:**

**Expected Output:**

**Hint:** Start a counter at 1, loop while it's <= 10, and remember to increment it inside the loop.

## Exercise 7. Filter a List With break and continue

**Concept:** break, continue, modulo

**Problem:** Walk through a list, printing only multiples of 5, skipping (not printing) any number above 150, and stopping entirely once a number exceeds 500.

**Given:**
```
numbers = [12, 75, 150, 180, 145, 525, 50]
```

**Expected Output:**
```
75
150
145
```

**Hint:** Check the 'stop' condition first with break, then the 'skip' condition with continue, then the actual filter.

````python
numbers = [12, 75, 150, 180, 145, 525, 50]

for item in numbers:
    if item > 500:
        break
    if item > 150:
        continue
    if item % 5 == 0:
        print(item)
````

**Explanation:** break ends the loop immediately once 525 is reached, so the trailing 50 is never even examined. continue jumps straight back to the next iteration when a number is over 150, skipping the print for that value only.

## Exercise 11. Reverse a String Without Slicing

**Concept:** string building, immutability

**Problem:** Reverse a string using a for loop, without using slice notation.

**Given:**
```

```

**Expected Output:**
```
Original: Python
Reversed: nohtyP
```

**Hint:** Build a new string by placing each character in front of what you've built so far, rather than appending to the end.

````python
original_str = "Python"
reversed_str = ""

for char in original_str:
    reversed_str = char + reversed_str

print(f"Original: {original_str}")
print(f"Reversed: {reversed_str}")
````

**Explanation:** Because strings are immutable, char + reversed_str creates a brand-new string on every pass. Prepending instead of appending is what flips the order — each new character pushes in front of the ones already collected.

## Exercise 12. Count Vowels and Consonants

**Concept:** membership test, str.isalpha()

**Problem:** Count the vowels and consonants in a sentence, ignoring spaces and punctuation.

**Given:**
```

```

**Expected Output:**
```
Vowels: 5
Consonants: 7
```

**Hint:** Lowercase the sentence first, use .isalpha() to skip non-letters, then test membership against a string of vowels.

````python
sentence = "Loops are Fun!"
vowels = "aeiou"
v_count = 0
c_count = 0

for char in sentence.lower():
    if char.isalpha():
        if char in vowels:
            v_count += 1
        else:
            c_count += 1

print(f"Vowels: {v_count}")
print(f"Consonants: {c_count}")
````

**Explanation:** .lower() lets one comparison handle both cases of a letter. .isalpha() filters out spaces and punctuation before the vowel/consonant classification even happens, keeping the counts accurate.

## Exercise 14. Reverse an Integer Mathematically

**Concept:** modulo + floor division combo

**Problem:** Reverse the digits of an integer using arithmetic rather than string slicing.

**Given:**
```
num = 76542
```

**Expected Output:**
```
24567
```

**Hint:** Peel off the last digit with % 10, then build the reversed number by shifting it left (x10) and adding that digit.

````python
num = 76542
reverse_number = 0

while num > 0:
    digit = num % 10
    reverse_number = (reverse_number * 10) + digit
    num = num // 10

print(reverse_number)
````

**Explanation:** num % 10 isolates the last digit. Multiplying reverse_number by 10 shifts its existing digits one place to the left, making room to add the newly-extracted digit at the end — effectively building the number in reverse.

## Exercise 15. Largest and Smallest Digit in a Number

**Concept:** digit extraction + running min/max

**Problem:** Find the largest and the smallest digit that appear in a given number.

**Given:**
```
num = 75869
```

**Expected Output:**
```
Largest digit: 9
Smallest digit: 5
```

**Hint:** Start largest at 0 and smallest at 9 so the first digit checked updates both, then compare each digit as it's extracted.

````python
num = 75869
largest = 0
smallest = 9

while num > 0:
    digit = num % 10
    if digit > largest:
        largest = digit
    if digit < smallest:
        smallest = digit
    num = num // 10

print("Largest digit:", largest)
print("Smallest digit:", smallest)
````

**Explanation:** Initializing largest at the lowest possible digit and smallest at the highest guarantees the very first digit examined replaces both. Each subsequent digit only replaces a tracker if it beats the current record.

## Exercise 16. Palindrome Number Check via Reversal

**Concept:** save-before-mutate pattern

**Problem:** Check whether a number is a palindrome by mathematically reversing it and comparing.

**Given:**
```
number = 121
```

**Expected Output:**
```
Yes. Given number is palindrome number
```

**Hint:** Save the original number before the loop destroys it, then compare that saved copy to the reversed result.

````python
number = 121
temp = number
reverse_num = 0

while number > 0:
    digit = number % 10
    reverse_num = (reverse_num * 10) + digit
    number = number // 10

if temp == reverse_num:
    print("Yes. Given number is palindrome number")
else:
    print("No. Given number is not palindrome")
````

**Explanation:** Since the reversal loop reduces number down to 0, its original value has to be stored in temp beforehand so there's still something to compare the final reversed result against.

## Exercise 17. Factorial via a Loop

**Concept:** iterative multiplication

**Problem:** Compute the factorial of a number using a loop, with guards for negative numbers and zero.

**Given:**
```
num = 5
```

**Expected Output:**
```
The factorial of 5 is 120
```

**Hint:** Start an accumulator at 1 and multiply it by every number from 1 to num.

````python
num = 5
factorial = 1

if num < 0:
    print("Factorial does not exist for negative numbers")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print(f"The factorial of {num} is {factorial}")
````

**Explanation:** factorial starts at 1 (the multiplicative identity) and each loop pass multiplies in the next integer, producing 1x2x3x4x5 = 120 by the time the loop ends.

## Exercise 18. Collatz Sequence

**Concept:** indeterminate while loop

**Problem:** Starting from a number, repeatedly halve it if even or turn it into 3n+1 if odd, printing the sequence until it reaches 1.

**Given:**
```
n = 6
```

**Expected Output:**
```
6, 3, 10, 5, 16, 8, 4, 2, 1
```

**Hint:** Loop while n != 1, applying the even/odd rule and updating n each pass.

````python
n = 6
print(n, end="")

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = (3 * n) + 1
    print(f", {n}", end="")
````

**Explanation:** Unlike a for loop with a known number of steps, this while loop's length isn't known ahead of time — it simply keeps applying the rule until the mathematical condition n == 1 is finally reached.

## Exercise 19. Armstrong Number Check

**Concept:** digit power sum

**Problem:** Check whether a number equals the sum of its own digits, each raised to the power of the digit count.

**Given:**
```
num = 153
```

**Expected Output:**
```
153 is an Armstrong number
```

**Hint:** Convert the number to a string to loop over its digits easily, using the string's length as the power.

````python
num = 153
num_str = str(num)
power = len(num_str)
total = 0

for digit in num_str:
    total += int(digit) ** power

if total == num:
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
````

**Explanation:** Turning the number into a string makes it easy to loop over individual digit characters. Each character is converted back to an int, raised to the digit-count power, and summed; if that sum equals the original number, it's an Armstrong number.

## Exercise 24. Hollow Square Pattern

**Problem:** Print a square outline of stars with an empty interior.

```

```

````python
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````

## Exercise 25. Diamond-ish Star Pyramid

**Concept:** two sequential loops for a symmetric shape

**Problem:** Print a pattern that grows to 5 stars wide and then shrinks back down.

**Given:**
```
rows = 5
```

**Expected Output:**
```
*
* *
* * *
* * * *
* * * * *
* * * *
* * *
* *
*
```

**Hint:** Use one loop to print the growing top half, then a separate loop counting backwards for the shrinking bottom half.

````python
rows = 5

for i in range(0, rows):
    for j in range(0, i + 1):
        print("*", end=' ')
    print("\r")

for i in range(rows - 1, 0, -1):
    for j in range(0, i):
        print("*", end=' ')
    print("\r")
````

**Explanation:** The first block is an ordinary growing triangle (1 up to 5 stars). The second block runs the same idea in reverse, starting at rows - 1 and counting down, to shrink the pattern back to a single star.

## Exercise 28. Filter a Dictionary by Value Threshold

**Concept:** dict.items() iteration

**Problem:** Given a dictionary of scores, build a new dictionary containing only entries above a threshold.

**Given:**
```
scores = {"Alice": 85, "Bob": 70, "Charlie": 95, "David": 60}, threshold = 75
```

**Expected Output:**
```
Passing Students: {'Alice': 85, 'Charlie': 95}
```

**Hint:** Loop with for name, score in scores.items() and only add entries whose score clears the threshold.

````python
scores = {"Alice": 85, "Bob": 70, "Charlie": 95, "David": 60}
passing_students = {}
threshold = 75

for name, score in scores.items():
    if score >= threshold:
        passing_students[name] = score

print(f"Passing Students: {passing_students}")
````

**Explanation:** .items() unpacks each dictionary entry into its key and value at once, so the loop can inspect the score while still knowing which name it belongs to. Only entries clearing the threshold get copied into the new dictionary.

## Exercise 31. Segregate Evens to Front, Odds to Back

**Concept:** list partitioning

**Problem:** Reorder a list so all even numbers come first, followed by all the odd numbers, in their original relative order.

**Given:**
```
[1, 2, 3, 4, 5, 6]
```

**Expected Output:**
```
Segregated List: [2, 4, 6, 1, 3, 5]
```

**Hint:** Sort numbers into two separate lists as you loop, then concatenate evens and odds at the end.

````python
numbers = [1, 2, 3, 4, 5, 6]
evens = []
odds = []

for n in numbers:
    if n % 2 == 0:
        evens.append(n)
    else:
        odds.append(n)

result = evens + odds
print("Segregated List:", result)
````

**Explanation:** Each number is routed into exactly one of the two lists based on its remainder mod 2. Concatenating evens + odds joins them so all evens appear before any odd number.

## Exercise 32. Rotate a List Left by k (Loop Version)

**Concept:** list.pop() and append() as a shift

**Problem:** Rotate a list to the left by k positions using repeated pop/append operations rather than slicing.

**Given:**
```
nums = [1, 2, 3, 4, 5], k = 2
```

**Expected Output:**
```
Rotated List: [3, 4, 5, 1, 2]
```

**Hint:** Repeat 'remove the first item, then append it at the end' exactly k times.

````python
nums = [1, 2, 3, 4, 5]
k = 2

for _ in range(k):
    first_element = nums.pop(0)
    nums.append(first_element)

print("Rotated List:", nums)
````

**Explanation:** pop(0) removes the front item and shifts everything else left by one. Appending that removed item at the end completes a single left-rotation; doing this k times produces a rotation by k positions. The underscore _ signals the loop variable itself isn't needed.

## Exercise 33. Word Frequency Counter

**Concept:** dictionary counting

**Problem:** Count how often each word occurs in a piece of text.

**Given:**
```
text = "apple banana apple orange banana apple"
```

**Expected Output:**
```
{'apple': 3, 'banana': 2, 'orange': 1}
```

**Hint:** Split the text into words, then check membership in the dictionary to decide whether to start or increment a count.

````python
text = "apple banana apple orange banana apple"
words = text.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
````

## Exercise 35. Perfect Number Check

**Concept:** divisor summation, loop-range optimization

**Problem:** Check whether a number equals the sum of its own proper divisors.

**Given:**
```
num = 28
```

**Expected Output:**
```
28 is a Perfect Number
```

**Hint:** No proper divisor can exceed half the number, so you only need to check up to num // 2.

````python
num = 28
divisor_sum = 0

for i in range(1, (num // 2) + 1):
    if num % i == 0:
        divisor_sum += i

if divisor_sum == num:
    print(f"{num} is a Perfect Number")
else:
    print(f"{num} is not a Perfect Number")
````

**Explanation:** Limiting the search to num // 2 avoids wasted checks, since no proper divisor of a number can be bigger than half of it. Every value that divides evenly gets added to divisor_sum, which is then compared back against the original number.

## Exercise 36. Binary String to Decimal (Manual)

**Concept:** positional notation, powers of two

**Problem:** Convert a binary string to its decimal value using a loop, without int(binary, 2).

**Given:**
```
binary_str = "1101"
```

**Expected Output:**
```
Decimal value: 13
```

**Hint:** Reverse the string so each character's index lines up with its power of two, then sum 2**index wherever there's a '1'.

````python
binary_str = "1101"
decimal_val = 0

reversed_binary = binary_str[::-1]

for i in range(len(reversed_binary)):
    if reversed_binary[i] == '1':
        decimal_val += 2 ** i

print("Decimal value:", decimal_val)
````

**Explanation:** Reversing the string lines up index 0 with the rightmost (least-significant) bit, so index i directly corresponds to the power 2**i. Summing 2**i for every '1' bit reconstructs the decimal value: 1*1 + 0*2 + 1*4 + 1*8 = 13.

## Exercise 37. Prime Numbers in a Range

**Concept:** nested loops with for...else

**Problem:** List every prime number within a given range.

**Given:**
```
start = 25, end = 50
```

**Expected Output:**
```
29
31
37
41
43
47
```

**Hint:** For each candidate number, try dividing it by every smaller number; if none divide evenly, it's prime.

````python
start = 25
end = 50

print(f"Prime numbers between {start} and {end} are:")

for num in range(start, end + 1):
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)
````

**Explanation:** The inner loop tries every possible divisor; break fires the moment a factor is found, disqualifying that number. The inner loop's else only runs when break was never triggered — meaning no divisor was found, so the number is prime.

## Exercise 39. Flatten a Nested List (Loop Version)

```
nested_list = [[10, 20], [30, 40], [50, 60]]
```

## Exercise 40. Find Coordinates in a 2D Matrix

**Concept:** enumerate() in nested loops

**Problem:** Given a 2D list, find the row and column index where a target value appears.

**Given:**
```
matrix = [[10, 20], [30, 40], [50, 60]], target = 30
```

**Expected Output:**
```
Target 30 found at Row: 1, Column: 0
```

**Hint:** Use enumerate() on both the outer and inner loops to track row and column indices simultaneously.

````python
matrix = [
    [10, 20],
    [30, 40],
    [50, 60]
]
target = 30

for r_idx, row in enumerate(matrix):
    for c_idx, val in enumerate(row):
        if val == target:
            print(f"Target {target} found at Row: {r_idx}, Column: {c_idx}")
            break
````

**Explanation:** enumerate() hands back both an index and a value together, avoiding the clumsier range(len(...)) pattern. The outer loop tracks the row index, the inner loop tracks the column index, and break exits early once the single match is found.
