## Concepts you'll need

This set uses a real company sales dataset to practice **Matplotlib**, Python's foundational 2D plotting library, usually paired with pandas for data loading.

- **The basic workflow** — call a plotting function (`plt.plot()`, `plt.bar()`, `plt.scatter()`, etc.) to draw, then `plt.xlabel()`/`plt.ylabel()`/`plt.title()` to label, then `plt.show()` to display. Everything between plot calls and `show()` accumulates onto the same figure.
- **Line plots — `plt.plot(x, y, ...)`** — the default chart type; style keywords include `color`, `linestyle` (`'--'` dashed, `':'` dotted), `linewidth`, `marker` (`'o'` circle), and `markerfacecolor`. Multiple `plt.plot()` calls before one `plt.show()` draw multiple lines on the same axes — a "multiline plot."
- **`plt.legend(loc=...)`** — shows a legend built from each plot call's `label=` argument; `loc` positions it (`'upper left'`, `'lower right'`, etc.).
- **`plt.xticks(values)` / `plt.yticks(values)`** — explicitly control which tick marks and labels appear on each axis, rather than leaving it to Matplotlib's automatic choices.
- **Other chart types** — `plt.scatter(x, y)` (unconnected points, good for showing individual data point spread), `plt.bar(x, height, width=...)` (bars; plotting two bars per x-position with offset `x` values, like `x - 0.25` and `x + 0.25`, produces grouped/side-by-side bars), `plt.hist(data, bins)` (frequency distribution — how many values fall in each range), `plt.pie(values, labels=..., autopct=...)` (proportions as percentages of a whole; `autopct='%1.1f%%'` formats each slice's percentage to one decimal place), `plt.stackplot(x, y1, y2, ...)` (layers cumulative areas on top of each other).
- **`plt.grid(True, linestyle=...)`** — adds gridlines to make reading values off the chart easier.
- **`plt.subplots(nrows, sharex=True)`** — creates multiple stacked (or gridded) charts in one figure, sharing one x-axis; returns a figure and an array of axes objects (`axarr[0]`, `axarr[1]`, ...) that each get their own `.plot()` and `.set_title()` calls instead of the top-level `plt.*` functions.
- **`plt.savefig(path, dpi=...)`** — saves the current figure to an image file; must be called **before** `plt.show()`, since `show()` can clear the figure afterward on some backends.
- **The stackplot legend trick** — `plt.stackplot()` doesn't generate its own legend entries, so the standard workaround is plotting empty, invisible lines (`plt.plot([], [], color=..., label=...)`) purely to register legend entries with matching colors before calling `stackplot()`.

## Exercise 1. Line Plot of Total Profit

**Concept:** the basic plt.plot() workflow

**Problem:** Plot total profit for each month as a line chart, with month number on the x-axis and profit on the y-axis.

**Given:**
```
company_sales_data.csv, 12 months of data
```

**Expected Output:**
```
A line chart rising and falling across the 12 months, peaking around months 11-12
```

**Hint:** plt.xticks(monthList) forces every month number to get its own tick mark, rather than relying on automatic tick selection.

````python
import pandas as pd
import matplotlib.pyplot as plt

# The dataset embedded directly so this notebook is fully self-contained
csv_data = """month_number,facecream,facewash,toothpaste,bathingsoap,shampoo,moisturizer,total_units,total_profit
1,2500,1500,5200,9200,1200,1500,21100,211000
2,2630,1200,5100,6100,2100,1200,18330,183300
3,2140,1340,4550,9550,3550,1340,22470,224700
4,3400,1130,5870,8870,1870,1130,22270,222700
5,3600,1740,4560,7760,1560,1740,20960,209600
6,2760,1555,4890,7490,1890,1555,20140,201400
7,2980,1120,4780,8980,1780,1120,29550,295500
8,3700,1400,5860,9960,2860,1400,36140,361400
9,3540,1780,6100,8100,2100,1780,23400,234000
10,1990,1890,8300,10300,2300,1890,26670,266700
11,2340,2100,7300,13300,2400,2100,41280,412800
12,2900,1760,7400,14400,1800,1760,30020,300200
"""

with open("company_sales_data.csv", "w") as f:
    f.write(csv_data)

df = pd.read_csv("company_sales_data.csv")
profitList = df['total_profit'].tolist()
monthList = df['month_number'].tolist()

plt.plot(monthList, profitList, label='Month-wise Profit data of last year')
plt.xlabel('Month number')
plt.ylabel('Profit in dollar')
plt.xticks(monthList)
plt.title('Company profit per month')
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.legend(loc='upper left')
plt.show()
````

**Explanation:** df['total_profit'].tolist() converts a single DataFrame column into a plain Python list, which plt.plot() accepts directly as x/y data. plt.xticks(monthList) forces exactly one tick per month (1 through 12) instead of Matplotlib's automatically-spaced default ticks, and plt.yticks([...]) similarly sets specific dollar-amount gridline positions on the y-axis for a cleaner, more readable scale.

## Exercise 2. Styled Line Plot

**Concept:** line style keywords: color, linestyle, marker, linewidth

**Problem:** Plot total profit with a dashed red line, circle markers, a specific legend location, and a thicker line width.

**Given:**
```
the same profit data as Exercise 1
```

**Expected Output:**
```
A dashed red line with circular markers, thicker than the default line width
```

**Hint:** markerfacecolor sets the FILL color of each marker, separate from the line's own color.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
profitList = df['total_profit'].tolist()
monthList = df['month_number'].tolist()

plt.plot(monthList, profitList, label='Profit data of last year',
          color='r', marker='o', markerfacecolor='k',
          linestyle='--', linewidth=3)

plt.xlabel('Month Number')
plt.ylabel('Profit in dollar')
plt.legend(loc='lower right')
plt.title('Company Sales data of last year')
plt.xticks(monthList)
plt.yticks([100000, 200000, 300000, 400000, 500000])
plt.show()
````

**Explanation:** color='r' sets the line itself to red, while marker='o' adds a circular marker at each actual data point, and markerfacecolor='k' fills those circles black — the line color and marker fill color are independent settings. linestyle='--' switches from a solid line to dashed, and linewidth=3 makes the line noticeably thicker than Matplotlib's default. loc='lower right' positions the legend box in that specific corner of the plot, away from where the data itself peaks.

## Exercise 3. Multiline Plot of All Products

**Concept:** multiple plt.plot() calls sharing one figure

**Problem:** Plot units sold per month for all six products as six separate lines on one chart.

**Given:**
```
facecream, facewash, toothpaste, bathingsoap, shampoo, moisturizer columns
```

**Expected Output:**
```
Six distinctly-colored, distinctly-labeled lines, one per product
```

**Hint:** Each product needs its OWN correct label — a common copy-paste mistake is reusing one product's label for all the lines.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
monthList = df['month_number'].tolist()
faceCremSalesData = df['facecream'].tolist()
faceWashSalesData = df['facewash'].tolist()
toothPasteSalesData = df['toothpaste'].tolist()
bathingsoapSalesData = df['bathingsoap'].tolist()
shampooSalesData = df['shampoo'].tolist()
moisturizerSalesData = df['moisturizer'].tolist()

plt.plot(monthList, faceCremSalesData,   label='Face cream Sales Data',   marker='o', linewidth=3)
plt.plot(monthList, faceWashSalesData,   label='Face Wash Sales Data',    marker='o', linewidth=3)
plt.plot(monthList, toothPasteSalesData, label='ToothPaste Sales Data',   marker='o', linewidth=3)
plt.plot(monthList, bathingsoapSalesData, label='Bathing Soap Sales Data', marker='o', linewidth=3)
plt.plot(monthList, shampooSalesData,     label='Shampoo Sales Data',      marker='o', linewidth=3)
plt.plot(monthList, moisturizerSalesData, label='Moisturizer Sales Data',  marker='o', linewidth=3)

plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.legend(loc='upper left')
plt.xticks(monthList)
plt.title('Sales data')
plt.show()
````

**Explanation:** Six separate plt.plot() calls, made before a single plt.show(), all accumulate onto the same set of axes — this is what makes it a 'multiline' plot rather than six separate charts. Each call needs its OWN distinct label= matching the product it actually represents; the original source exercise had a copy-paste bug labeling the bathing soap, shampoo, and moisturizer lines all as 'ToothPaste Sales Data', which is corrected here so the legend accurately identifies all six lines.

## Exercise 4. Scatter Plot of Toothpaste Sales

**Concept:** plt.scatter() for unconnected individual data points

**Problem:** Show toothpaste sales as a scatter plot (individual points, not connected by lines), with a dashed grid.

**Given:**
```
the toothpaste column
```

**Expected Output:**
```
12 individual scattered points, one per month, with dashed gridlines
```

**Hint:** grid()'s linestyle parameter accepts the same style codes as plot()'s linestyle: '--' for dashed.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
monthList = df['month_number'].tolist()
toothPasteSalesData = df['toothpaste'].tolist()

plt.scatter(monthList, toothPasteSalesData, label='Tooth paste Sales data')
plt.xlabel('Month Number')
plt.ylabel('Number of units Sold')
plt.legend(loc='upper left')
plt.title('Tooth paste Sales data')
plt.xticks(monthList)
plt.grid(True, linewidth=1, linestyle="--")
plt.show()
````

**Explanation:** plt.scatter() plots each (month, sales) pair as an independent point with no connecting line between them, which is more appropriate than plt.plot() when you want to emphasize individual data points rather than a continuous trend. plt.grid(True, linewidth=1, linestyle="--") overlays light dashed gridlines across the plot, making it easier to read off approximate values for any given point.

## Exercise 5. Grouped Bar Chart of Face Cream vs. Face Wash

**Concept:** side-by-side bars via offset x-positions

**Problem:** Show face cream and face wash sales as two side-by-side bars per month.

**Given:**
```
the facecream and facewash columns
```

**Expected Output:**
```
Two bars per month, side by side, one color per product
```

**Hint:** Offsetting one bar's x-positions by -0.25 and the other's by +0.25 (with matching widths) places them side by side instead of overlapping.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
monthList = df['month_number'].tolist()
faceCremSalesData = df['facecream'].tolist()
faceWashSalesData = df['facewash'].tolist()

plt.bar([a - 0.25 for a in monthList], faceCremSalesData, width=0.25, label='Face Cream sales data', align='edge')
plt.bar([a + 0.25 for a in monthList], faceWashSalesData, width=-0.25, label='Face Wash sales data', align='edge')
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.legend(loc='upper left')
plt.xticks(monthList)
plt.grid(True, linewidth=1, linestyle="--")
plt.title('Facewash and facecream sales data')
plt.show()
````

**Explanation:** Without any offset, two plt.bar() calls at the same x-positions would draw directly on top of each other. Shifting the face cream bars left by 0.25 and the face wash bars right by 0.25 — with matching bar widths of 0.25 — places them side by side at each month instead. align='edge' anchors each bar's edge (rather than its center) at the given x-position, which combined with the negative width on the second call (width=-0.25) makes it grow leftward back toward the tick mark, completing the paired layout.

## Exercise 6. Bar Chart of Bathing Soap Sales, Saved to Disk

**Concept:** plt.savefig() to export a chart as an image file

**Problem:** Show bathing soap sales as a bar chart and save it to disk as a PNG file.

**Given:**
```
the bathingsoap column
```

**Expected Output:**
```
A bar chart, plus a saved file 'sales_data_of_bathingsoap.png' on disk
```

**Hint:** savefig() must be called BEFORE show() — on some backends, show() can clear the current figure afterward.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
monthList = df['month_number'].tolist()
bathingsoapSalesData = df['bathingsoap'].tolist()

plt.bar(monthList, bathingsoapSalesData)
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.xticks(monthList)
plt.grid(True, linewidth=1, linestyle="--")
plt.title('Bathing soap sales data')
plt.savefig('sales_data_of_bathingsoap.png', dpi=150)
plt.show()

import os
print("Saved file exists:", os.path.exists('sales_data_of_bathingsoap.png'))
print("File size:", os.path.getsize('sales_data_of_bathingsoap.png'), "bytes")
````

**Explanation:** plt.savefig(path, dpi=150) writes the CURRENT figure to disk as an image file — the format (PNG here) is inferred automatically from the file extension. dpi=150 sets the resolution (dots per inch); higher values produce a larger, higher-quality image file. The call must happen BEFORE plt.show(), since on some Matplotlib backends, show() can clear or close the figure afterward, leaving nothing left to save.

## Exercise 7. Histogram of Total Profit

**Concept:** plt.hist() — frequency distribution across value ranges (bins)

**Problem:** Show the distribution of monthly total profit values as a histogram, grouped into custom profit ranges.

**Given:**
```
the total_profit column
```

**Expected Output:**
```
A histogram showing how many months fall into each profit range bucket
```

**Hint:** Unlike a bar chart (one bar per category), a histogram's bins group a CONTINUOUS range of values — the profit_range list defines the bin EDGES.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
profitList = df['total_profit'].tolist()

profit_range = [150000, 175000, 200000, 225000, 250000, 300000, 350000, 425000]
plt.hist(profitList, profit_range, label='Profit data')
plt.xlabel('profit range in dollar')
plt.ylabel('Number of months in that range')
plt.legend(loc='upper left')
plt.xticks(profit_range, rotation=45)
plt.title('Profit data')
plt.show()
````

**Explanation:** Unlike plt.bar(), where each bar represents one distinct, pre-existing category, plt.hist() instead sorts continuous numeric values into ranges called bins, and each bar's height shows how many values fell within that bin's range. Passing profit_range as the second argument defines the exact bin EDGES rather than letting Matplotlib choose automatic bin boundaries, giving fine control over how the profit values are grouped. (The bin edges are extended slightly beyond the source page's original list, which stopped just short of covering the actual maximum profit value in the data, so every month is correctly captured in a bin.)

## Exercise 8. Pie Chart of Total Yearly Sales per Product

**Concept:** plt.pie() with autopct for percentage labels

**Problem:** Show each product's total units sold for the year as a percentage slice of a pie chart.

**Given:**
```
the sum of each product column across all 12 months
```

**Expected Output:**
```
A pie chart with 6 slices, each labeled with its product name and percentage of the whole
```

**Hint:** autopct='%1.1f%%' formats each slice's computed percentage to one decimal place, with a literal % sign appended.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")

labels = ['FaceCream', 'FaceWash', 'ToothPaste', 'Bathing soap', 'Shampoo', 'Moisturizer']
salesData = [df['facecream'].sum(), df['facewash'].sum(), df['toothpaste'].sum(),
             df['bathingsoap'].sum(), df['shampoo'].sum(), df['moisturizer'].sum()]

plt.axis("equal")
plt.pie(salesData, labels=labels, autopct='%1.1f%%')
plt.legend(loc='lower right')
plt.title('Sales data')
plt.show()
````

**Explanation:** .sum() on each product column collapses all 12 monthly values down to one yearly total, giving plt.pie() the six values it needs to compute proportions from. plt.axis("equal") forces the plot's x and y scales to match, which is what keeps the pie perfectly circular instead of stretched into an oval. autopct='%1.1f%%' is a printf-style format string: %1.1f formats the auto-computed percentage to one decimal place, and the trailing %% escapes to a literal percent sign in the displayed label — so each slice shows something like '15.2%'.

## Exercise 9. Subplots: Bathing Soap and Face Wash Stacked

**Concept:** plt.subplots() for multiple stacked charts sharing an x-axis

**Problem:** Show bathing soap and face wash sales as two separate but vertically-stacked charts sharing one x-axis.

**Given:**
```
the bathingsoap and facewash columns
```

**Expected Output:**
```
Two stacked line charts, one above the other, sharing the same x-axis scale
```

**Hint:** axarr[0] and axarr[1] are individual Axes objects, each with its own .plot() and .set_title() — different from the top-level plt.plot().

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
monthList = df['month_number'].tolist()
bathingsoap = df['bathingsoap'].tolist()
faceWashSalesData = df['facewash'].tolist()

f, axarr = plt.subplots(2, sharex=True)
axarr[0].plot(monthList, bathingsoap, label='Bathingsoap Sales Data', color='k', marker='o', linewidth=3)
axarr[0].set_title('Sales data of a Bathingsoap')
axarr[1].plot(monthList, faceWashSalesData, label='Face Wash Sales Data', color='r', marker='o', linewidth=3)
axarr[1].set_title('Sales data of a facewash')

plt.xticks(monthList)
plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.show()
````

**Explanation:** plt.subplots(2, sharex=True) creates a figure containing 2 separate chart areas (Axes) stacked vertically, and sharex=True locks both to the identical x-axis range and tick marks, so panning or zooming one would move the other in sync. axarr is an array of those Axes objects — axarr[0] and axarr[1] each get their own independent .plot() and .set_title() calls, distinct from the module-level plt.plot() used everywhere else in this notebook, since here we're addressing each subplot specifically rather than 'whatever the current active plot is'.

## Exercise 10. Stack Plot of All Product Sales

**Concept:** plt.stackplot() with the empty-line legend trick

**Problem:** Show all six products' monthly sales as a stacked area chart, with each product visually layered on top of the others.

**Given:**
```
all six product columns
```

**Expected Output:**
```
A stacked area chart where each product's band sits on top of the cumulative total of the products before it
```

**Hint:** stackplot() itself generates no legend entries — the workaround is plotting invisible empty lines with matching colors purely to register legend labels.

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("company_sales_data.csv")
monthList = df['month_number'].tolist()

faceCremSalesData = df['facecream'].tolist()
faceWashSalesData = df['facewash'].tolist()
toothPasteSalesData = df['toothpaste'].tolist()
bathingsoapSalesData = df['bathingsoap'].tolist()
shampooSalesData = df['shampoo'].tolist()
moisturizerSalesData = df['moisturizer'].tolist()

plt.plot([], [], color='m', label='Face Cream', linewidth=5)
plt.plot([], [], color='c', label='Face wash', linewidth=5)
plt.plot([], [], color='r', label='Tooth paste', linewidth=5)
plt.plot([], [], color='k', label='Bathing soap', linewidth=5)
plt.plot([], [], color='g', label='Shampoo', linewidth=5)
plt.plot([], [], color='y', label='Moisturizer', linewidth=5)

plt.stackplot(monthList, faceCremSalesData, faceWashSalesData, toothPasteSalesData,
              bathingsoapSalesData, shampooSalesData, moisturizerSalesData,
              colors=['m', 'c', 'r', 'k', 'g', 'y'])

plt.xlabel('Month Number')
plt.ylabel('Sales units in Number')
plt.title('All product sales data using stack plot')
plt.legend(loc='upper left')
plt.show()
````

**Explanation:** plt.stackplot() layers each product's values as a colored band stacked on top of the cumulative total of the bands below it, making the chart's total height at any month represent the sum across all six products. Crucially, stackplot() generates NO legend entries of its own, which is why six plt.plot([], [], color=..., label=..., linewidth=5) calls come first — each plots literally zero data points (empty x and y lists), so nothing is drawn, but each still registers one colored legend entry, deliberately matching the colors= list passed to the actual stackplot() call afterward. The result is a chart with an accurate, correctly-colored legend despite stackplot() itself being unable to produce one directly.
