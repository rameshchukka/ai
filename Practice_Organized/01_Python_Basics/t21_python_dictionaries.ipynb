# Python Dictionaries

A **dictionary** stores data as KEY-VALUE pairs — think of a real dictionary, where a word (key) maps to its definition (value). Looking something up by key is fast and direct, unlike a list where you'd have to search item by item.

- **Ordered** (Python 3.7+) — dictionaries remember insertion order; earlier Python 3.6 and below did not guarantee this.
- **Unique keys** — assigning to an existing key OVERWRITES its value rather than adding a duplicate entry.
- **Mutable** — items can be added, removed, or changed after creation.
- Keys must be immutable/hashable (string, number, tuple); values can be absolutely anything, including duplicates.

## Creating a Dictionary

Curly braces with `key: value` pairs, the `dict()` constructor, or from a sequence of (key, value) pairs.

```python
person = {"name": "Jessa", "country": "USA", "telephone": 1178}

print(person)



person2 = dict({"name": "Jessa", "country": "USA", "telephone": 1178})

print(person2)



person3 = dict([("name", "Mark"), ("country", "USA"), ("telephone", 1178)])

print(person3)



sample_dict = {"name": "Jessa", 10: "Mobile"}   # mixed key types are fine

print(sample_dict)



person4 = {"name": "Jessa", "telephones": [1178, 2563, 4569]}   # a value can be a list

print(person4)
```

### Empty Dictionary

```python
emptydict = {}

print(type(emptydict))
```

## Accessing Elements

By key inside `[]`, or via `.get(key)`. The key difference: `[]` raises `KeyError` for a missing key, while `.get()` returns `None` (or a default you specify) instead — safer when a key might not exist.

```python
person = {"name": "Jessa", "country": "USA", "telephone": 1178}



print(person['name'])

print(person.get('telephone'))

print(person.get('zip'))              # missing key -> None, no error

print(person.get('zip', 'N/A'))       # missing key -> your own default
```

```python
try:

    print(person['zip'])

except KeyError as e:

    print("KeyError:", e)
```

### Getting All Keys, Values, or Items



```python
person = {"name": "Jessa", "country": "USA", "telephone": 1178}



print(person.keys())

print(type(person.keys()))



print(person.values())

print(type(person.values()))



print(person.items())

print(type(person.items()))
```

### Iterating a Dictionary

```python
person = {"name": "Jessa", "country": "USA", "telephone": 1178}



print('key', ':', 'value')

for key in person:

    print(key, ':', person[key])
```

```python
person = {"name": "Jessa", "country": "USA", "telephone": 1178}



for key_value in person.items():

    print(key_value[0], key_value[1])



# more idiomatic: unpack directly in the for statement

for key, value in person.items():

    print(key, value)
```

## Length of a Dictionary

```python
person = {"name": "Jessa", "country": "USA", "telephone": 1178}

print(len(person))
```

## Adding Items

Direct key assignment for ONE new key; `.update()` for one OR several at once.

```python
person = {"name": "Jessa", 'country': "USA", "telephone": 1178}



person["weight"] = 50

person.update({"height": 6})

print(person)
```

```python
person = {"name": "Jessa", 'country': "USA"}



person.update({"weight": 50, "height": 6})

print(person)



person.update([("city", "Texas"), ("company", "Google")])   # a list of tuples works too

print(person)
```

### `setdefault()` — Add Only If Missing

If the key already exists, `setdefault()` just returns its current value, unchanged. If the key is missing, it INSERTS the key with the given default (or `None` if no default given) and returns that.

```python
person_details = {"name": "Jessa", "country": "USA", "telephone": 1178}



person_details.setdefault('state', 'Texas')     # key missing -> inserted

person_details.setdefault("zip")                # key missing, no default -> inserted as None

person_details.setdefault('country', 'Canada')  # key EXISTS -> unchanged, 'Canada' is ignored



for key, value in person_details.items():

    print(key, ':', value)
```

## Modifying Values

Direct key assignment, or `.update()` with the existing key.

```python
person = {"name": "Jessa", "country": "USA"}



person["country"] = "Canada"

print(person['country'])



person.update({"country": "USA"})

print(person['country'])
```

## Removing Items

| Method | Description |
|---|---|
| `pop(key)` | Removes AND returns the value for `key`; `KeyError` if missing |
| `popitem()` | Removes AND returns the LAST-inserted (key, value) pair |
| `del d[key]` | Removes a specific key |
| `clear()` | Empties the dictionary |
| `del d` | Deletes the dictionary variable itself |

```python
person = {'name': 'Jessa', 'country': 'USA', 'telephone': 1178, 'weight': 50, 'height': 6}



deleted_item = person.popitem()

print(deleted_item)

print(person)



deleted_item = person.pop('telephone')

print(deleted_item)

print(person)



del person['weight']

print(person)



person.clear()

print(person)
```

## Checking If a Key Exists

`in` checks against the dictionary's KEYS by default — `key in some_dict` works directly, `key in some_dict.keys()` is equivalent but more explicit/verbose.

```python
person = {'name': 'Jessa', 'country': 'USA', 'telephone': 1178}



key_name = 'country'

if key_name in person:

    print("country name is", person[key_name])

else:

    print("Key not found")
```

## Joining Two Dictionaries

### `.update()`

```python
dict1 = {'Jessa': 70, 'Arul': 80, 'Emma': 55}

dict2 = {'Kelly': 68, 'Harry': 50, 'Olivia': 66}



dict1.update(dict2)

print(dict1)
```

### `**` Unpacking — merge into a brand NEW dictionary

```python
student_dict1 = {'Aadya': 1, 'Arul': 2}

student_dict2 = {'Harry': 5, 'Olivia': 6}

student_dict3 = {'Nancy': 7, 'Perry': 9}



student_dict = {**student_dict1, **student_dict2, **student_dict3}

print(student_dict)
```

### Overlapping Keys — the LATER value wins

```python
dict1 = {'Jessa': 70, 'Arul': 80, 'Emma': 55}

dict2 = {'Kelly': 68, 'Harry': 50, 'Emma': 66}   # 'Emma' also appears here



dict1.update(dict2)

print(dict1['Emma'])   # dict2's value for 'Emma' overwrote dict1's
```

## Copying a Dictionary

`.copy()` and `dict()` both create a genuine, independent copy.

```python
dict1 = {'Jessa': 70, 'Emma': 55}



dict2 = dict1.copy()

print(dict2)



dict3 = dict(dict1)

print(dict3)



dict4 = dict(dict1.items())

print(dict4)
```

### Assignment (`=`) Is NOT a Copy

Same rule as lists and sets: `dict2 = dict1` just gives a second name to the SAME object. Modifying one affects the other, since there's really only one dictionary.

```python
dict1 = {'Jessa': 70, 'Emma': 55}



dict2 = dict1   # NOT a copy

dict2.update({'Jessa': 90})



print(dict2)

print(dict1)   # changed too - same underlying object
```

## Nested Dictionaries

A dictionary can hold another dictionary as one of its values — access the inner value with chained keys: `outer['key1']['key2']`.

```python
address = {"state": "Texas", 'city': 'Houston'}

person = {'name': 'Jessa', 'company': 'Google', 'address': address}



print("person:", person)

print("City:", person['address']['city'])



print("Person details")

for key, value in person.items():

    if key == 'address':

        print("Person Address")

        for nested_key, nested_value in value.items():

            print(" ", nested_key, ':', nested_value)

    else:

        print(key, ':', value)
```

### Multiple Nested Dictionaries in One Dictionary

```python
jessa = {'name': 'Jessa', 'state': 'Texas', 'city': 'Houston', 'marks': 75}

emma  = {'name': 'Emma',  'state': 'Texas', 'city': 'Dallas',  'marks': 60}

kelly = {'name': 'Kelly', 'state': 'Texas', 'city': 'Austin',  'marks': 85}



class_six = {'student1': jessa, 'student2': emma, 'student3': kelly}



print("Student 3 name:", class_six['student3']['name'])

print("Student 3 marks:", class_six['student3']['marks'])



print("\nClass details\n")

for key, value in class_six.items():

    print(key)

    for nested_key, nested_value in value.items():

        print(" ", nested_key, ':', nested_value)

    print()
```

## Sorting a Dictionary

`sorted()` on a dict sorts by KEY by default; `.items()` sorts the (key, value) pairs (by key, since tuples compare element-by-element); `.values()` sorts just the values.

```python
dict1 = {'c': 45, 'b': 95, 'a': 35}



print(sorted(dict1.items()))

print(sorted(dict1))

print(sorted(dict1.values()))
```

## Dictionary Comprehension

`{key: value for key, value in iterable if condition}`

```python
numbers = [1, 3, 5, 2, 8]

even_squares = {x: x ** 2 for x in numbers if x % 2 == 0}

print(even_squares)
```

Two SEPARATE iterables can be zipped together inside the comprehension.

```python
telephone_book = [1178, 4020, 5786]

persons = ['Jessa', 'Emma', 'Kelly']



telephone_directory = {key: value for key, value in zip(persons, telephone_book)}

print(telephone_directory)
```

## Built-in Functions With Dictionaries

### `max()` and `min()` — operate on the KEYS only, never the values

```python
sample = {1: 'aaa', 2: 'bbb', 3: 'AAA'}

print('Maximum Key', max(sample))

print('Minimum Key', min(sample))
```

### `all()` and `any()` — evaluated on the KEYS' truthiness (same rules as lists/sets)

**Note:** two identical keys in one dict LITERAL silently collapse to just one entry (the later value wins) — the two examples below deliberately use genuinely DISTINCT keys to correctly demonstrate 'all true' vs. 'one false'.

```python
dict1 = {1: 'a', 2: 'b'}       # both keys (1, 2) are truthy

dict2 = {0: 'a', 1: 'b'}       # key 0 is falsy

dict3 = {}                     # empty

dict4 = {'0': False}           # the STRING '0' is truthy (non-empty strings always are)



print('All True Keys::', all(dict1))

print('One False Key', all(dict2))

print('Empty Dictionary', all(dict3))

print('With "0" as a string key', all(dict4))
```

```python
dict1 = {1: 'a', 2: 'b'}

dict2 = {0: 'a', 1: 'b'}

dict3 = {}

dict5 = {0: False}             # the only key is falsy



print('All True Keys::', any(dict1))

print('One False Key ::', any(dict2))

print('Empty Dictionary ::', any(dict3))

print('all false ::', any(dict5))
```

## When to Use a Dictionary

Whenever data naturally maps ONE thing to ANOTHER — names to phone numbers, product IDs to prices, usernames to profiles. Lookups by key are O(1) on average (hash-table based), dramatically faster than searching a list item by item for large collections.

## Summary of Dictionary Operations

Assuming `d1 = {'a': 10, 'b': 20, 'c': 30}` and `d2 = {'d': 40, 'e': 50, 'f': 60}`:

| Operation | Description |
|---|---|
| `dict({'a': 10})` | create via constructor |
| `d1.get('a')` | safe value lookup |
| `d1.keys()` / `.values()` / `.items()` | views of keys / values / pairs |
| `len(d1)` | item count |
| `d1['d'] = 40` | add a new key |
| `d1.update({...})` | add/overwrite multiple keys |
| `d1.setdefault('g', 70)` | add only if missing |
| `d1.pop('b')` | remove and return by key |
| `d1.popitem()` | remove and return the last-inserted pair |
| `d1.clear()` | empty the dict |
| `'key' in d1` | key existence check |
| `{**d1, **d2}` | merge into a new dict |
| `d1.copy()` | genuine independent copy |
| `max(d1)` / `min(d1)` | largest/smallest KEY |

```python
d1 = {'a': 10, 'b': 20, 'c': 30}

d2 = {'d': 40, 'e': 50, 'f': 60}



print(d1.get('a'))

print(list(d1.keys()))

print(len(d1))

print({**d1, **d2})

print(max(d1), min(d1))
```

## Practice Exercises

1. Build a dictionary mapping 5 country names to their capital cities, then print each pair as `"CountryName -> Capital"`.
2. Given `scores = {'Alice': 85, 'Bob': 72, 'Carol': 91}`, find the name of the student with the HIGHEST score (not just the highest score itself — use `max()` with a `key=` function).
3. Write a function `merge_counts(d1, d2)` that merges two dictionaries mapping words to counts, ADDING the counts together when a word appears in both (rather than one overwriting the other).
4. Given a nested dictionary of employees with department info, write a loop that prints every employee grouped by department.
5. Using dictionary comprehension, build a dictionary mapping each word in a sentence to its length, for words longer than 3 letters only.

### Exercise 1 solution

```python
capitals = {

    "France": "Paris",

    "Japan": "Tokyo",

    "Egypt": "Cairo",

    "Brazil": "Brasilia",

    "Canada": "Ottawa",

}

for country, capital in capitals.items():

    print(f"{country} -> {capital}")
```

### Exercise 2 solution

```python
scores = {'Alice': 85, 'Bob': 72, 'Carol': 91}

top_student = max(scores, key=scores.get)

print(top_student, scores[top_student])
```

### Exercise 3 solution

```python
def merge_counts(d1, d2):

    result = d1.copy()

    for word, count in d2.items():

        result[word] = result.get(word, 0) + count

    return result



a = {"apple": 3, "banana": 2}

b = {"banana": 5, "cherry": 1}

print(merge_counts(a, b))
```

### Exercise 4 solution

```python
employees = {

    "Alice": {"department": "Engineering"},

    "Bob":   {"department": "Marketing"},

    "Carol": {"department": "Engineering"},

}



by_department = {}

for name, info in employees.items():

    dept = info["department"]

    by_department.setdefault(dept, []).append(name)



for dept, names in by_department.items():

    print(dept, ":", names)
```

### Exercise 5 solution

```python
sentence = "the quick brown fox jumps over the lazy dog"

word_lengths = {word: len(word) for word in sentence.split() if len(word) > 3}

print(word_lengths)
```
