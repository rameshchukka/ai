## `3_whileIfFactors.py` -- While If Factors

⌨️ **Needs keyboard input** -- this cell calls `input()`; Jupyter will prompt you inline when you run it.

```python
print('Please enter a number')
num = int(input())

# candidate will change in the while loop
# candidate is a potential factor
candidate = 1

while candidate <= num:
    # if num divided by candidate has no remainder
    if num % candidate == 0:
        print(candidate, 'is a factor of', num)
    candidate += 1
```

## `6_primeTest.py` -- Prime Test

⌨️ **Needs keyboard input** -- this cell calls `input()`; Jupyter will prompt you inline when you run it.

```python
import time

num = int(input('Provide a number to test:'))
tic = time.clock()
prime = True

for i in range(2, num):
    # if you can divide num by i and get no remainder
    # then num has a factor and is not a prime
    if num % i == 0:
        prime = False

toc = time.clock()
print('time taken:',toc-tic)

if prime == True:
    print(num, 'is a prime number')
else:
    print(num, 'is not a prime number')
```

## `7_nestedLoop.py` -- Nested Loop

```python
coordinates = []

for x in range(0, 5):
    for y in range(0, 5):
        c = str(x) + ',' + str(y)
        coordinates.append(c)

print(coordinates)
```
