## `4_stringInterpolaton.py` -- String Interpolaton

```python
person = 'Jeff'
pet_name = 'Rex'
pet_type = 'dog'

print(person, 'has a', pet_type, 'and his name is', pet_name, '.')

print(f'{person} has a {pet_type} and his name is {pet_name}.')
m = '{} has a {} and his name is {}.'
print(m.format(person, pet_type, pet_name))
print('%s has a %s and his name is %s.' % (person, pet_type, pet_name))
```

## `9_listFunctions.py` -- List Functions

```python
rainfall = [68, 58, 0, 99, 200, 39, 73, 43, 68, 0, 19]

min_rainfall = min(rainfall)
max_rainfall = max(rainfall)
average_rainfall = sum(rainfall) / len(rainfall)

print('minimum rainfal:', min_rainfall)
print('maximum rainfal:', max_rainfall)
print('average rainfall:', average_rainfall)
print(sorted(rainfall,reverse=True))
```

## `10bools.py` -- Bools

```python
raining = True
daytime = True
hot_outside = False

raining = not raining

if daytime and not raining and not hot_outside:
    print('Its a nice day lets go out')

if raining and not hot_outside or not raining and hot_outside:
    print('lets stay in')

if raining and hot_outside:
    print('lets dance in the rain')
```
