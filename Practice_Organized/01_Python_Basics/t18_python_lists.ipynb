# Python Lists

A **list** is Python's ordered, general-purpose, resizable collection type. Four defining properties:

- **Mutable** — items can be added, removed, or changed after creation.
- **Ordered** — items keep the position they were inserted in; new items go at the end by default.
- **Heterogeneous** — a single list can freely mix ints, strings, floats, even other lists.
- **Duplicates allowed** — the same value can appear more than once.

This notebook covers creating, accessing, iterating, modifying, and removing list items, copying, built-in functions, nested lists, and list comprehensions.

## Creating a List

Two ways: the `list()` constructor, or square-bracket literal syntax `[]`.

```python
my_list1 = list((1, 2, 3))

print(my_list1)



my_list2 = [1, 2, 3]

print(my_list2)



my_list3 = [1.0, 'Jessa', 3]   # heterogeneous - mixed types are fine

print(my_list3)



my_list4 = list()

print(my_list4)



my_list5 = []

print(my_list5)
```

## Length of a List

```python
my_list = [1, 2, 3]

print(len(my_list))
```

## Accessing Items: Indexing

Each item has a unique POSITION, accessible via `[index]`. Positive indices count from the left starting at 0; negative indices count from the right starting at -1.

list positive and negative indexing

```python
my_list = [10, 20, 'Jessa', 12.50, 'Emma']

print(my_list[1])   # 2nd element

print(my_list[4])   # 5th element
```

An out-of-range index raises `IndexError`; a non-integer index raises `TypeError`.

```python
my_list = [10, 20, 'Jessa']

try:

    print(my_list[10])

except IndexError as e:

    print("IndexError:", e)



try:

    print(my_list[1.5])

except TypeError as e:

    print("TypeError:", e)
```

## Negative Indexing

```python
my_list = [10, 20, 'Jessa', 12.50, 'Emma']

print(my_list[-1])   # last element

print(my_list[-2])   # second-to-last

print(my_list[-4])   # 4th from the end
```

## List Slicing

`list[start:stop:step]` extracts a RANGE of items — `stop` is always excluded, just like `range()`.

```python
my_list = [10, 20, 'Jessa', 12.50, 'Emma', 25, 50]

print(my_list[2:5])
```

```python
my_list = [5, 8, 'Tom', 7.50, 'Emma']



print(my_list[:4])     # first four items

print(my_list[::2])    # every second item

print(my_list[::-1])   # reversed

print(my_list[3:])     # from index 3 to the end
```

## Iterating a List

Directly with `for`, or paired with `range(len(...))` when you specifically need the index too.

```python
my_list = [5, 8, 'Tom', 7.50, 'Emma']

for item in my_list:

    print(item)
```

```python
my_list = [5, 8, 'Tom', 7.50, 'Emma']

for i in range(0, len(my_list)):

    print(my_list[i])
```

## Adding Elements

### `append(item)` — add ONE item at the end

```python
my_list = list([5, 8, 'Tom', 7.50])



my_list.append('Emma')

print(my_list)



my_list.append([25, 50, 75])   # appends the WHOLE list as a single nested element

print(my_list)
```

### `insert(index, item)` — add ONE item at a specific position

```python
my_list = list([5, 8, 'Tom', 7.50])



my_list.insert(2, 25)

print(my_list)



my_list.insert(3, [25, 50, 75])

print(my_list)
```

### `extend(iterable)` — add MULTIPLE items, unpacked, at the end

```python
my_list = list([5, 8, 'Tom', 7.50])

my_list.extend([25, 75, 100])

print(my_list)
```

## Modifying Items

Assign a new value directly to an index (or a slice, to replace a whole RANGE at once).

```python
my_list = list([2, 4, 6, 8, 10, 12])



my_list[0] = 20

print(my_list)



my_list[1:4] = [40, 60, 80]

print(my_list)



my_list[3:] = [80, 100, 120]

print(my_list)
```

### Modifying Every Item

```python
my_list = list([2, 4, 6, 8])

for i in range(len(my_list)):

    square = my_list[i] * my_list[i]

    my_list[i] = square

print(my_list)
```

## Removing Elements

| Method | Description |
|---|---|
| `remove(item)` | Removes the FIRST occurrence of a value |
| `pop(index)` | Removes AND returns the item at a position (last item if no index given) |
| `clear()` | Empties the list entirely |
| `del list_name` | Deletes the list variable itself |

### `remove()` — by value, first match only

```python
my_list = list([2, 4, 6, 8, 10, 12])

my_list.remove(6)

my_list.remove(8)

print(my_list)
```

`remove()` on a value not present raises `ValueError` (not `KeyError`, despite what you might expect from dicts).

```python
my_list = [1, 2, 3]

try:

    my_list.remove(99)

except ValueError as e:

    print("ValueError:", e)
```

### Removing EVERY occurrence of a value

```python
my_list = list([6, 4, 6, 6, 8, 12])

while 6 in my_list:

    my_list.remove(6)

print(my_list)
```

### `pop()` — by index, returns what was removed

```python
my_list = list([2, 4, 6, 8, 10, 12])

my_list.pop(2)

print(my_list)



my_list.pop()   # no index -> removes the LAST item

print(my_list)
```

### Removing a RANGE with `del` + slicing

```python
my_list = list([2, 4, 6, 8, 10, 12])

del my_list[2:5]

print(my_list)



my_list = list([2, 4, 6, 8, 10, 12])

del my_list[3:]

print(my_list)
```

### Removing everything

```python
my_list = list([2, 4, 6, 8, 10, 12])

my_list.clear()

print(my_list)



del my_list

try:

    print(my_list)

except NameError as e:

    print("NameError:", e)
```

## Finding an Element: `index()`

Returns the position of the FIRST match, or raises `ValueError` if it isn't found.

```python
my_list = list([2, 4, 6, 8, 10, 12])

print(my_list.index(8))



try:

    my_list.index(100)

except ValueError as e:

    print("ValueError:", e)
```

## Concatenating Two Lists

`+` builds a brand NEW combined list; `.extend()` modifies the FIRST list in place.

```python
my_list1 = [1, 2, 3]

my_list2 = [4, 5, 6]



my_list3 = my_list1 + my_list2

print(my_list3)



my_list1.extend(my_list2)

print(my_list1)
```

## Copying a List

### Assignment (`=`) — this is NOT a copy at all

**Important correction to a common misconception:** writing `new_list = my_list1` does NOT create any kind of copy — not a deep copy, not a shallow copy. It just gives a SECOND NAME to the exact same list object already in memory. Both names point at the identical object, so a change made through either name is visible through the other, since there's really only ever one list involved.

```python
my_list1 = [1, 2, 3]



new_list = my_list1   # NOT a copy - new_list and my_list1 are now two names for ONE object

print(new_list)



my_list1.append(4)



print(my_list1)

print(new_list)         # changed too! because it's literally the same object

print(new_list is my_list1)  # True confirms they're the same object, not separate copies
```

### `.copy()` — a genuine (shallow) copy

`.copy()` allocates a real, independent list object with the same starting contents. Changes to the original afterward do NOT appear in the copy, because they're now two separate objects. This is called a **shallow** copy — it copies the list's own structure, but if the list contains NESTED mutable objects (like an inner list), those inner objects are still shared between the original and the copy (this nuance is covered in more depth in the Dictionaries and Copying material later in this series).

```python
my_list1 = [1, 2, 3]



new_list = my_list1.copy()   # genuine copy - a separate object

print(new_list)



my_list1.append(4)



print(my_list1)

print(new_list)              # unaffected - it's a separate list

print(new_list is my_list1)  # False confirms they are different objects
```

## List Operations: `sort()` and `reverse()`

Both modify the list IN PLACE and return `None` — don't write `mylist = mylist.sort()`, that would overwrite your list with `None`.

```python
mylist = [3, 2, 1]

mylist.sort()

print(mylist)
```

```python
mylist = [3, 4, 5, 6, 1]

mylist.reverse()

print(mylist)
```

## Built-in Functions That Work on Lists

### `max()` and `min()`

```python
mylist = [3, 4, 5, 6, 1]

print(max(mylist))

print(min(mylist))
```

### `sum()`

```python
mylist = [3, 4, 5, 6, 1]

print(sum(mylist))
```

### `all()` — True only if EVERY item is truthy

| Contents | Result |
|---|---|
| All truthy | `True` |
| One or more falsy | `False` |
| All falsy | `False` |
| Empty list | `True` (vacuously — there's nothing to be false) |

```python
print("all() All True values::", all([1, 1, True]))

print("all() with One false value ::", all([0, 1, True, 1]))

print("all() with all false values ::", all([0, 0, False]))

print("all() Empty list ::", all([]))
```

### `any()` — True if AT LEAST ONE item is truthy

| Contents | Result |
|---|---|
| All truthy | `True` |
| One or more truthy | `True` |
| All falsy | `False` |
| Empty list | `False` |

```python
print("any() True values::", any([1, 1, True]))

print("any() One true value ::", any([0, 1, True, 1]))

print("any() all false values ::", any([0, 0, False]))

print("any() Empty list ::", any([]))
```

## Nested Lists

A list can contain another list as one of its elements. Accessing an element inside a NESTED list needs two index lookups chained together — `outer[i][j]`.

```python
mylist = [3, 4, 5, 6, 3, [1, 2, 3], 4]

print(mylist)

print(mylist[5])       # the inner list itself

print(mylist[5][1])    # the 2nd element WITHIN the inner list
```

```python
nestedlist = [[2, 4, 6, 8, 10], [1, 3, 5, 7, 9]]



print("Accessing the third element of the second list", nestedlist[1][2])



for i in nestedlist:

    print("list", i, "elements")

    for j in i:

        print(j)
```

## List Comprehension

A compact way to build a new list from an existing iterable, with an optional filter — see the full Comprehensions notebook elsewhere in this series for the deep dive.

`[expression for variable in inputList if condition]`

```python
inputList = [4, 7, 11, 13, 18, 20]

squareList = [var**2 for var in inputList if var % 2 == 0]

print(squareList)
```

```python
squarelist1 = [s**2 for s in range(10) if s % 2 == 0]

print(squarelist1)
```

## Summary of List Operations

Assuming `l1 = [10, 20, 30, 40, 50]` and `l2 = [60, 70, 80, 60]`:

| Operation | Description |
|---|---|
| `x in l1` | membership test |
| `x not in l1` | negative membership test |
| `l1 + l2` | concatenation (new list) |
| `l1 * 5` | repeat the list 5 times |
| `l1[i]` | index access |
| `l1[i:j]` | slice |
| `l1[i:j:k]` | slice with step |
| `len(l1)` | item count |
| `l2.count(60)` | count occurrences (2, for `l2`) |
| `l1.index(30)` | first position of a value (2) |
| `min(l1)` / `max(l1)` | smallest/largest value |
| `l1.append(x)` | add one item at the end |
| `l1[i] = x` | modify by index |
| `l1.remove(x)` | remove first occurrence of a value |
| `l1.pop(i)` | remove and return item at index i |
| `l1.clear()` | empty the list |
| `l1.copy()` | genuine (shallow) copy |

```python
l1 = [10, 20, 30, 40, 50]

l2 = [60, 70, 80, 60]



print(30 in l1)

print(l1 + l2)

print(l1 * 2)

print(l1[2])

print(l1[0:2])

print(l1[0:4:2])

print(len(l1))

print(l2.count(60))

print(l1.index(30))

print(min(l1), max(l1))
```

### Exercise 1 solution

```python
nums = [5, 3, 8, 1, 9, 2]

nums.sort(reverse=True)

print(nums)
```

### Exercise 2 solution

```python
def remove_duplicates_preserve_order(lst):

    seen = set()

    result = []

    for item in lst:

        if item not in seen:

            seen.add(item)

            result.append(item)

    return result



print(remove_duplicates_preserve_order([3, 1, 4, 1, 5, 9, 2, 6, 5, 3]))
```

### Exercise 3 solution

```python
grid = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

diagonal_sum = sum(grid[i][i] for i in range(len(grid)))

print(diagonal_sum)
```

### Exercise 4 solution

```python
original = [1, 2, [3, 4]]

shallow = original.copy()



original[2].append(5)   # mutating the SHARED inner list



print("original:", original)

print("shallow :", shallow)

print("Inner list is shared -> the change shows up in BOTH, even though .copy() was used.")
```
