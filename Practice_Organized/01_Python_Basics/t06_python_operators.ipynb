# Python Operators

An **operator** is a symbol that performs an operation on one or more **operands** (the values it acts on) and produces a result — `x + y` is an expression where `+` is the operator and `x`, `y` are the operands.

Python groups its operators into seven families:
1. Arithmetic
2. Relational (comparison)
3. Assignment
4. Logical
5. Membership
6. Identity
7. Bitwise

This notebook walks through all seven, then closes with the operator precedence table that determines evaluation order in a compound expression.

## 1. Arithmetic Operators

`+ - * / // % **` — perform the familiar mathematical operations.

```python
x = 10

y = 40

print(x + y)  # addition



# + also works on strings, as concatenation

name = "Kelly"

surname = "Ault"

print(surname + " " + name)
```

```python
x = 10

y = 40

print(y - x)  # subtraction
```

```python
x, y, z = 2, 4, 5

print(x * y)      # multiplication

print(x * y * z)



# * also works on strings/lists, as repetition

name = "Jessa"

print(name * 3)
```

```python
x, y, z = 2, 4, 8

print(y / x)       # division - ALWAYS returns a float, even for exact division

print(z / y / x)



try:

    print(z / 0)

except ZeroDivisionError as e:

    print("ZeroDivisionError:", e)
```

```python
x, y, z = 2, 4, 2.2



print(y / x)     # normal division

print(y // x)    # floor division: drops everything after the decimal point



print(y / z)     # normal division

print(y // z)    # floor division - result is FLOAT here, since one operand (z) is a float
```

```python
x, y = 15, 4

print(x % y)   # modulus: the REMAINDER of x / y
```

```python
num = 2

print(num ** 2)   # 2 * 2

print(num ** 3)   # 2 * 2 * 2
```

## 2. Relational (Comparison) Operators

`> < == != >= <=` compare two values and always produce a `bool`. Python also supports **chained comparisons** — `x > y > z` reads naturally as 'x > y AND y > z', evaluated efficiently without computing `y` twice.

```python
x, y, z = 10, 5, 2



print(x > y)        # Greater than

print(x > y > z)     # chained: x > y AND y > z



print(x < y)         # Less than

print(y < x)



print(x == y)        # Equal to

print(x == 10)



print(x != y)        # Not equal to

print(10 != x)



print(x >= y)        # Greater than or equal to

print(10 >= x)



print(x <= y)        # Less than or equal to

print(10 <= x)
```

## 3. Assignment Operators

`=` assigns a value. The compound forms (`+=`, `-=`, `*=`, ...) are shorthand that combine an arithmetic operation with reassignment in one step — `a += 5` means exactly `a = a + 5`, just more concise.

```python
a, b = 4, 2

a += b

print(a)   # 4 + 2
```

```python
a = 4

a -= 2

print(a)
```

```python
a = 4

a *= 2

print(a)
```

```python
a = 4

a /= 2

print(a)   # note: /= still gives a float result, same as /
```

```python
a = 4

a **= 2

print(a)
```

```python
a = 5

a %= 2

print(a)
```

```python
a = 4

a //= 2

print(a)
```

## 4. Logical Operators

`and`, `or`, `not` combine or invert boolean expressions — the building blocks of every conditional you'll write.

```python
print(True and False)

print(True and True)

print(False and False)

print(False and True)



# a realistic use

a, b = 2, 4

if a > 0 and b > 0:

    print(a * b)

else:

    print("Do nothing")
```

### `and`/`or` on Non-Boolean Values

Python's `and`/`or` don't always return `True`/`False` when given non-boolean operands — they return one of the ORIGINAL operands, following short-circuit rules:
- `and` returns the FIRST operand if it's falsy, otherwise the SECOND operand.
- `or` returns the FIRST operand if it's truthy, otherwise the SECOND operand.

For two nonzero numbers, this means `and` always yields the second value, and `or` always yields the first.

```python
print(10 and 20)    # both truthy -> and returns the SECOND

print(10 and 5)

print(100 and 300)

print()

print(10 or 20)      # first is truthy -> or returns the FIRST

print(10 or 5)

print(100 or 300)
```

```python
a, b = 2, 4

if a > 0 or b < 0:

    print(a + b)

else:

    print("Do nothing")
```

```python
print(not False)

print(not True)



a = True

if not a:

    print(a)

else:

    print("Do nothing")
```

`not` on a non-boolean value evaluates the value's **truthiness** first, then inverts it. `0` is the only falsy integer, so `not 0` is `True` and `not` of any other integer is `False`.

```python
print(not 10)   # False - 10 is truthy

print(not 1)    # False - 1 is truthy

print(not 5)    # False - 5 is truthy

print(not 0)    # True  - 0 is the only falsy int
```

## 5. Membership Operators

`in` / `not in` check whether a value exists inside a sequence (string, list, tuple, set, dict keys, ...).

```python
my_list = [11, 15, 21, 29, 50, 70]

number = 15

if number in my_list:

    print("number is present")

else:

    print("number is not present")
```

```python
my_tuple = (11, 15, 21, 29, 50, 70)

number = 35

if number not in my_tuple:

    print("number is not present")

else:

    print("number is present")
```

## 6. Identity Operators

`is` / `is not` compare **object identity** (are both names pointing at the exact SAME object in memory) — not equal VALUE. This is subtly different from `==`, which compares values. Two variables can hold equal values while being different objects (common for larger numbers, lists, etc.).

```python
x = 10

y = 11

z = 10

print(x is y)      # different values -> different objects

print(x is z)      # same small int - CPython often reuses (caches) the same object
```

```python
x = 10

y = 11

z = 10

print(x is not y)

print(x is not z)
```

```python
# == (equal value) vs is (same object) - a classic point of confusion

a = 1000

b = 1000

print("a == b:", a == b)   # True - same VALUE, always

print("a is b:", a is b)   # implementation-dependent! Might be True OR False depending on

                            # how/where the two literals were compiled and whether CPython's

                            # constant-folding happened to reuse one object for both.
```

## 7. Bitwise Operators

Bitwise operators work on the individual binary digits (bits) of an integer, and always return the result back in ordinary decimal form. Python has 6: `& | ^ ~ << >>`.

Bitwise AND bit-by-bit diagram

```python
a, b, c = 7, 4, 5

# binary: a=0111, b=0100, c=0101

print(a & b)   # 0111 & 0100 = 0100 = 4

print(a & c)   # 0111 & 0101 = 0101 = 5

print(b & c)   # 0100 & 0101 = 0100 = 4
```

```python
a, b, c = 7, 4, 5

print(a | b)   # 0111 | 0100 = 0111 = 7

print(a | c)   # 0111 | 0101 = 0111 = 7

print(b | c)   # 0100 | 0101 = 0101 = 5
```

```python
a, b, c = 7, 4, 5

# XOR: 1 only where the bits DIFFER between the two operands

print(a ^ c)   # 0111 ^ 0101 = 0010 = 2

print(b ^ c)   # 0100 ^ 0101 = 0001 = 1
```

```python
a, b, c = 7, 4, 3

# ~x is equivalent to -(x + 1) - it flips every bit, including the sign

print(~a, ~b, ~c)
```

```python
# left-shift: shifts bits left, filling with 0s on the right - equivalent to multiplying by 2**n

print(4 << 2)   # 4 * (2**2) = 16

print(5 << 3)   # 5 * (2**3) = 40
```

```python
# right-shift: shifts bits right, discarding bits that fall off the end - equivalent to // (2**n)

print(4 >> 2)   # 4 // (2**2) = 1

print(5 >> 2)   # 5 // (2**2) = 1
```

## Operator Precedence

When an expression mixes several operators, **precedence** determines which runs first — exactly like the order of operations (PEMDAS) in ordinary math. Parentheses always win, and can be added anywhere to force a specific order or simply to make intent clearer to a reader.

```python
result = (10 - 4) * 2 + (10 + 2)

print(result)

# Step by step: (10-4)=6, (10+2)=12, then 6*2=12, then 12+12=24
```

Highest to lowest precedence:

| Level | Operator | Meaning |
|---|---|---|
| 1 (highest) | `()` | Parentheses |
| 2 | `**` | Exponent |
| 3 | `+x`, `-x`, `~x` | Unary plus/minus, bitwise NOT |
| 4 | `* / // %` | Multiplication, division, floor division, modulus |
| 5 | `+ -` | Addition, subtraction |
| 6 | `<< >>` | Bitwise shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `\|` | Bitwise OR |
| 10 | `== != > >= < <=` | Comparison |
| 11 | `is`, `is not`, `in`, `not in` | Identity, membership |
| 12 | `not` | Logical NOT |
| 13 | `and` | Logical AND |
| 14 (lowest) | `or` | Logical OR |

## Summary

- **Arithmetic**: `+ - * / // % **` — remember `/` always returns a float, and `//` floors the result.
- **Relational**: `> < == != >= <=` — always return `bool`; chainable (`x > y > z`).
- **Assignment**: `=` plus shorthand compound forms (`+=`, `-=`, ...).
- **Logical**: `and`/`or` return one of their actual operands (not always a plain bool) when given non-boolean values; `not` inverts truthiness.
- **Membership**: `in`/`not in` test sequence containment.
- **Identity**: `is`/`is not` compare object identity, not value — different from `==`.
- **Bitwise**: `& | ^ ~ << >>` operate on binary representations.
- Precedence determines evaluation order in mixed expressions — when in doubt, add parentheses.

### Exercise 1 solution

```python
a, b = 17, 5

print("a + b =", a + b)

print("a - b =", a - b)

print("a * b =", a * b)

print("a / b =", a / b)

print("a // b =", a // b)

print("a % b =", a % b)

print("a ** b =", a ** b)
```

### Exercise 4 solution

```python
mult = 6 << 2   # 6 * (2**2) = 6 * 4

div = 20 >> 2   # 20 // (2**2) = 20 / 4

print(mult, mult == 6 * 4)

print(div, div == 20 // 4)
```

```python
list_a = [1, 2, 3]

list_b = [1, 2, 3]



print("Same object (is):", list_a is list_b)   # False - two separate list objects

print("Equal value (==):", list_a == list_b)     # True - same contents
```
